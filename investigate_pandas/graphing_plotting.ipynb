```python
# Setup imports & load dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import time
import os
df = pd.read_csv('data/raw_transactions.csv', na_values=['Nan', ''])
```

## Section 11: Graphing & Outliers

### Built-in Chart Plotting
**Explanation**: Fast wrapper functions built on Matplotlib. Useful for checking data distributions (histograms) and identifying outliers (boxplots).

**Syntax**: `df.plot.scatter(x, y)` / `df.boxplot(column)`

```python
df['transaction_amount'].plot.hist(bins=15)
plt.title("Transaction Amount Histogram")
plt.show()

df.boxplot(column='transaction_amount')
plt.title("Transaction Amount Boxplot")
plt.show()
```
