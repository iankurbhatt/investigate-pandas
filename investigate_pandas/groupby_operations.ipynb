```python
# Setup imports & load dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import time
import os
df = pd.read_csv('data/raw_transactions.csv', na_values=['Nan', ''])
```

## Section 9: Column Operations & Feature Engineering

### Method Chaining Assign
**Explanation**: Declares new columns dynamically without modifying the original DataFrame. Crucial for designing clean, fluent method chains.

**Syntax**: `df.assign(new_col=lambda x: x.col * factor)`

```python
df_features = df.assign(tax_amount=lambda x: x.transaction_amount * 0.1)
print(df_features[['transaction_amount', 'tax_amount']].head(3))
```

## Section 10: Groupby

### Groupby Aggregations
**Explanation**: Implements the Split-Apply-Combine pattern. Groups rows based on category values and computes custom aggregates (e.g. sum, mean) per group.

**Syntax**: `df.groupby('col').agg(new_name=('target_col', 'func'))`

```python
print(df.groupby('region')['transaction_amount'].agg(['count', 'sum', 'mean']))
```
