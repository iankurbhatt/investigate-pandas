Exhaustive revision guide to Pandas Matplotlib wrappers: line, bar, scatter, hist, boxplot, and subplots on raw_transactions.csv.

### Key 5-Year Interview Concepts Covered:
- **Line & Bar Charts**: Dedicated cell for `df.plot.line()` and `df.plot.bar()`.
- **Distributions & Correlation**: Dedicated cell for `df.plot.scatter()`, `df.plot.hist()`, and `df.boxplot()`.
- **Multi-Panel Views**: Dedicated cell for subplots with `subplots=True`.

This interactive revision guide uses `data/raw_transactions.csv` for all real-world code examples.

```python
# Setup imports & dataset loading
import pandas as pd
import numpy as np
import sys
import time
import os
import matplotlib.pyplot as plt

# Load raw transactions dataset
csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
df = pd.read_csv(csv_path)
print(f"Pandas Version: {pd.__version__}")
print(f"Loaded raw_transactions.csv: {df.shape[0]} rows, {df.shape[1]} columns")
```

### Line Charts with `df.plot.line()`
**Explanation**: Plots cumulative daily transaction revenue.

**Syntax**: `daily_ts.plot.line(title='Daily Revenue')`

```python
daily_rev = df.assign(date=pd.to_datetime(df['transaction_date'], format='mixed', errors='coerce')).dropna().set_index('date').resample('D')['transaction_amount'].sum()
ax = daily_rev.head(30).plot.line(title='30-Day Transaction Revenue Trend')
plt.close()
```

### Bar Charts with `df.plot.bar()`
**Explanation**: Plots total transaction spend by region.

**Syntax**: `df.groupby('region')['transaction_amount'].sum().plot.bar()`

```python
ax = df.groupby('region')['transaction_amount'].sum().plot.bar(title='Regional Spend', rot=15)
plt.close()
```

### Scatter Plots with `df.plot.scatter()`
**Explanation**: Plots transaction amount versus account age.

**Syntax**: `df.plot.scatter(x='account_age_months', y='transaction_amount')`

```python
ax = df.head(500).plot.scatter(x='account_age_months', y='transaction_amount', alpha=0.5, title='Account Age vs Spend')
plt.close()
```

### Histograms with `df.plot.hist()`
**Explanation**: Plots distribution of transaction amounts.

**Syntax**: `df['transaction_amount'].plot.hist(bins=30)`

```python
ax = df['transaction_amount'].plot.hist(bins=30, title='Transaction Amount Distribution')
plt.close()
```

### Boxplots with `df.boxplot()`
**Explanation**: Draws boxplots comparing transaction amount distributions across card types.

**Syntax**: `df.boxplot(column='transaction_amount', by='card_type')`

```python
ax = df.boxplot(column='transaction_amount', by='card_type')
plt.close()
```

### Multi-Panel Subplots: `subplots=True`
**Explanation**: Generates multi-panel subplots comparing regional card metrics.

**Syntax**: `regional_metrics.plot(subplots=True)`

```python
reg_metrics = df.groupby('region')[['transaction_amount', 'account_age_months']].mean()
axes = reg_metrics.plot(subplots=True, figsize=(8, 4))
plt.close()
```

### Q1: Formatting Financial Visualization Figures
**Explanation**: Customize the Matplotlib figure with grid lines, dollar labels, and bold titles.

**Syntax**: `ax.yaxis.set_major_formatter('${x:,.0f}')`

```python
fig, ax = plt.subplots(figsize=(6, 3))
df.groupby('card_type')['transaction_amount'].mean().plot.bar(ax=ax, grid=True, title='Average Spend by Card')
ax.set_ylabel('Mean Amount ($)')
plt.close()
print('Plotting API execution verified!')
```
