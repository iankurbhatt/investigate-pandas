# 13. Plotting & Exploratory Visualization API: Beginner Guide

### 📌 Overview & Architectural Context
Welcome to **13. Plotting & Exploratory Visualization API**. Visual inspection is an indispensable phase of exploratory data analysis (EDA). This notebook demonstrates how to generate publication-grade visual figures directly from Pandas DataFrames, covering trend line charts, categorical bar distributions, frequency histograms, continuous KDE density estimates, boxplots, pair-wise scatter matrices, and dual secondary y-axes.

### 📚 Key Concepts Covered in this Notebook:
- [x] 🔹 Line Plotting: `DataFrame.plot.line()`
- [x] 🔹 Categorical Bar Plotting: `DataFrame.plot.bar()`
- [x] 🔹 Scatter Plotting: `DataFrame.plot.scatter()`
- [x] 🔹 Distribution Histogram: `DataFrame.plot.hist()`
- [x] 🔹 Five-Number Summary Boxplots: `DataFrame.boxplot()`
- [x] 🔹 Multi-Panel Subplot Layouts: `DataFrame.plot(subplots=True)`


```python
# Setup imports & dataset loading
import pandas as pd
import numpy as np
import sys
import time
import os
import matplotlib.pyplot as plt

# Load raw transactions dataset
csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
df = pd.read_csv(csv_path)
print(f"Pandas Version: {pd.__version__}")
print(f"Loaded raw_transactions.csv: {df.shape[0]} rows, {df.shape[1]} columns")
```

### 🔹 Line Plotting: `DataFrame.plot.line()`
- **What it does:** Generates a 2D line plot visualizing continuous sequences, trends, and time series across index coordinates.
- **Syntax:** `DataFrame.plot.line()`
  - **Parameters:**
    - `row_label` (*hashable*): Row label.
    - `col_label` (*hashable*): Column label.
- **Key Note:** Pandas plotting API wraps Matplotlib directly, allowing seamless integration with custom `plt` styling commands.
- **Dataset Application & Code Demonstration:** Applies Line Plotting on fintech records using columns `transaction_amount`, `transaction_date` to demonstrate real-world execution.


```python
daily_rev = df.assign(date=pd.to_datetime(df['transaction_date'], format='mixed', errors='coerce')).dropna().set_index('date').resample('D')['transaction_amount'].sum()
ax = daily_rev.head(30).plot.line(title='30-Day Transaction Revenue Trend')
plt.close()
```

### 🔹 Categorical Bar Plotting: `DataFrame.plot.bar()`
- **What it does:** Generates vertical bar plots comparing discrete categories and aggregate metrics.
- **Syntax:** `DataFrame.plot.bar()`
  - **Parameters:**
    - `row_label` (*hashable*): Row label.
    - `col_label` (*hashable*): Column label.
- **Key Note:** Use `stacked=True` to visualize compositional category breakdowns within each primary group.
- **Dataset Application & Code Demonstration:** Applies Categorical Bar Plotting on fintech records using columns `region`, `transaction_amount` to demonstrate real-world execution.


```python
ax = df.groupby('region')['transaction_amount'].sum().plot.bar(title='Regional Spend', rot=15)
plt.close()
```

### 🔹 Scatter Plotting: `DataFrame.plot.scatter()`
- **What it does:** Generates a 2D scatter plot exploring correlation, clustering, and relationships between two continuous variables.
- **Syntax:** `DataFrame.plot.scatter()`
  - **Parameters:**
    - `row_label` (*hashable*): Row label.
    - `col_label` (*hashable*): Column label.
- **Key Note:** Specifying `c='numeric_col'` automatically attaches a continuous colorbar scale to the plot.
- **Dataset Application & Code Demonstration:** Applies Scatter Plotting on fintech records using columns `account_age_months`, `transaction_amount` to demonstrate real-world execution.


```python
ax = df.head(500).plot.scatter(x='account_age_months', y='transaction_amount', alpha=0.5, title='Account Age vs Spend')
plt.close()
```

### 🔹 Distribution Histogram: `DataFrame.plot.hist()`
- **What it does:** Generates a histogram estimating the underlying probability distribution and frequency of numerical data.
- **Syntax:** `DataFrame.plot.hist()`
  - **Parameters:**
    - `row_label` (*hashable*): Row label.
    - `col_label` (*hashable*): Column label.
- **Key Note:** Experiment with `bins=30` or `bins=50` to uncover multi-modal distributions masked by coarse binning.
- **Dataset Application & Code Demonstration:** Applies Distribution Histogram on fintech records using columns `transaction_amount` to demonstrate real-world execution.


```python
ax = df['transaction_amount'].plot.hist(bins=30, title='Transaction Amount Distribution')
plt.close()
```

### 🔹 Five-Number Summary Boxplots: `DataFrame.boxplot()`
- **What it does:** Generates box-and-whisker plots displaying minimum, first quartile (Q1), median, third quartile (Q3), maximum, and outlier points.
- **Syntax:** `DataFrame.boxplot()`
  - **Parameters:**
    - `row_label` (*hashable*): Row label.
    - `col_label` (*hashable*): Column label.
- **Key Note:** Points plotted beyond the whiskers (`1.5 * IQR`) represent statistical anomalies and outlier observations.
- **Dataset Application & Code Demonstration:** Applies Five-Number Summary Boxplots on fintech records using columns `card_type`, `transaction_amount` to demonstrate real-world execution.


```python
ax = df.boxplot(column='transaction_amount', by='card_type')
plt.close()
```

### 🔹 Multi-Panel Subplot Layouts: `DataFrame.plot(subplots=True)`
- **What it does:** Renders each numeric column onto its own dedicated, isolated sub-axes panel in a shared figure grid.
- **Syntax:** `DataFrame.plot(subplots=True)`
  - **Parameters:**
    - `row_label` (*hashable*): Row label.
    - `col_label` (*hashable*): Column label.
- **Key Note:** Setting `sharey=False` allows each variable to scale to its own dynamic range without being flattened by high-magnitude peers.
- **Dataset Application & Code Demonstration:** Applies Multi-Panel Subplot Layouts on fintech records using columns `account_age_months`, `region`, `transaction_amount` to demonstrate real-world execution.


```python
reg_metrics = df.groupby('region')[['transaction_amount', 'account_age_months']].mean()
axes = reg_metrics.plot(subplots=True, figsize=(8, 4))
plt.close()
```

### 🔍 Scenario: Q1: Formatting Financial Visualization Figures
- **Objective:** Q1: Formatting Financial Visualization Figures
- **Approach:** Customize the Matplotlib figure with grid lines, dollar labels, and bold titles.
- **Syntax:** `ax.yaxis.set_major_formatter('${x:,.0f}')`

```python
fig, ax = plt.subplots(figsize=(6, 3))
df.groupby('card_type')['transaction_amount'].mean().plot.bar(ax=ax, grid=True, title='Average Spend by Card')
ax.set_ylabel('Mean Amount ($)')
plt.close()
print('Plotting API execution verified!')
```
