## Section 2: Data Structures & Series Alignment

### Series Creation
**Explanation**: A Series is a 1-dimensional labeled array capable of holding any data type (integers, strings, floating point numbers, Python objects). Axis labels are collectively referred to as the index.

**Syntax**: `pd.Series(data, index)`

```python
# Series Creation (with custom indices)
s = pd.Series(data=['DSA', 'ML', 'DS'], index=[10, 20, 30])
print(s)
```

### Series Index Alignment
**Explanation**: When performing operations (like addition) on two Series, Pandas automatically aligns elements based on their index labels. If an index label exists in one Series but not the other, the resulting value will be NaN. To avoid this, use methods like `.add()` with a `fill_value`.

**Syntax**: `s1 + s2` or `s1.add(s2, fill_value=0)`

```python
s1 = pd.Series([100, 200], index=['TX1', 'TX2'])
s2 = pd.Series([300, 400], index=['TX2', 'TX3'])
print(s1 + s2)
```

## Section 3: Read and Write Ingestion

### Read CSV
**Explanation**: Ingests tabular data from a CSV file into a DataFrame. Using `na_values` converts custom string representations (like 'Nan' or blank spaces) to actual Pandas null objects (`NaN`), ensuring consistent numeric processing.

**Syntax**: `pd.read_csv(filepath, na_values=['Nan', ''])`

```python
# Load transactions dataset with custom missing value list
df = pd.read_csv('data/raw_transactions.csv', na_values=['Nan', ''])
print("Shape:", df.shape)
df.head(3)
```

### Column Data Types
**Explanation**: Returns the data type of each column. Checking dtypes is crucial for memory tuning (e.g., ensuring integer columns aren't parsed as float64 due to missing values).

**Syntax**: `df.dtypes`

```python
print(df.dtypes)
```

### DataFrame Shape
**Explanation**: Returns a tuple representing the dimensionality of the DataFrame: (number of rows, number of columns).

**Syntax**: `df.shape`

```python
print("Rows, Cols:", df.shape)
```

### DataFrame Info
**Explanation**: Provides a concise summary of the DataFrame, including the index range, columns, non-null counts, data types, and memory usage. Excellent for a quick sanity check after data cleaning.

**Syntax**: `df.info()`

```python
df.info()
```

### Raw Array Access
**Explanation**: Extracts the underlying data as a raw NumPy n-dimensional array, bypassing the labels and index structures. Essential when interfacing with scikit-learn or custom numerical optimization packages.

**Syntax**: `df.values`

```python
print(df.head(2).values)
```

### Save to CSV & Excel
**Explanation**: Exports the DataFrame to CSV and Excel formats. Storing subsets or summaries is common in data pipelines. Use `index=False` to prevent pandas from exporting the index as an unnamed column.

**Syntax**: `df.to_csv('file.csv', index=False)` / `df.to_excel('file.xlsx', index=False)`

```python
df_sub = df.head(5)
df_sub.to_csv('tx_sub.csv', index=False)
df_sub.to_excel('tx_sub.xlsx', index=False)
try:
    os.remove('tx_sub.csv')
    os.remove('tx_sub.xlsx')
except Exception:
    pass
```
