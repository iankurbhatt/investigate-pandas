```python
# Setup imports & load dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import time
import os
df = pd.read_csv('data/raw_transactions.csv', na_values=['Nan', ''])
```

## Section 5: Observation, Slicing, and Subsetting

### Head, Tail, and Sample
**Explanation**: Quick inspection tools. `.head()` fetches the first $N$ rows, `.tail()` gets the last $N$ rows, and `.sample()` retrieves random records to check formatting variation.

**Syntax**: `df.head(n)` / `df.tail(n)` / `df.sample(n)`

```python
print("Head:\n", df.head(2))
print("\nTail:\n", df.tail(2))
print("\nSample:\n", df.sample(2))
```

### Largest & Smallest
**Explanation**: Highly optimized methods to retrieve the top $N$ largest or smallest records. Much faster than sorting the entire DataFrame and calling `.head(N)`.

**Syntax**: `df.nlargest(n, 'col')` / `df.nsmallest(n, 'col')`

```python
print("Largest 3 Amounts:\n", df.nlargest(3, 'transaction_amount'))
print("\nSmallest 3 Amounts:\n", df.nsmallest(3, 'transaction_amount'))
```

### Boolean Indexing
**Explanation**: Filters rows by evaluating a boolean expression across the DataFrame. Rows resolving to `True` are retained.

**Syntax**: `df[df.col > value]`

```python
print(df[df.transaction_amount > 1200].head(3))
```

### Selecting Columns
**Explanation**: Accesses single or multiple features. Passing a single string returns a Series; passing a list of strings returns a DataFrame.

**Syntax**: `df['col']` / `df[['col1', 'col2']]`

```python
print("Single Column:\n", df['customer_id'].head(2))
print("\nMultiple Columns:\n", df[['customer_id', 'transaction_amount']].head(2))
```

### Filtering Columns by Regex
**Explanation**: Selects rows or columns based on regular expression pattern matches on labels. Great for filtering large numbers of features.

**Syntax**: `df.filter(regex='pattern')`

```python
print(df.filter(regex='.*_id|^transaction_').head(2))
```

### Subsetting with .loc & .iloc
**Explanation**: Selection methods. `.loc` references index/column labels (names), and `.iloc` references zero-based index/column integer positions.

**Syntax**: `df.loc[labels]` / `df.iloc[positions]`

```python
print("loc slicing columns:\n", df.loc[0:3, 'transaction_id':'transaction_amount'])
print("\niloc indexing positions:\n", df.iloc[0:3, [0, 2]])
```

### Single Value Retrieval with .at & .iat
**Explanation**: Extremely optimized accessors designed specifically for fetching or setting a single scalar value. Much faster than using `.loc` or `.iloc`.

**Syntax**: `df.at[row_label, col_label]` / `df.iat[row_pos, col_pos]`

```python
print("at (label):", df.at[1, 'transaction_amount'])
print("iat (position):", df.iat[1, 2])
```

### Replacing with .where()
**Explanation**: Retains elements where the condition is True, and replaces elements where it resolves to False with a default value (usually `NaN`).

**Syntax**: `df.where(condition, other_value)`

```python
print(df['transaction_amount'].where(df['transaction_amount'] > 500).head(5))
```

### Query Expressions
**Explanation**: Evaluates boolean conditions written as string queries. Highly readable and optimized for performance under the hood using numexpr.

**Syntax**: `df.query('col1 > value and col2 == "text"')`

```python
print(df.query('transaction_amount > 1000 and region == "North"').head(3))
```
