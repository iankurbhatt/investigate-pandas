### 🌟 What is Pandas & Its Core Data Structures?
**Pandas** is the standard library for data manipulation and analysis in Python. Built on top of NumPy, it provides two primary structures: the 1D labeled **Series** and the 2D labeled **DataFrame**. Pandas makes loading, inspecting, filtering, and summarizing tabular data intuitive and expressive.

This interactive guide loads and works directly with `data/raw_transactions.csv`, giving you real-world hands-on practice.

### 📚 Key Concepts Covered in this Notebook:
- **Core Data Structures**: Covers `pd.Series()` and `pd.DataFrame()`.
- **Inspecting Properties**: Covers `.shape`, `.dtypes`, `.index`, `.columns`, `.values`, `.axes`, `.info()`, and `.describe()`.


```python
# Setup imports & dataset loading
import pandas as pd
import numpy as np
import sys
import time
import os
import sqlite3
import matplotlib.pyplot as plt

# Load raw transactions dataset
csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
df = pd.read_csv(csv_path)
print(f"Pandas Version: {pd.__version__}")
print(f"Loaded raw_transactions.csv: {df.shape[0]} rows, {df.shape[1]} columns")
print(df.head(2))
```

### 🔹 Constructing Series: `pd.Series()`
A Series is a 1D labeled array holding a single column of homogeneous data with an explicit Index. Pandas simplifies tabular workflows, offering expressive one-line operations for filtering, aggregating, joining, and cleaning datasets. **Tip:** Inspect your data types early with `df.dtypes` to ensure numeric values weren't accidentally parsed as text.

**Syntax:** `pd.Series(data=df['transaction_amount'].values, name='amount')`


```python
amount_series = pd.Series(df['transaction_amount'].dropna().values[:5], name='sample_amount')
print('Amount Series:\n', amount_series)
```

### 🔹 Constructing DataFrame: `pd.DataFrame()`
A DataFrame is a 2D labeled tabular structure with heterogeneous column data types. Pandas simplifies tabular workflows, offering expressive one-line operations for filtering, aggregating, joining, and cleaning datasets. **Tip:** Inspect your data types early with `df.dtypes` to ensure numeric values weren't accidentally parsed as text.

**Syntax:** `pd.DataFrame(data=subset_dict)`


```python
sample_df = pd.DataFrame({
    'tx_id': df['transaction_id'][:5],
    'amount': df['transaction_amount'][:5],
    'is_fraud': df['is_fraud'][:5]
})
print('Constructed DataFrame:\n', sample_df)
```
