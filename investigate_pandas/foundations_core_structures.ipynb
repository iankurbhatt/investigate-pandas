# 01. Foundations & Core Data Structures (5+ Years Interview Guide)
Exhaustive revision guide to one-dimensional labeled Series, two-dimensional DataFrames, and property inspection on raw_transactions.csv.

### Key 5-Year Interview Concepts Covered:
- **Core Data Structures**: Dedicated cell for `pd.Series()` and `pd.DataFrame()`.
- **Inspecting Properties**: Dedicated cell for `.shape`, `.dtypes`, `.index`, `.columns`, `.values`, `.axes`, `.info()`, and `.describe()`.

This interactive revision guide uses `data/raw_transactions.csv` for all real-world code examples.

```python
# Setup imports & dataset loading
import pandas as pd
import numpy as np
import sys
import time
import os
import sqlite3
import matplotlib.pyplot as plt

# Load raw transactions dataset
csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
df = pd.read_csv(csv_path)
print(f"Pandas Version: {pd.__version__}")
print(f"Loaded raw_transactions.csv: {df.shape[0]} rows, {df.shape[1]} columns")
print(df.head(2))
```

### Constructing Series: `pd.Series()`
**Explanation**: A Series is a 1D labeled array holding a single column of homogeneous data with an explicit Index.

**Syntax**: `pd.Series(data=df['transaction_amount'].values, name='amount')`

```python
amount_series = pd.Series(df['transaction_amount'].dropna().values[:5], name='sample_amount')
print('Amount Series:\n', amount_series)
```

### Constructing DataFrame: `pd.DataFrame()`
**Explanation**: A DataFrame is a 2D labeled tabular structure with heterogeneous column data types.

**Syntax**: `pd.DataFrame(data=subset_dict)`

```python
sample_df = pd.DataFrame({
    'tx_id': df['transaction_id'][:5],
    'amount': df['transaction_amount'][:5],
    'is_fraud': df['is_fraud'][:5]
})
print('Constructed DataFrame:\n', sample_df)
```

### Inspecting Dimensions: `.shape`
**Explanation**: Returns `(num_rows, num_columns)` in $O(1)$ time.

**Syntax**: `df.shape`

```python
print('Raw Transactions Shape (rows, cols):', df.shape)
```

### Inspecting Column Dtypes: `.dtypes`
**Explanation**: Lists data types for each column in the raw transactions table.

**Syntax**: `df.dtypes`

```python
print('Column Data Types (dtypes):\n', df.dtypes)
```

### Inspecting Row Index: `.index`
**Explanation**: Returns the row index RangeIndex or Index object.

**Syntax**: `df.index`

```python
print('Row Index:', df.index)
```

### Inspecting Column Labels: `.columns`
**Explanation**: Returns an Index object of all transaction column headers.

**Syntax**: `df.columns`

```python
print('Transaction Columns:', df.columns.tolist())
```

### Extracting Raw NumPy Array: `.values` & `.to_numpy()`
**Explanation**: Extracts raw underlying NumPy arrays for machine learning ingestion.

**Syntax**: `df[['transaction_amount', 'is_fraud']].to_numpy()`

```python
print('Raw NumPy 2D Array Slice:\n', df[['transaction_amount', 'is_fraud']].dropna().to_numpy()[:4])
```

### Inspecting Axes: `.axes`
**Explanation**: Returns `[df.index, df.columns]`.

**Syntax**: `df.axes`

```python
print('DataFrame Axes:', [df.axes[0], df.axes[1].tolist()])
```

### Summary & Memory: `.info()`
**Explanation**: Prints non-null counts, dtypes, and exact deep memory usage.

**Syntax**: `df.info(memory_usage='deep')`

```python
df.info(memory_usage='deep')
```

### Descriptive Summary: `.describe()`
**Explanation**: Calculates summary statistics across transaction amounts and account ages.

**Syntax**: `df.describe()`

```python
print('Summary Statistics:\n', df.describe())
```
