### 🔹 Series Construction: `pd.Series()`
- **What it does:** Constructs a one-dimensional ndarray with axis labels (index) and homogeneous data type.
- **Syntax:** `pd.Series()`
- **Key Note:** A Series is essentially a labeled 1D NumPy array with integrated index alignment and missing value handling.
- **Dataset Application & Code Demonstration:** Applies Series Construction on fintech records using columns `transaction_amount` to demonstrate real-world execution.


```python
amount_series = pd.Series(df['transaction_amount'].dropna().values[:5], name='sample_amount')
print('Amount Series:\n', amount_series)
```

### 🔹 DataFrame Construction: `pd.DataFrame()`
- **What it does:** Constructs a two-dimensional, size-mutable, potentially heterogeneous tabular data structure with labeled axes (rows and columns).
- **Syntax:** `pd.DataFrame()`
- **Key Note:** Creating a DataFrame from a dictionary of lists/arrays assigns dictionary keys as column names.
- **Dataset Application & Code Demonstration:** Applies DataFrame Construction on fintech records using columns `is_fraud`, `transaction_amount`, `transaction_id` to demonstrate real-world execution.


```python
sample_df = pd.DataFrame({
    'tx_id': df['transaction_id'][:5],
    'amount': df['transaction_amount'][:5],
    'is_fraud': df['is_fraud'][:5]
})
print('Constructed DataFrame:\n', sample_df)
```

### 🔹 Dimension Inspection: `DataFrame.shape`
- **What it does:** Returns a tuple of integers indicating the size of the array along each dimension.
- **Syntax:** `ndarray.shape / DataFrame.shape / Series.shape`
  - **Parameters:**
    - `object`: The NumPy array or Pandas structure whose dimensional tuple is inspected.
- **Key Note:** `.shape` is a property; modifying `ndarray.shape` directly reshapes in-place without memory copying if contiguous.
- **Dataset Application & Code Demonstration:** Executes Dimension Inspection on the transaction DataFrame (`df`) to inspect and transform tabular features.


```python
print('Raw Transactions Shape (rows, cols):', df.shape)
```

### 🔹 Data Types Inspection: `DataFrame.dtypes`
- **What it does:** Returns the data type object (dtype) describing the element type and byte order.
- **Syntax:** `ndarray.dtype / Series.dtype`
  - **Parameters:**
    - `object`: Target array or series.
- **Key Note:** Homogeneous data types ensure maximum SIMD CPU cache alignment.
- **Dataset Application & Code Demonstration:** Executes Data Types Inspection on the transaction DataFrame (`df`) to inspect and transform tabular features.


```python
print('Column Data Types (dtypes):\n', df.dtypes)
```

### 🔹 Row Index Inspection: `DataFrame.index`
- **What it does:** Returns the index (row labels) object of the DataFrame.
- **Syntax:** `DataFrame.index`
- **Key Note:** Default row indices are memory-efficient `RangeIndex(start=0, stop=n, step=1)` objects.
- **Dataset Application & Code Demonstration:** Executes Row Index Inspection on the transaction DataFrame (`df`) to inspect and transform tabular features.


```python
print('Row Index:', df.index)
```

### 🔹 Column Index Inspection: `DataFrame.columns`
- **What it does:** Returns the column labels of the DataFrame as an Index object.
- **Syntax:** `DataFrame.columns`
- **Key Note:** Modifying `.columns` directly (`df.columns = [c.lower() for c in df.columns]`) renames all columns in-place without copying data.
- **Dataset Application & Code Demonstration:** Executes Column Index Inspection on the transaction DataFrame (`df`) to inspect and transform tabular features.


```python
print('Transaction Columns:', df.columns.tolist())
```

### 🔹 NumPy Array Extraction: `DataFrame.to_numpy()`
- **What it does:** Converts the DataFrame to a NumPy ndarray representation.
- **Syntax:** `DataFrame.to_numpy()`
- **Key Note:** Prefer `.to_numpy()` over `.values` in modern Pandas, as `.to_numpy()` provides explicit control over dtypes and memory copying.
- **Dataset Application & Code Demonstration:** Applies NumPy Array Extraction on fintech records using columns `is_fraud`, `transaction_amount` to demonstrate real-world execution.


```python
print('Raw NumPy 2D Array Slice:\n', df[['transaction_amount', 'is_fraud']].dropna().to_numpy()[:4])
```

### 🔹 Axes Inspection: `DataFrame.axes`
- **What it does:** Returns a list representing the axes of the DataFrame: `[row_index, col_index]`.
- **Syntax:** `DataFrame.axes`
- **Key Note:** `df.axes[0]` corresponds to rows (`df.index`) and `df.axes[1]` corresponds to columns (`df.columns`).
- **Dataset Application & Code Demonstration:** Executes Axes Inspection on the transaction DataFrame (`df`) to inspect and transform tabular features.


```python
print('DataFrame Axes:', [df.axes[0], df.axes[1].tolist()])
```

### 🔹 Structural Summary: `DataFrame.info()`
- **What it does:** Prints a concise summary of a DataFrame including index dtype, column dtypes, non-null values, and memory usage.
- **Syntax:** `DataFrame.info()`
- **Key Note:** Passing `memory_usage='deep'` provides accurate heap memory measurement for object columns.
- **Dataset Application & Code Demonstration:** Executes Structural Summary on the transaction DataFrame (`df`) to inspect and transform tabular features.


```python
df.info(memory_usage='deep')
```

### 🔹 Descriptive Statistics: `DataFrame.describe()`
- **What it does:** Generates descriptive summary statistics that summarize the central tendency, dispersion, and shape of a dataset's distribution.
- **Syntax:** `DataFrame.describe()`
- **Key Note:** Passing `include='all'` produces statistical metrics for both numeric and categorical/string columns in a single table.
- **Dataset Application & Code Demonstration:** Executes Descriptive Statistics on the transaction DataFrame (`df`) to inspect and transform tabular features.


```python
print('Summary Statistics:\n', df.describe())
```
