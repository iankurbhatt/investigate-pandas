```python
# Setup imports & dataset loading
import pandas as pd
import numpy as np
import sys
import time
import os
import sqlite3
import matplotlib.pyplot as plt

# Load raw transactions dataset
csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
df = pd.read_csv(csv_path)
print(f"Pandas Version: {pd.__version__}")
print(f"Loaded raw_transactions.csv: {df.shape[0]} rows, {df.shape[1]} columns")
print(df.head(2))
```

### 🔹 Constructing Series: `pd.Series()`
- **What it does:** A Series is a 1D labeled array holding a single column of homogeneous data with an explicit Index.
- **Syntax:** `pd.Series(data=df['transaction_amount'].values, name='amount')`
- **Key Note:** Inspect your data types early with `df.dtypes` to ensure numeric values weren't accidentally parsed as text.

```python
amount_series = pd.Series(df['transaction_amount'].dropna().values[:5], name='sample_amount')
print('Amount Series:\n', amount_series)
```

### 🔹 Constructing DataFrame: `pd.DataFrame()`
- **What it does:** A DataFrame is a 2D labeled tabular structure with heterogeneous column data types.
- **Syntax:** `pd.DataFrame(data=subset_dict)`
- **Operation:** `sample_df = pd.DataFrame({`
- **Key Note:** Inspect your data types early with `df.dtypes` to ensure numeric values weren't accidentally parsed as text.

```python
sample_df = pd.DataFrame({
    'tx_id': df['transaction_id'][:5],
    'amount': df['transaction_amount'][:5],
    'is_fraud': df['is_fraud'][:5]
})
print('Constructed DataFrame:\n', sample_df)
```

### 🔹 Inspecting Dimensions: `.shape`
- **What it does:** Returns `(num_rows, num_columns)` in $O(1)$ time.
- **Syntax:** `df.shape`
- **Operation:** `print('Raw Transactions Shape (rows, cols):', df.shape)`
- **Key Note:** Inspect your data types early with `df.dtypes` to ensure numeric values weren't accidentally parsed as text.

```python
print('Raw Transactions Shape (rows, cols):', df.shape)
```

### 🔹 Inspecting Column Dtypes: `.dtypes`
- **What it does:** Lists data types for each column in the raw transactions table.
- **Syntax:** `df.dtypes`
- **Operation:** `print('Column Data Types (dtypes):\n', df.dtypes)`
- **Key Note:** Prefer `isinstance(obj, ClassName)` over `type(obj) == ClassName` when checking types in conditional logic.

```python
print('Column Data Types (dtypes):\n', df.dtypes)
```

### 🔹 Inspecting Row Index: `.index`

### 🔹 Inspecting Column Labels: `.columns`

### 🔹 Extracting Raw NumPy Array: `.values` & `.to_numpy()`
- **What it does:** Extracts raw underlying NumPy arrays for machine learning ingestion.
- **Syntax:** `df[['transaction_amount', 'is_fraud']].to_numpy()`
