```python
# Setup imports & load dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import time
import os
df = pd.read_csv('data/raw_transactions.csv', na_values=['Nan', ''])
```

## Section 6: Combining Datasets

### Merging (Joins)
**Explanation**: Combines two DataFrames horizontally based on matching keys. Supports SQL-style joins: left, right, inner, and outer.

**Syntax**: `pd.merge(df1, df2, how='left|right|inner|outer', on='key')`

```python
df_l = df.head(3)[['transaction_id', 'customer_id']]
df_r = df.head(3)[['transaction_id', 'transaction_amount']]
print(pd.merge(df_l, df_r, on='transaction_id', how='inner'))
```

### Concatenations
**Syntax**: `pd.concat([df1, df2], axis=0|1)`

**Explanation**: glues DataFrames along an axis. `axis=0` stacks them vertically (rows), and `axis=1` aligns and joins them horizontally (columns).

```python
df_c1 = pd.DataFrame({'A': [1, 2]})
df_c2 = pd.DataFrame({'A': [3, 4]})
print("Row concat:\n", pd.concat([df_c1, df_c2], axis=0, ignore_index=True))
print("\nCol concat:\n", pd.concat([df_c1, df_c2], axis=1))
```
