```python
# Setup imports & load dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import time
import os
df = pd.read_csv('data/raw_transactions.csv', na_values=['Nan', ''])
```

## Section 7: Descriptive & Statistical Analysis

### Summaries
**Explanation**: Calculates descriptive statistics (mean, counts, percentiles) for numerical columns. Passing `include=['O']` compiles summaries for string/object categories.

**Syntax**: `df.describe()` / `df.describe(include=['O'])`

```python
print("Describe:\n", df.describe())
print("\nUnique Regions:", df.region.unique())
print("\nRegion Frequencies:\n", df.region.value_counts())
```

### Aggregation & Covariance/Correlation
**Explanation**: Core statistical indicators. `.corr()` measures linear relationship strength, and `.cov()` measures scale-dependent variance relationship.

**Syntax**: `df.mean()` / `df.corr()` / `df.cov()`

```python
print("Mean Amount:", df['transaction_amount'].mean())
print("Correlation:\n", df[['transaction_amount', 'account_age_months']].corr())
print("Covariance:\n", df[['transaction_amount', 'account_age_months']].cov())
```

## Section 8: Handling Missing Values

### Null Checks, Fill & Drop
**Explanation**: Essential cleaning methods. Identifies null values (`.isnull()`), fills missing values using a specified strategy (`.fillna()`), or drops rows with nulls (`.dropna()`).

**Syntax**: `df.isnull().sum()` / `df.fillna(val)` / `df.dropna()`

```python
print("Null counts per column:\n", df.isnull().sum())
df_filled_demo = df['transaction_amount'].fillna(df['transaction_amount'].median())
print("Filled amount Null counts:", df_filled_demo.isnull().sum())
```
