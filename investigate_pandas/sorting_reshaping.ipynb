```python
# Setup imports & load dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import time
import os
df = pd.read_csv('data/raw_transactions.csv', na_values=['Nan', ''])
```

## Section 4: Sorting, Reindexing, Renaming, Reshaping, & Dropping

### Sorting Values
**Explanation**: Rearranges rows based on the values in one or more specified columns. Can sort in ascending or descending order.

**Syntax**: `df.sort_values(by='col', ascending=True)`

```python
print(df.sort_values('transaction_amount', ascending=False).head(3))
```

### Sorting by Index
**Explanation**: Rearranges rows or columns based on their index labels rather than the values inside the cells.

**Syntax**: `df.sort_index(ascending=True)`

```python
print(df.sort_index(ascending=False).head(3))
```

### Reset Index
**Explanation**: Resets the index labels to the default integer sequence (0, 1, 2, ...). Setting `drop=True` discards the old index labels completely instead of inserting them as a new column.

**Syntax**: `df.reset_index(drop=True, inplace=True)`

```python
df_sorted = df.sort_values('transaction_amount')
df_sorted.reset_index(drop=True, inplace=True)
print(df_sorted.head(3))
```

### Renaming Columns
**Explanation**: Modifies specific column names using a mapper dictionary. This is useful for normalizing casing, removing spaces, or renaming columns to match a target schema.

**Syntax**: `df.rename(columns={'old':'new'}, inplace=True)`

```python
df_rename = df.rename(columns={'region': 'GEOGRAPHY', 'transaction_amount': 'AMOUNT'})
print(df_rename.columns)
```

### Reshaping: Melt
**Explanation**: Transforms a DataFrame from wide format to long format by unpivoting columns into key-value row pairs. Perfect for normalizing tables for visualization or ingestion.

**Syntax**: `pd.melt(df, id_vars=['id'], value_vars=['col1'])`

```python
df_melt = pd.melt(df.head(2), id_vars=['transaction_id'], value_vars=['region', 'transaction_amount'])
print(df_melt)
```

### Reshaping: Pivot
**Explanation**: Performs a spreadsheet-style pivot aggregation. Reshapes the DataFrame based on column values and aggregates the outputs using functions like mean, sum, or count.

**Syntax**: `df.pivot_table(index='row_col', columns='col_col', values='val_col', aggfunc='mean')`

```python
print(df.pivot_table(index='region', values='transaction_amount', aggfunc='mean'))
```

### Dropping Columns & Rows
**Explanation**: Removes features or observations. `axis=1` deletes columns, and `axis=0` deletes rows. Modifying in-place or returning a new DataFrame is controlled by parameters.

**Syntax**: `df.drop(columns=['col'], axis=1)` / `df.drop([index_labels], axis=0)`

```python
print("Drop region col:\n", df.drop(columns=['region']).head(2))
print("\nDrop rows 0 & 2:\n", df.drop([0, 2], axis=0).head(2))
```
