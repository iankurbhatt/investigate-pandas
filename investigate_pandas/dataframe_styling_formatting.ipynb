# 15. DataFrame Styling & Formatting: Beginner Guide

### 🌟 What is DataFrame Styling & Visual Formatting in Pandas?
Pandas DataFrames can be styled visually for executive reports and notebooks using **`df.style`**. You can highlight maximum/minimum values, display gradient color heatmaps, format currency and percentages, and export publication-ready HTML tables.

This interactive guide loads and works directly with `data/raw_transactions.csv`, giving you real-world hands-on practice.

### 📚 Key Concepts Covered in this Notebook:
- **Highlighting Elements**: Covers `df.style.highlight_max()` and `df.style.highlight_null()`.
- **Custom Formatting**: Covers `df.style.format({'price': '${:.2f}'})`.
- **Color Gradients**: Covers `df.style.background_gradient(cmap='Blues')`.


```python
# Setup imports & dataset loading
import pandas as pd
import numpy as np
import sys
import time
import os
import matplotlib.pyplot as plt

# Load raw transactions dataset
csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
df = pd.read_csv(csv_path)
print(f"Pandas Version: {pd.__version__}")
print(f"Loaded raw_transactions.csv: {df.shape[0]} rows, {df.shape[1]} columns")
```

### 🔹 Highlighting Extrema with `highlight_max()` & `highlight_null()`
Applies CSS styling to flag peak revenue and missing data in regional summaries. Pandas simplifies tabular workflows, offering expressive one-line operations for filtering, aggregating, joining, and cleaning datasets. **Tip:** Inspect your data types early with `df.dtypes` to ensure numeric values weren't accidentally parsed as text.

**Syntax:** `summary_df.style.highlight_max(color='lightgreen').highlight_null(color='pink')`


```python
reg_summary = df.pivot_table(index='region', columns='card_type', values='transaction_amount', aggfunc='mean')
styled_max = reg_summary.style.highlight_max(axis=0, color='lightgreen')
print('Style Object Created:', styled_max)
```

### 🔹 Currency & Percentage Formatting with `df.style.format()`
Formats numeric spending averages as currency strings for executive reporting. Pandas simplifies tabular workflows, offering expressive one-line operations for filtering, aggregating, joining, and cleaning datasets. **Tip:** Inspect your data types early with `df.dtypes` to ensure numeric values weren't accidentally parsed as text.

**Syntax:** `reg_summary.style.format('${:,.2f}')`


```python
styled_fmt = reg_summary.style.format('${:,.2f}')
print('Currency Formatted Table Ready for HTML Export.')
```

### 🔹 Heatmap Color Gradients with `df.style.background_gradient()`
Renders a continuous background color gradient highlighting high-spending regions. Pandas simplifies tabular workflows, offering expressive one-line operations for filtering, aggregating, joining, and cleaning datasets. **Tip:** `.apply()` runs a standard Python loop row-by-row. Whenever possible, use built-in vectorized operations (`df['a'] + df['b']`) which run up to 100x faster!.

**Syntax:** `reg_summary.style.background_gradient(cmap='YlGnBu')`


```python
styled_heatmap = reg_summary.style.background_gradient(cmap='YlGnBu')
print('Heatmap Gradient Initialized.')
```
