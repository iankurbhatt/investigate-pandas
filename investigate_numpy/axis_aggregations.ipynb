# 05. Axis-Based Aggregations: Beginner Guide

### 🌟 What Are Multi-Axis Aggregations in NumPy?
When analyzing multi-dimensional arrays, you often need to calculate metrics across rows or columns. In NumPy, `axis=0` reduces across rows (producing column summaries), while `axis=1` reduces across columns (producing row summaries) for functions like `sum()`, `mean()`, and `std()`.

This interactive guide loads and works directly with `data/raw_transactions.csv`, giving you real-world hands-on practice.

### 📚 Key Concepts Covered in this Notebook:
- **Statistical Reductions**: Covers `np.sum()`, `np.mean()`, `np.std()`, `np.var()`, and `np.min()`/`np.max()`.
- **Directional Arguments**: Dedicated cell comparing `axis=0` (across rows) vs `axis=1` (across columns).
- **Dimension Retention**: Covers `keepdims=True`.
- **Extrema Index Tracking**: Covers `np.argmax()` and `np.argmin()`.


```python
# Setup imports & dataset loading from raw_transactions.csv
import numpy as np
import pandas as pd
import sys
import time
import os

# Load raw transactions and extract aligned NumPy numeric arrays
csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
raw_df = pd.read_csv(csv_path)
clean_raw = raw_df.dropna(subset=['transaction_amount', 'is_fraud', 'account_age_months']).reset_index(drop=True)
amounts = clean_raw['transaction_amount'].to_numpy(dtype=np.float64)
fraud_flags = clean_raw['is_fraud'].to_numpy(dtype=np.int8)
account_ages = clean_raw['account_age_months'].to_numpy(dtype=np.float32)

print(f"NumPy Version: {np.__version__}")
print(f"Loaded from {csv_path} ({len(amounts)} clean aligned rows):")
print(f"- amounts array: shape {amounts.shape}, dtype {amounts.dtype}")
print(f"- fraud_flags array: shape {fraud_flags.shape}, dtype {fraud_flags.dtype}")
print(f"- account_ages array: shape {account_ages.shape}, dtype {account_ages.dtype}")
```

### 🔹 Summation: `np.sum()`
Computes total revenue across all transactions in the dataset. NumPy runs in compiled C memory buffers, enabling mathematical operations across millions of numbers simultaneously in milliseconds. **Tip:** NumPy operations are optimized for homogeneous numeric data, offering massive speed improvements over standard Python loops.

### 🔹 Directional Reductions (`axis=0` vs `axis=1`)
Computes column averages (`axis=0`) vs transaction row sums (`axis=1`) on a multi-feature matrix. NumPy runs in compiled C memory buffers, enabling mathematical operations across millions of numbers simultaneously in milliseconds. **Tip:** NumPy & Pandas Axis Rule: `axis=0` operates along rows (downwards), while `axis=1` operates along columns (horizontally).

**Syntax:** `tx_mat.mean(axis=0)`


```python
tx_mat = np.column_stack([amounts[:100], account_ages[:100]])
print('Column Feature Means (axis=0):', tx_mat.mean(axis=0).round(2))
print('Row Aggregations (axis=1, first 3 rows):', tx_mat.sum(axis=1)[:3].round(2))
```

### 🔹 Dimension Retention with `keepdims=True`
Retains size-1 dimensions for direct broadcasting without reshaping. NumPy runs in compiled C memory buffers, enabling mathematical operations across millions of numbers simultaneously in milliseconds. **Tip:** NumPy operations are optimized for homogeneous numeric data, offering massive speed improvements over standard Python loops.

**Syntax:** `tx_mat.mean(axis=0, keepdims=True)`


```python
kept_means = tx_mat.mean(axis=0, keepdims=True)
print('Shape with keepdims=True:', kept_means.shape)
```

### 🔍 Scenario: Q1: Identifying Peak Spend Days via 2D Reshaped Extrema

**Approach:** Locate the peak daily transaction volume index across a reshaped weekly matrix.
**Syntax:** `np.argmax(weekly_matrix.sum(axis=1))`


```python
weekly_data = amounts[:700].reshape(100, 7)
peak_day_indices = np.argmax(weekly_data, axis=1)
print('Peak spending day of the week for first 5 weeks:', peak_day_indices[:5])
```
