# 05. Axis-Based Aggregations (5+ Years Interview Guide)
Exhaustive revision guide to directional reductions (axis=0 vs 1), keepdims=True, and extrema tracking (argmax/argmin) on raw_transactions.csv.

### Key 5-Year Interview Concepts Covered:
- **Statistical Reductions**: Dedicated cell for `np.sum()`, `np.mean()`, `np.std()`, `np.var()`, and `np.min()`/`np.max()`.
- **Directional Arguments**: Dedicated cell comparing `axis=0` (across rows) vs `axis=1` (across columns).
- **Dimension Retention**: Dedicated cell for `keepdims=True`.
- **Extrema Index Tracking**: Dedicated cell for `np.argmax()` and `np.argmin()`.

This interactive revision guide loads and operates directly on `data/raw_transactions.csv` using dedicated cells per method.

```python
# Setup imports & dataset loading from raw_transactions.csv
import numpy as np
import pandas as pd
import sys
import time
import os

# Load raw transactions and extract aligned NumPy numeric arrays
csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
raw_df = pd.read_csv(csv_path)
clean_raw = raw_df.dropna(subset=['transaction_amount', 'is_fraud', 'account_age_months']).reset_index(drop=True)
amounts = clean_raw['transaction_amount'].to_numpy(dtype=np.float64)
fraud_flags = clean_raw['is_fraud'].to_numpy(dtype=np.int8)
account_ages = clean_raw['account_age_months'].to_numpy(dtype=np.float32)

print(f"NumPy Version: {np.__version__}")
print(f"Loaded from {csv_path} ({len(amounts)} clean aligned rows):")
print(f"- amounts array: shape {amounts.shape}, dtype {amounts.dtype}")
print(f"- fraud_flags array: shape {fraud_flags.shape}, dtype {fraud_flags.dtype}")
print(f"- account_ages array: shape {account_ages.shape}, dtype {account_ages.dtype}")
```

### Summation: `np.sum()`
**Explanation**: Computes total revenue across all transactions in the dataset.

**Syntax**: `np.sum(amounts)`

```python
print(f'Total Transaction Amount (np.sum): ${np.sum(amounts):,.2f}')
```

### Mean Average: `np.mean()`
**Explanation**: Computes arithmetic mean transaction size.

**Syntax**: `np.mean(amounts)`

```python
print(f'Mean Transaction Amount: ${np.mean(amounts):.2f}')
```

### Standard Deviation & Variance: `np.std()` & `np.var()`
**Explanation**: Measures dispersion in customer spending amounts.

**Syntax**: `np.std(amounts)` / `np.var(amounts)`

```python
print(f'Spending Standard Deviation: ${np.std(amounts):.2f}')
print(f'Spending Variance: {np.var(amounts):.2f}')
```

### Extrema Values: `np.min()` & `np.max()`
**Explanation**: Finds smallest and largest transaction amounts in raw_transactions.csv.

**Syntax**: `np.min(amounts)` / `np.max(amounts)`

```python
print(f'Min Transaction Amount: ${np.min(amounts):.2f}')
print(f'Max Transaction Amount: ${np.max(amounts):.2f}')
```

### Directional Reductions (`axis=0` vs `axis=1`)
**Explanation**: Computes column averages (`axis=0`) vs transaction row sums (`axis=1`) on a multi-feature matrix.

**Syntax**: `tx_mat.mean(axis=0)` vs `tx_mat.sum(axis=1)`

```python
tx_mat = np.column_stack([amounts[:100], account_ages[:100]])
print('Column Feature Means (axis=0):', tx_mat.mean(axis=0).round(2))
print('Row Aggregations (axis=1, first 3 rows):', tx_mat.sum(axis=1)[:3].round(2))
```

### Dimension Retention with `keepdims=True`
**Explanation**: Retains size-1 dimensions for direct broadcasting without reshaping.

**Syntax**: `tx_mat.mean(axis=0, keepdims=True)`

```python
kept_means = tx_mat.mean(axis=0, keepdims=True)
print('Shape with keepdims=True:', kept_means.shape)
```

### Extrema Index Tracking with `np.argmax()`
**Explanation**: Locates the exact integer row index of the highest value transaction.

**Syntax**: `np.argmax(amounts)`

```python
max_idx = np.argmax(amounts)
print(f'Largest transaction at index {max_idx} with amount ${amounts[max_idx]:.2f}')
```

### Extrema Index Tracking with `np.argmin()`
**Explanation**: Locates the exact row index of the smallest transaction.

**Syntax**: `np.argmin(amounts)`

```python
min_idx = np.argmin(amounts)
print(f'Smallest transaction at index {min_idx} with amount ${amounts[min_idx]:.2f}')
```

### Q1: Identifying Peak Spend Days via 2D Reshaped Extrema
**Explanation**: Locate the peak daily transaction volume index across a reshaped weekly matrix.

**Syntax**: `np.argmax(weekly_matrix.sum(axis=1))`

```python
weekly_data = amounts[:700].reshape(100, 7)
peak_day_indices = np.argmax(weekly_data, axis=1)
print('Peak spending day of the week for first 5 weeks:', peak_day_indices[:5])
```
