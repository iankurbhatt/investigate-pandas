This interactive revision guide loads and operates directly on `data/raw_transactions.csv` using dedicated cells per method.

```python
# Setup imports & dataset loading from raw_transactions.csv
import numpy as np
import pandas as pd
import sys
import time
import os

# Load raw transactions and extract aligned NumPy numeric arrays
csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
raw_df = pd.read_csv(csv_path)
clean_raw = raw_df.dropna(subset=['transaction_amount', 'is_fraud', 'account_age_months']).reset_index(drop=True)
amounts = clean_raw['transaction_amount'].to_numpy(dtype=np.float64)
fraud_flags = clean_raw['is_fraud'].to_numpy(dtype=np.int8)
account_ages = clean_raw['account_age_months'].to_numpy(dtype=np.float32)

print(f"NumPy Version: {np.__version__}")
print(f"Loaded from {csv_path} ({len(amounts)} clean aligned rows):")
print(f"- amounts array: shape {amounts.shape}, dtype {amounts.dtype}")
print(f"- fraud_flags array: shape {fraud_flags.shape}, dtype {fraud_flags.dtype}")
print(f"- account_ages array: shape {account_ages.shape}, dtype {account_ages.dtype}")
```

### Matrix Multiplication with the `@` Operator
**Explanation**: Multiplies transaction feature matrix by weight vector.

**Syntax**: `X @ weights`

```python
X = np.column_stack([amounts[:100]/100, account_ages[:100]/10])
w = np.array([0.6, 0.4])
risk_scores = X @ w
print('Linear Composite Risk Scores Head:', risk_scores[:5].round(3))
```

### Dot Products with `np.dot()`
**Explanation**: Computes vector dot products for portfolio valuation.

**Syntax**: `np.dot(amounts[:5], np.ones(5))`

```python
print('Total Spend via Dot Product:', np.dot(amounts[:5], np.ones(5)).round(2))
```

### Solving Linear Systems with `np.linalg.solve()`
**Explanation**: Solves ordinary least squares system equations $X^T X \beta = X^T y$.

**Syntax**: `np.linalg.solve(A, b)`

```python
A = (X.T @ X)
b = (X.T @ amounts[:100])
beta = np.linalg.solve(A, b)
print('Fitted OLS Beta Coefficients:', beta.round(4))
```

### Matrix Inversion with `np.linalg.inv()`
**Explanation**: Inverts covariance matrix $(X^T X)^{-1}$.

**Syntax**: `np.linalg.inv(cov_matrix)`

```python
A_inv = np.linalg.inv(A)
print('Inverse Covariance Matrix (A^-1):\n', A_inv)
```

### Eigenvalue Decomposition with `np.linalg.eig()`
**Explanation**: Computes eigenvalues and principal directions from transaction covariance matrix.

**Syntax**: `w, v = np.linalg.eig(cov_mat)`

```python
cov_mat = np.cov(X.T)
eig_vals, eig_vecs = np.linalg.eig(cov_mat)
print('Covariance Eigenvalues:', eig_vals.round(4))
print('Eigenvectors:\n', eig_vecs.round(4))
```

### Singular Value Decomposition with `np.linalg.svd()`
**Explanation**: Factorizes transaction feature matrix into $U \Sigma V^T$ for PCA dimensionality reduction.

**Syntax**: `U, S, Vt = np.linalg.svd(X, full_matrices=False)`

```python
U, S, Vt = np.linalg.svd(X, full_matrices=False)
print('Singular Values (S):', S.round(4))
print('Principal Component Weights (Vt):\n', Vt.round(4))
```

### QR Decomposition with `np.linalg.qr()`
**Explanation**: Factorizes $X = QR$ where $Q$ is orthonormal and $R$ is upper triangular.

**Syntax**: `Q, R = np.linalg.qr(X)`

```python
Q, R = np.linalg.qr(X)
print('Orthogonal Q Shape:', Q.shape, 'Upper Triangular R Shape:', R.shape)
```
