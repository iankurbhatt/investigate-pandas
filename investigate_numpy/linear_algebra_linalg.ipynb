# 11. Linear Algebra & Matrix Decompositions: Beginner Guide

### 🌟 What is Linear Algebra & Matrix Decomposition in NumPy?
Linear algebra powers machine learning and quantitative modeling. NumPy's `np.linalg` module provides optimized routines for matrix multiplication (`@`), matrix inversion, solving systems of linear equations, and performing matrix decompositions (SVD, QR, Eigenvalues).

This interactive guide loads and works directly with `data/raw_transactions.csv`, giving you real-world hands-on practice.

### 📚 Key Concepts Covered in this Notebook:
- **Matrix Multiplication**: Covers `@` operator, `np.matmul()`, and `np.dot()`.
- **Linear Solvers**: Covers `np.linalg.solve()` and `np.linalg.inv()`.
- **Matrix Factorizations**: Covers `np.linalg.eig()`, `np.linalg.svd()`, and `np.linalg.qr()`.


```python
# Setup imports & dataset loading from raw_transactions.csv
import numpy as np
import pandas as pd
import sys
import time
import os

# Load raw transactions and extract aligned NumPy numeric arrays
csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
raw_df = pd.read_csv(csv_path)
clean_raw = raw_df.dropna(subset=['transaction_amount', 'is_fraud', 'account_age_months']).reset_index(drop=True)
amounts = clean_raw['transaction_amount'].to_numpy(dtype=np.float64)
fraud_flags = clean_raw['is_fraud'].to_numpy(dtype=np.int8)
account_ages = clean_raw['account_age_months'].to_numpy(dtype=np.float32)

print(f"NumPy Version: {np.__version__}")
print(f"Loaded from {csv_path} ({len(amounts)} clean aligned rows):")
print(f"- amounts array: shape {amounts.shape}, dtype {amounts.dtype}")
print(f"- fraud_flags array: shape {fraud_flags.shape}, dtype {fraud_flags.dtype}")
print(f"- account_ages array: shape {account_ages.shape}, dtype {account_ages.dtype}")
```

### 🔹 Matrix Multiplication with the `@` Operator
Multiplies transaction feature matrix by weight vector. NumPy runs in compiled C memory buffers, enabling mathematical operations across millions of numbers simultaneously in milliseconds. **Tip:** NumPy operations are optimized for homogeneous numeric data, offering massive speed improvements over standard Python loops.

**Syntax:** `X @ weights`


```python
X = np.column_stack([amounts[:100]/100, account_ages[:100]/10])
w = np.array([0.6, 0.4])
risk_scores = X @ w
print('Linear Composite Risk Scores Head:', risk_scores[:5].round(3))
```

### 🔹 Dot Products with `np.dot()`
Computes vector dot products for portfolio valuation. NumPy runs in compiled C memory buffers, enabling mathematical operations across millions of numbers simultaneously in milliseconds. **Tip:** NumPy operations are optimized for homogeneous numeric data, offering massive speed improvements over standard Python loops.

**Syntax:** `np.dot(amounts[:5], np.ones(5))`


```python
print('Total Spend via Dot Product:', np.dot(amounts[:5], np.ones(5)).round(2))
```

### 🔹 Solving Linear Systems with `np.linalg.solve()`
Solves ordinary least squares system equations $X^T X \beta = X^T y$. NumPy runs in compiled C memory buffers, enabling mathematical operations across millions of numbers simultaneously in milliseconds. **Tip:** NumPy operations are optimized for homogeneous numeric data, offering massive speed improvements over standard Python loops.

**Syntax:** `np.linalg.solve(A, b)`


```python
A = (X.T @ X)
b = (X.T @ amounts[:100])
beta = np.linalg.solve(A, b)
print('Fitted OLS Beta Coefficients:', beta.round(4))
```

### 🔹 Matrix Inversion with `np.linalg.inv()`
Inverts covariance matrix $(X^T X)^{-1}$. NumPy runs in compiled C memory buffers, enabling mathematical operations across millions of numbers simultaneously in milliseconds. **Tip:** NumPy operations are optimized for homogeneous numeric data, offering massive speed improvements over standard Python loops.

**Syntax:** `np.linalg.inv(cov_matrix)`


```python
A_inv = np.linalg.inv(A)
print('Inverse Covariance Matrix (A^-1):\n', A_inv)
```

### 🔹 Eigenvalue Decomposition with `np.linalg.eig()`
Computes eigenvalues and principal directions from transaction covariance matrix. NumPy runs in compiled C memory buffers, enabling mathematical operations across millions of numbers simultaneously in milliseconds. **Tip:** NumPy operations are optimized for homogeneous numeric data, offering massive speed improvements over standard Python loops.

**Syntax:** `w, v = np.linalg.eig(cov_mat)`


```python
cov_mat = np.cov(X.T)
eig_vals, eig_vecs = np.linalg.eig(cov_mat)
print('Covariance Eigenvalues:', eig_vals.round(4))
print('Eigenvectors:\n', eig_vecs.round(4))
```

### 🔹 Singular Value Decomposition with `np.linalg.svd()`
Factorizes transaction feature matrix into $U \Sigma V^T$ for PCA dimensionality reduction. NumPy runs in compiled C memory buffers, enabling mathematical operations across millions of numbers simultaneously in milliseconds. **Tip:** NumPy operations are optimized for homogeneous numeric data, offering massive speed improvements over standard Python loops.

**Syntax:** `U, S, Vt = np.linalg.svd(X, full_matrices=False)`


```python
U, S, Vt = np.linalg.svd(X, full_matrices=False)
print('Singular Values (S):', S.round(4))
print('Principal Component Weights (Vt):\n', Vt.round(4))
```

### 🔍 Scenario: QR Decomposition with `np.linalg.qr()`

**Approach:** Factorizes $X = QR$ where $Q$ is orthonormal and $R$ is upper triangular.
**Syntax:** `Q, R = np.linalg.qr(X)`


```python
Q, R = np.linalg.qr(X)
print('Orthogonal Q Shape:', Q.shape, 'Upper Triangular R Shape:', R.shape)
```

### 🔍 Scenario: Q1: Portfolio Risk Variance Calculation

**Approach:** Given transaction asset weights $w$, calculate total portfolio variance $w^T \Sigma w$.
**Syntax:** `w.T @ cov_mat @ w`


```python
port_var = w.T @ cov_mat @ w
print('Total Portfolio Variance:', round(port_var, 6))
```
