# 01. Array Foundations & Memory Mechanics (5+ Years Interview Guide)
Exhaustive revision guide to homogeneous memory allocation, creation routines, and ndarray properties on raw_transactions.csv.

### Key 5-Year Interview Concepts Covered:
- **Homogeneous Memory Allocation**: Flat contiguous C-memory buffer storing raw binary bytes without PyObject wrapper overhead.
- **Array Creation Routines**: Dedicated cell for `np.array()`, `np.zeros()`, `np.ones()`, `np.empty()`, and `np.full()`.
- **Sequence Generation**: Dedicated cell for `np.arange()`, `np.linspace()`, and `np.logspace()`.
- **Inspecting Properties**: Dedicated cell for `.ndim`, `.shape`, `.size`, `.dtype`, `.itemsize`, and `.nbytes`.

This interactive revision guide loads and operates directly on `data/raw_transactions.csv` using dedicated cells per method.

```python
# Setup imports & dataset loading from raw_transactions.csv
import numpy as np
import pandas as pd
import sys
import time
import os

# Load raw transactions and extract aligned NumPy numeric arrays
csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
raw_df = pd.read_csv(csv_path)
clean_raw = raw_df.dropna(subset=['transaction_amount', 'is_fraud', 'account_age_months']).reset_index(drop=True)
amounts = clean_raw['transaction_amount'].to_numpy(dtype=np.float64)
fraud_flags = clean_raw['is_fraud'].to_numpy(dtype=np.int8)
account_ages = clean_raw['account_age_months'].to_numpy(dtype=np.float32)

print(f"NumPy Version: {np.__version__}")
print(f"Loaded from {csv_path} ({len(amounts)} clean aligned rows):")
print(f"- amounts array: shape {amounts.shape}, dtype {amounts.dtype}")
print(f"- fraud_flags array: shape {fraud_flags.shape}, dtype {fraud_flags.dtype}")
print(f"- account_ages array: shape {account_ages.shape}, dtype {account_ages.dtype}")
```

### Constructing Arrays: `np.array()`
**Explanation**: Creates an `ndarray` from transaction amount lists.

**Syntax**: `np.array(amounts[:10], dtype=np.float64)`

```python
amt_sample = np.array(amounts[:5], dtype=np.float64)
print('Constructed Amount Array:', amt_sample)
print('Dtype:', amt_sample.dtype)
```

### Allocating Zeros: `np.zeros()`
**Explanation**: Allocates zeroed memory of shape matching transaction subsets.

**Syntax**: `np.zeros(len(amounts), dtype=np.float32)`

```python
zeros_weights = np.zeros((4, 4), dtype=np.float64)
print('Zeros Weight Matrix (4, 4):\n', zeros_weights)
```

### Allocating Ones: `np.ones()`
**Explanation**: Allocates ones initialized array for scaling factors.

**Syntax**: `np.ones(shape=(3, 3), dtype=np.int32)`

```python
ones_scale = np.ones((2, 3), dtype=np.int32)
print('Ones Array:\n', ones_scale)
```

### Uninitialized Fast Memory: `np.empty()`
**Explanation**: Fastest allocation routine returning raw uninitialized RAM bytes.

**Syntax**: `np.empty(shape=(2, 2))`

```python
empty_buf = np.empty((2, 2), dtype=np.float64)
print('Empty Uninitialized Memory:\n', empty_buf)
```

### Constant-Filled Array: `np.full()`
**Explanation**: Fills an array of transaction count with default risk score 0.5.

**Syntax**: `np.full(shape=(5,), fill_value=0.5)`

```python
default_risk = np.full(shape=(5,), fill_value=0.5)
print('Default Risk Scores:', default_risk)
```

### Stepped Range Sequence: `np.arange()`
**Explanation**: Generates integer transaction index offsets.

**Syntax**: `np.arange(0, 10, 2)`

```python
idx_offsets = np.arange(0, 10, 2)
print('Stepped Range:', idx_offsets)
```

### Linear Evenly Spaced Grids: `np.linspace()`
**Explanation**: Generates evenly spaced threshold intervals between min and max transaction amount.

**Syntax**: `np.linspace(amounts.min(), amounts.max(), num=5)`

```python
spend_grid = np.linspace(amounts.min(), amounts.max(), num=5)
print('Linspace Grid from Min to Max Transaction Amount:', spend_grid.round(2))
```

### Logarithmic Grids: `np.logspace()`
**Explanation**: Generates geometrically spaced transaction cutoff points.

**Syntax**: `np.logspace(1, 4, num=4)`

```python
log_cutoffs = np.logspace(1, 4, num=4)
print('Logarithmic Cutoffs ($10 to $10,000):', log_cutoffs)
```

### Inspecting Dimensions: `.ndim`
**Explanation**: Returns array dimension rank.

**Syntax**: `amounts.ndim`

```python
print('Amounts Array Dimensions (ndim):', amounts.ndim)
```

### Inspecting Shape: `.shape`
**Explanation**: Returns total transaction count tuple.

**Syntax**: `amounts.shape`

```python
print('Amounts Array Shape:', amounts.shape)
```

### Inspecting Total Elements: `.size`
**Explanation**: Returns total transaction elements count.

**Syntax**: `amounts.size`

```python
print('Total Elements in Amounts:', amounts.size)
```

### Inspecting Dtype: `.dtype`
**Explanation**: Returns binary primitive data type.

**Syntax**: `amounts.dtype`

```python
print('Amounts Data Type:', amounts.dtype)
```

### Inspecting Item Size: `.itemsize`
**Explanation**: Returns bytes per transaction element (8 bytes for float64).

**Syntax**: `amounts.itemsize`

```python
print('Bytes per Element (itemsize):', amounts.itemsize)
```

### Inspecting Buffer Memory: `.nbytes`
**Explanation**: Calculates total raw RAM consumed by the transaction amounts vector.

**Syntax**: `amounts.nbytes`

```python
print(f'Total Memory Footprint (nbytes): {amounts.nbytes} bytes ({amounts.nbytes / 1024:.2f} KB)')
```

### Q1: Memory Consumption: Python Float List vs NumPy Transaction Vector
**Explanation**: Compare memory footprint of transaction amounts in Python float list vs NumPy float64 ndarray.

**Syntax**: `sys.getsizeof(py_list) + sum(sys.getsizeof(x) for x in py_list)` vs `amounts.nbytes`

```python
py_amounts = list(amounts)
py_mem = sys.getsizeof(py_amounts) + sum(sys.getsizeof(x) for x in py_amounts)
np_mem = amounts.nbytes
print(f'Python List Memory ({len(amounts)} floats): {py_mem / 1024:.1f} KB')
print(f'NumPy ndarray Memory ({len(amounts)} floats): {np_mem / 1024:.1f} KB ({(1 - np_mem/py_mem)*100:.1f}% RAM reduction)')
```
