```python
# Setup imports & dataset loading from raw_transactions.csv
import numpy as np
import pandas as pd
import sys
import time
import os

# Load raw transactions and extract aligned NumPy numeric arrays
csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
raw_df = pd.read_csv(csv_path)
clean_raw = raw_df.dropna(subset=['transaction_amount', 'is_fraud', 'account_age_months']).reset_index(drop=True)
amounts = clean_raw['transaction_amount'].to_numpy(dtype=np.float64)
fraud_flags = clean_raw['is_fraud'].to_numpy(dtype=np.int8)
account_ages = clean_raw['account_age_months'].to_numpy(dtype=np.float32)

print(f"NumPy Version: {np.__version__}")
print(f"Loaded from {csv_path} ({len(amounts)} clean aligned rows):")
print(f"- amounts array: shape {amounts.shape}, dtype {amounts.dtype}")
print(f"- fraud_flags array: shape {fraud_flags.shape}, dtype {fraud_flags.dtype}")
print(f"- account_ages array: shape {account_ages.shape}, dtype {account_ages.dtype}")
```

### 🔹 Array Construction: `np.array()`
- **What it does:** Constructs an N-dimensional array (ndarray) from an array-like sequence with specified data type and memory order.
- **Syntax:** `np.array()`
- **Key Note:** NumPy arrays enforce homogeneous data types, enabling SIMD vectorization and high CPU cache performance.
- **Dataset Application & Code Demonstration:** Applies Array Construction across the extracted numeric transaction `amounts` array to compute performance metrics.


```python
amt_sample = np.array(amounts[:5], dtype=np.float64)
print('Constructed Amount Array:', amt_sample)
print('Dtype:', amt_sample.dtype)
```

### 🔹 Zero-Filled Array: `np.zeros()`
- **What it does:** Returns a new array of given shape and type, initialized with zeros.
- **Syntax:** `np.zeros()`
- **Key Note:** `np.zeros` zeroes out memory pages, ensuring reproducible zero initialization.
- **Dataset Application & Code Demonstration:** Demonstrates Zero-Filled Array with practical fintech data structures and variables in the following code block.


```python
zeros_weights = np.zeros((4, 4), dtype=np.float64)
print('Zeros Weight Matrix (4, 4):\n', zeros_weights)
```

### 🔹 Ones-Filled Array: `np.ones()`
- **What it does:** Returns a new array of given shape and type, filled with ones.
- **Syntax:** `np.ones()`
- **Key Note:** Useful for initializing identity weights, unit scaling vectors, and multiplication identities.
- **Dataset Application & Code Demonstration:** Demonstrates Ones-Filled Array with practical fintech data structures and variables in the following code block.


```python
ones_scale = np.ones((2, 3), dtype=np.int32)
print('Ones Array:\n', ones_scale)
```

### 🔹 Uninitialized Memory Allocation: `np.empty()`
- **What it does:** Returns a new array of given shape and type without initializing memory buffer entries (contains uninitialized values).
- **Syntax:** `np.empty()`
- **Key Note:** `np.empty()` is the fastest array allocation method because it skips zero-filling RAM; always overwrite before reading.
- **Dataset Application & Code Demonstration:** Demonstrates Uninitialized Memory Allocation with practical fintech data structures and variables in the following code block.


```python
empty_buf = np.empty((2, 2), dtype=np.float64)
print('Empty Uninitialized Memory:\n', empty_buf)
```

### 🔹 Constant Value Array: `np.full()`
- **What it does:** Returns a new array of given shape and type, filled with a specified scalar constant value.
- **Syntax:** `np.full()`
- **Key Note:** Equivalent to `np.empty(shape) * fill_value` but faster and type-safe.
- **Dataset Application & Code Demonstration:** Demonstrates Constant Value Array with practical fintech data structures and variables in the following code block.


```python
default_risk = np.full(shape=(5,), fill_value=0.5)
print('Default Risk Scores:', default_risk)
```

### 🔹 Stepped Range Sequence: `np.arange()`
- **What it does:** Generates integer elements index offsets.
- **Syntax:** `np.arange()`
- **Key Note:** Use `np.arange(start, stop, step)` when you know the step size. Use `np.linspace(start, stop, num)` when you know the exact number of points you want.
- **Dataset Application & Code Demonstration:** Demonstrates Stepped Range Sequence with practical fintech data structures and variables in the following code block.


```python
idx_offsets = np.arange(0, 10, 2)
print('Stepped Range:', idx_offsets)
```

### 🔹 Linear Evenly Spaced Grids: `np.linspace()`
- **What it does:** Generates evenly spaced threshold intervals between min and max numerical values.
- **Syntax:** `np.linspace()`
- **Key Note:** Use `np.arange(start, stop, step)` when you know the step size. Use `np.linspace(start, stop, num)` when you know the exact number of points you want.
- **Dataset Application & Code Demonstration:** Applies Linear Evenly Spaced Grids across the extracted numeric transaction `amounts` array to compute performance metrics.


```python
spend_grid = np.linspace(amounts.min(), amounts.max(), num=5)
print('Linspace Grid from Min to Max Transaction Amount:', spend_grid.round(2))
```

### 🔹 Logarithmic Grids: `np.logspace()`
- **What it does:** Generates geometrically spaced elements cutoff points.
- **Syntax:** `np.logspace()`
- **Key Note:** NumPy operations are optimized for homogeneous numeric data, offering massive speed improvements over standard Python loops.
- **Dataset Application & Code Demonstration:** Demonstrates Logarithmic Grids with practical fintech data structures and variables in the following code block.


```python
log_cutoffs = np.logspace(1, 4, num=4)
print('Logarithmic Cutoffs ($10 to $10,000):', log_cutoffs)
```

### 🔹 Inspecting Dimensions: `.ndim`
- **What it does:** Returns an integer representing the number of array dimensions (axes).
- **Syntax:** `ndarray.ndim`
  - **Parameters:**
    - `ndarray`: The N-dimensional NumPy array whose dimension count is inspected.
- **Key Note:** 1D array has `.ndim == 1`, 2D matrix has `.ndim == 2`, 3D tensor has `.ndim == 3`.
- **Dataset Application & Code Demonstration:** Applies Inspecting Dimensions across the extracted numeric transaction `amounts` array to compute performance metrics.


```python
print('Amounts Array Dimensions (ndim):', amounts.ndim)
```

### 🔹 Inspecting Shape: `.shape`
- **What it does:** Returns a tuple of integers indicating the size of the array along each dimension.
- **Syntax:** `ndarray.shape / DataFrame.shape / Series.shape`
  - **Parameters:**
    - `object`: The NumPy array or Pandas structure whose dimensional tuple is inspected.
- **Key Note:** `.shape` is a property; modifying `ndarray.shape` directly reshapes in-place without memory copying if contiguous.
- **Dataset Application & Code Demonstration:** Applies Inspecting Shape across the extracted numeric transaction `amounts` array to compute performance metrics.


```python
print('Amounts Array Shape:', amounts.shape)
```

### 🔹 Inspecting Total Elements: `.size`
- **What it does:** Returns the total number of elements contained across all dimensions.
- **Syntax:** `ndarray.size / Series.size`
  - **Parameters:**
    - `object`: Target array or series.
- **Key Note:** For 2D array of shape (4, 4), `.size == 16`.
- **Dataset Application & Code Demonstration:** Applies Inspecting Total Elements across the extracted numeric transaction `amounts` array to compute performance metrics.


```python
print('Total Elements in Amounts:', amounts.size)
```

### 🔹 Inspecting Dtype: `.dtype`
- **What it does:** Returns the data type object (dtype) describing the element type and byte order.
- **Syntax:** `ndarray.dtype / Series.dtype`
  - **Parameters:**
    - `object`: Target array or series.
- **Key Note:** Homogeneous data types ensure maximum SIMD CPU cache alignment.
- **Dataset Application & Code Demonstration:** Applies Inspecting Dtype across the extracted numeric transaction `amounts` array to compute performance metrics.


```python
print('Amounts Data Type:', amounts.dtype)
```

### 🔹 Inspecting Item Size: `.itemsize`
- **What it does:** Returns the total number of elements contained across all dimensions.
- **Syntax:** `ndarray.size / Series.size`
  - **Parameters:**
    - `object`: Target array or series.
- **Key Note:** For 2D array of shape (4, 4), `.size == 16`.
- **Dataset Application & Code Demonstration:** Applies Inspecting Item Size across the extracted numeric transaction `amounts` array to compute performance metrics.


```python
print('Bytes per Element (itemsize):', amounts.itemsize)
```

### 🔹 Inspecting Buffer Memory: `.nbytes`
- **What it does:** Returns the total number of bytes consumed by the elements of the array in RAM.
- **Syntax:** `ndarray.nbytes`
  - **Parameters:**
    - `ndarray`: Target array.
- **Key Note:** Does not include memory consumed by array container object overhead.
- **Dataset Application & Code Demonstration:** Applies Inspecting Buffer Memory across the extracted numeric transaction `amounts` array to compute performance metrics.


```python
print(f'Total Memory Footprint (nbytes): {amounts.nbytes} bytes ({amounts.nbytes / 1024:.2f} KB)')
```

### 🔍 Scenario: Q1: Memory Consumption: Python Float List vs NumPy Transaction Vector
- **Objective:** Q1: Memory Consumption: Python Float List vs NumPy Transaction Vector
- **Approach:** Compare memory footprint of transaction amounts in Python float list vs NumPy float64 ndarray.
- **Syntax:** `sys.getsizeof(py_list) + sum(sys.getsizeof(x) for x in py_list)` vs `amounts.nbytes`

```python
py_amounts = list(amounts)
py_mem = sys.getsizeof(py_amounts) + sum(sys.getsizeof(x) for x in py_amounts)
np_mem = amounts.nbytes
print(f'Python List Memory ({len(amounts)} floats): {py_mem / 1024:.1f} KB')
print(f'NumPy ndarray Memory ({len(amounts)} floats): {np_mem / 1024:.1f} KB ({(1 - np_mem/py_mem)*100:.1f}% RAM reduction)')
```
