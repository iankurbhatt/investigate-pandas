# 01. Array Foundations & Memory Mechanics: Beginner Guide

### 🌟 What Are NumPy Array Foundations & Memory Mechanics?
**NumPy (Numerical Python)** is the core library for numerical computing in Python. Standard Python lists store scattered object references, which makes math slow. NumPy introduces the **`ndarray` (N-Dimensional Array)**—a flat, contiguous block of memory where all elements share the same data type. This notebook teaches how memory is allocated, how to initialize arrays, and how to inspect array properties.

This interactive guide loads and works directly with `data/raw_transactions.csv`, giving you real-world hands-on practice.

### 📚 Key Concepts Covered in this Notebook:
- **Homogeneous Memory Allocation**: Flat contiguous C-memory buffer storing raw binary bytes without PyObject wrapper overhead.
- **Array Creation Routines**: Covers `np.array()`, `np.zeros()`, `np.ones()`, `np.empty()`, and `np.full()`.
- **Sequence Generation**: Covers `np.arange()`, `np.linspace()`, and `np.logspace()`.
- **Inspecting Properties**: Covers `.ndim`, `.shape`, `.size`, `.dtype`, `.itemsize`, and `.nbytes`.


```python
# Setup imports & dataset loading from raw_transactions.csv
import numpy as np
import pandas as pd
import sys
import time
import os

# Load raw transactions and extract aligned NumPy numeric arrays
csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
raw_df = pd.read_csv(csv_path)
clean_raw = raw_df.dropna(subset=['transaction_amount', 'is_fraud', 'account_age_months']).reset_index(drop=True)
amounts = clean_raw['transaction_amount'].to_numpy(dtype=np.float64)
fraud_flags = clean_raw['is_fraud'].to_numpy(dtype=np.int8)
account_ages = clean_raw['account_age_months'].to_numpy(dtype=np.float32)

print(f"NumPy Version: {np.__version__}")
print(f"Loaded from {csv_path} ({len(amounts)} clean aligned rows):")
print(f"- amounts array: shape {amounts.shape}, dtype {amounts.dtype}")
print(f"- fraud_flags array: shape {fraud_flags.shape}, dtype {fraud_flags.dtype}")
print(f"- account_ages array: shape {account_ages.shape}, dtype {account_ages.dtype}")
```

### 🔹 Constructing Arrays: `np.array()`
Creates an `ndarray` from transaction amount lists. ### 🔍 Scenario: Q1: Memory Consumption: Python Float List vs NumPy Transaction Vector

**Approach:** Compare memory footprint of transaction amounts in Python float list vs NumPy float64 ndarray.
**Syntax:** `sys.getsizeof(py_list) + sum(sys.getsizeof(x) for x in py_list)` vs `amounts.nbytes`


```python
py_amounts = list(amounts)
py_mem = sys.getsizeof(py_amounts) + sum(sys.getsizeof(x) for x in py_amounts)
np_mem = amounts.nbytes
print(f'Python List Memory ({len(amounts)} floats): {py_mem / 1024:.1f} KB')
print(f'NumPy ndarray Memory ({len(amounts)} floats): {np_mem / 1024:.1f} KB ({(1 - np_mem/py_mem)*100:.1f}% RAM reduction)')
```
