### 1. Set Uniqueness
**Explanation**: Unordered, duplicate-free hash collections.

**Syntax**:
```python
set_variable = {value_one, value_two}
```



````python
unique_card_providers = {1, 1, 2}
print(unique_card_providers)
````

### 2. Set membership search O(1) checks
**Explanation**: Containment checks are O(1) complexity.

**Syntax**:
```python
value in set_object
```



````python
unique_card_providers = {1, 2}
print(1 in unique_card_providers)
````

### 3. Adding set items
**Explanation**: Insert items to sets.

**Syntax**:
```python
set_object.add(value)
```



````python
unique_card_providers = {1}
unique_card_providers.add(2)
print(unique_card_providers)
````

### 4. Removing set items safely
**Explanation**: Discard removes items safely; remove raises errors.

**Syntax**:
```python
set_object.discard(value)
set_object.remove(value)
```



````python
unique_card_providers = {1}
unique_card_providers.discard(9)
print(unique_card_providers)
````

### 5. Set pop operations
**Explanation**: Pops arbitrary elements from sets.

**Syntax**:
```python
value = set_object.pop()
```



````python
unique_card_providers = {1, 2}
print(unique_card_providers.pop(), unique_card_providers)
````

### 6. Clearing sets
**Explanation**: Removes all set elements.

**Syntax**:
```python
set_object.clear()
```



````python
unique_card_providers = {1, 2}
unique_card_providers.clear()
print(unique_card_providers)
````

### 13. Immutable frozenset keys
**Explanation**: Frozensets are hashable set keys.

**Syntax**:


### 14. Set comprehensions syntax
**Explanation**: Construct sets inline.

**Syntax**:

````python
print({x % 3 for x in range(10)})
````

### 15. Converting sequences to set

## Section 3: Fintech Interview Questions

### Q1: Extract the unique card types from the first 50 transactions, and verify if they are a subset of {'Visa', 'MasterCard', 'Amex', 'Discover'}.

````python
# Solution:
cards = set()
with open(csv_path, 'r') as f:
    f.readline()
    for _ in range(50):
        row = f.readline().strip().split(',')
        cards.add(row[4].strip())
        
allowed = {'Visa', 'MasterCard', 'Amex', 'Discover', 'Maestro'}
print('Unique Cards:', cards)
print('Is subset?:', cards.issubset(allowed))
````

### Q2: Find set symmetric difference of merchant IDs between first 50 transactions and the next 50 transactions.

````python
# Solution:
m1 = set()
m2 = set()
with open(csv_path, 'r') as f:
    f.readline()
    for i in range(100):
        row = f.readline().strip().split(',')
        if i < 50: m1.add(row[2])
        else: m2.add(row[2])
        
print('Symmetric Difference count:', len(m1 ^ m2))
````
