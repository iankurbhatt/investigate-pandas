### 1. Set Uniqueness & Hash Table Architecture
**Explanation**: A Python set (`PySetObject`) is an unordered collection of distinct, hashable objects implemented as a hash table with dummy value pointers. Duplicate elements are automatically discarded on insertion. Because sets rely on hashing, all elements must be immutable and implement `__hash__()`.

**Syntax**: `unique_set = {elem1, elem2, elem3}` / `empty_set = set()`

```python
unique_card_providers = {1, 1, 2}
print(unique_card_providers)
```

### 2. O(1) Membership Testing Complexity
**Explanation**: Testing `item in my_set` runs in O(1) constant average time via hash lookup, compared to O(N) linear time for lists and tuples. In interview scenarios requiring deduplication or filtering against an allowed list of IDs, converting search collections to a `set` eliminates O(N²) bottlenecks.

**Syntax**: `if item in allowed_set: ...  # O(1) lookup`

```python
unique_card_providers = {1, 2}
print(1 in unique_card_providers)
```

### 3. Adding Elements (`add()` vs `update()`)
**Explanation**: `.add(elem)` inserts a single hashable element into the set in O(1) time. `.update(*iterables)` takes one or more iterables and adds all their individual elements to the set in-place.

**Syntax**: `my_set.add(item)` / `my_set.update(list_a, list_b)`

```python
unique_card_providers = {1}
unique_card_providers.add(2)
print(unique_card_providers)
```

### 4. Removing Elements Safely (`remove()` vs `discard()`)
**Explanation**: `.remove(elem)` removes the specified element in O(1) time, but raises a `KeyError` if the element is not found. In contrast, `.discard(elem)` removes the element if present and silently does nothing if absent, preventing unnecessary `try-except` boilerplate.

**Syntax**: `my_set.discard(item)  # Safe deletion` / `my_set.remove(item)`

```python
unique_card_providers = {1}
unique_card_providers.discard(9)
print(unique_card_providers)
```

### 5. Arbitrary Element Removal (`pop()`)
**Explanation**: `.pop()` removes and returns an arbitrary element from the set in O(1) time (raises `KeyError` if empty). Because sets are unordered, you cannot predict which element will be popped.

**Syntax**: `elem = my_set.pop()`

```python
unique_card_providers = {1, 2}
print(unique_card_providers.pop(), unique_card_providers)
```

### 6. Clearing Sets (`clear()`)
**Explanation**: `.clear()` removes all elements from the set in-place, resetting its length to 0 while keeping the same memory object.

**Syntax**: `my_set.clear()`

```python
unique_card_providers = {1, 2}
unique_card_providers.clear()
print(unique_card_providers)
```

### 7. Set Union Algebra (`|` and `union()`)
**Explanation**: The union of two sets returns a new set containing all elements present in either set. The operator `set_a | set_b` requires both operands to be sets, while `set_a.union(iterable)` accepts any arbitrary iterable (lists, tuples, generators).

**Syntax**: `combined = set_a | set_b` / `set_a.union(iterable)`

```python
print({1} | {2})
```

### 8. Set Intersection Algebra (`&` and `intersection()`)
**Explanation**: The intersection returns a new set containing only elements present in both sets. In database terms, this is equivalent to an `INNER JOIN` on keys, running in O(min(len(s1), len(s2))) time.

**Syntax**: `shared = set_a & set_b` / `set_a.intersection(iterable)`

```python
print({1, 2} & {2, 3})
```

### 9. Set Difference Algebra (`-` and `difference()`)
**Explanation**: The difference `set_a - set_b` returns a new set with elements that exist in `set_a` but NOT in `set_b`. This is standard for identifying missing IDs, revoked permissions, or dropped records in reconciliation pipelines.

**Syntax**: `diff = set_a - set_b` / `set_a.difference(iterable)`

```python
print({1, 2} - {2, 3})
```

### 10. Set Symmetric Difference Algebra (`^` and `symmetric_difference()`)
**Explanation**: The symmetric difference `set_a ^ set_b` returns elements that exist in exactly one of the sets, but not both (XOR operation). It is equivalent to `(set_a | set_b) - (set_a & set_b)` and is used to detect dataset divergence.

**Syntax**: `diverged = set_a ^ set_b` / `set_a.symmetric_difference(iterable)`

```python
print({1, 2} ^ {2, 3})
```

### 11. Subsets and Supersets (`<=`, `<`, `>=`, `>`)
**Explanation**: `set_a.issubset(set_b)` (or `set_a <= set_b`) checks if every element in `set_a` is contained in `set_b`. `set_a.issuperset(set_b)` (or `set_a >= set_b`) checks if `set_a` contains all elements of `set_b`. Strict comparisons `<` and `>` check proper subset/superset relationships (must be strictly unequal).

**Syntax**: `is_sub = set_a <= set_b` / `set_a.issubset(set_b)`

```python
print({1}.issubset({1, 2}))
```

### 12. Disjoint Sets Validation (`isdisjoint()`)
**Explanation**: `set_a.isdisjoint(set_b)` returns `True` if `set_a` and `set_b` have no elements in common (their intersection is empty). CPython optimizes this by iterating over the smaller set and performing O(1) lookups in the larger set, returning `False` on the first shared element.

**Syntax**: `has_no_overlap = set_a.isdisjoint(set_b)`

```python
print({1}.isdisjoint({2}))
```

### 13. Immutable Sets with `frozenset`
**Explanation**: A `frozenset` is an immutable, hashable version of a set. Because its contents cannot be modified after creation, it can be used as a dictionary key or stored as an element within another set (enabling sets of sets).

**Syntax**: `frozen_key = frozenset([1, 2, 3])`

```python
frozenset_security_key = frozenset([1, 2])
security_dict = {frozenset_security_key: 'OK'}
print(security_dict[frozenset_security_key])
```

### 14. Set Comprehensions
**Explanation**: Set comprehensions `{expr for item in iterable if cond}` construct deduplicated sets declaratively in a single optimized pass. They are ideal for extracting unique attributes (such as distinct countries or status codes) from raw data streams.

**Syntax**: `{row['region'] for row in transactions if row['amount'] > 100}`

```python
print({x % 3 for x in range(10)})
```

### 15. Sequence Conversion to Set (`set()`)
**Explanation**: The `set(iterable)` constructor consumes any iterable and returns a set containing only unique elements. This is the fastest way in Python to eliminate duplicates from a list in O(N) time (note that it does not preserve original element ordering; use `dict.fromkeys(list)` if insertion order must be preserved).

**Syntax**: `unique_items = set(list_with_duplicates)`

```python
print(set('abac'))
```

### Q2: Dataset Divergence via Symmetric Difference
**Explanation**: **Scenario**: Compare merchant IDs between the first 50 transactions and the next 50 transactions to identify merchant IDs unique to each batch using symmetric difference.

**Syntax**: `divergent_merchants = batch1_merchants ^ batch2_merchants`

```python
# Solution:
m1 = set()
m2 = set()
with open(csv_path, 'r') as f:
    f.readline()
    for i in range(100):
        row = f.readline().strip().split(',')
        if i < 50: m1.add(row[2])
        else: m2.add(row[2])
        
print('Symmetric Difference count:', len(m1 ^ m2))
```
