```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
print('Sample Record:', transactions[0])
```

### 🔹 Conditional Branching: `if`
- **What it does:** Executes code block if the evaluated boolean predicate is True.
- **Syntax:** `if`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Applies Conditional Branching on fintech records using columns `transaction_amount`, `transaction_id` to demonstrate real-world execution.


```python
amt = float(transactions[0]['transaction_amount'])
if amt > 1000.0:
    print(f'Transaction {transactions[0]["transaction_id"]} is over $1,000!')
```

### 🔹 Conditional Branching: `elif`
- **What it does:** Evaluates alternative condition branch when prior `if` evaluates to False.
- **Syntax:** `elif`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Demonstrates Conditional Branching with practical fintech data structures and variables in the following code block.


```python
if amt > 1500.0:
    tier = 'Platinum'
elif amt > 500.0:
    tier = 'Gold'
print('Assigned Tier via elif:', tier)
```

### 🔹 Conditional Branching: `else`
- **What it does:** Default fallback executed when all preceding `if` and `elif` checks fail.
- **Syntax:** `else`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Demonstrates Conditional Branching with practical fintech data structures and variables in the following code block.


```python
if amt > 2000.0:
    category = 'Mega'
else:
    category = 'Regular'
print('Assigned Category via else:', category)
```

### 🔹 Ternary Conditional Expression: `val1 if cond else val2`
- **What it does:** Evaluates single-line inline conditional expression.
- **Syntax:** `val1 if cond else val2`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Applies Ternary Conditional Expression on fintech records using columns `is_fraud` to demonstrate real-world execution.


```python
status_label = 'FRAUD' if int(transactions[0]['is_fraud']) == 1 else 'CLEARED'
print('Ternary Label:', status_label)
```

### 🔹 Structural Pattern Matching: `match target:`
- **What it does:** Introduced in Python 3.10 for structural unpacking and multi-pattern matching.
- **Syntax:** `match target:`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Applies Structural Pattern Matching on fintech records using columns `card_type` to demonstrate real-world execution.


```python
tx_record = transactions[0]
match tx_record['card_type']:
    case 'Visa':
        print('Matched Visa processing flow')
    case 'MasterCard':
        print('Matched MasterCard processing flow')
    case _:
        print('Matched other card issuer')
```

### 🔹 Pattern Matching with Guard Clauses: `case pattern if guard:`
- **What it does:** Adds conditional guard logic to structural pattern cases.
- **Syntax:** `case pattern if guard:`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Applies Pattern Matching with Guard Clauses on fintech records using columns `card_type`, `transaction_amount` to demonstrate real-world execution.


```python
match tx_record:
    case {'card_type': 'Visa', 'transaction_amount': a} if float(a) > 500:
        print('Matched High-Value Visa with Guard Clause!')
    case _:
        print('Standard match')
```

### 🔹 Pattern Matching Wildcard: `case _:`
- **What it does:** Catch-all default case matching any unmatched pattern structure.
- **Syntax:** `case _:`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Demonstrates Pattern Matching Wildcard with practical fintech data structures and variables in the following code block.


```python
match 'UNKNOWN_CODE':
    case '200': print('OK')
    case _ : print('Wildcard default caught unhandled code')
```

### 🔹 Null Placeholder Statement: `pass`
- **What it does:** Syntactic null operation required where a statement block is syntactically required but no action is needed.
- **Syntax:** `pass`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Applies Null Placeholder Statement on fintech records using columns `is_fraud` to demonstrate real-world execution.


```python
for t in transactions[:3]:
    if int(t['is_fraud']) == 0:
        pass  # No-op placeholder
    else:
        print('Fraud found')
```

### 🔹 Reference Deletion: `del var_name`
- **What it does:** Unbinds variable name and decrements target object reference count in Python.
- **Syntax:** `del var_name`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Demonstrates Reference Deletion with practical fintech data structures and variables in the following code block.


```python
temp_token = 'SECRET_KEY_123'
print('Token defined:', temp_token)
del temp_token
print('Token deleted from namespace successfully.')
```

### 🔹 Collection Element Deletion: `del lst[0]` & `del d[key]`
- **What it does:** Deletes elements from mutable lists or dictionaries in-place.
- **Syntax:** `del lst[0]`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Applies Collection Element Deletion on fintech records using columns `transaction_id` to demonstrate real-world execution.


```python
mock_tx = transactions[0].copy()
print('Keys before del:', list(mock_tx.keys())[:3])
del mock_tx['transaction_id']
print('Keys after del transaction_id:', list(mock_tx.keys())[:3])
```

### 🔍 Scenario: Q1: Pattern Matching vs Dict Dispatch Table Performance
- **Objective:** Q1: Pattern Matching vs Dict Dispatch Table Performance
- **Approach:** Explain why dictionary dispatch tables (`dispatch.get(key, default)()`) have $O(1)$ time complexity while match/case statements provide structural type checking.

```python
dispatch = {'Visa': lambda: 'Fee 1.5%', 'Amex': lambda: 'Fee 2.5%'}
print('Dispatch table lookup:', dispatch.get('Visa', lambda: 'Fee 2.0%')())
```
