# Control Flow, Branching & Pattern Matching (5+ Years Interview Guide)
Exhaustive revision guide to if/elif/else branching, ternary expressions, structural pattern matching (match/case), pass, and reference deletion (del) on raw_transactions.csv.

### Key 5-Year Interview Concepts Covered:
- **Conditional Branching**: Dedicated cell for `if`, `elif`, and `else`.
- **Ternary Evaluation**: Dedicated cell for inline conditionals (`val1 if condition else val2`).
- **Structural Pattern Matching**: Dedicated cell for Python 3.10+ `match` and `case`.
- **Null & Deletion Statements**: Dedicated cell for `pass` and `del`.

This interactive revision guide uses `data/raw_transactions.csv` with individual dedicated cells per method.

```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
print('Sample Record:', transactions[0])
```

### Multi-Way Branching: `if`, `elif`, `else`
**Explanation**: Routes execution across mutually exclusive condition branches.

**Syntax**: `if cond1: ... elif cond2: ... else: ...`

```python
for t in transactions[:3]:
    amt = float(t['transaction_amount'])
    if amt > 1000.0:
        tier = 'VIP'
    elif amt > 250.0:
        tier = 'Standard'
    else:
        tier = 'Micro'
    print(f"{t['transaction_id']} (${amt:.2f}) -> {tier}")
```

### Ternary Conditional Expression (`val1 if cond else val2`)
**Explanation**: Evaluates an inline single-line conditional expression.

**Syntax**: `result = 'FRAUD' if is_fraud else 'LEGIT'`

```python
fraud_labels = [
    'FRAUD' if int(t['is_fraud']) == 1 else 'LEGIT'
    for t in transactions[:5]
]
print('Ternary Evaluated Labels:', fraud_labels)
```

### Structural Pattern Matching: `match` and `case` (Python 3.10+)
**Explanation**: Matches data against structural patterns, extracting components and binding variables with guard clauses (`if`).

**Syntax**: `match value: case pattern if guard: ...`

```python
def classify_card(tx_record):
    match tx_record:
        case {'card_type': 'Visa', 'is_fraud': '1'}:
            return 'Visa_Fraud_Alert'
        case {'card_type': 'Amex', 'transaction_amount': amt} if float(amt) > 1000:
            return 'Amex_High_Value'
        case {'transaction_status': 'Failed'}:
            return 'Transaction_Failed'
        case _:
            return 'Standard_Transaction'

for t in transactions[:4]:
    print(f"{t['transaction_id']} -> {classify_card(t)}")
```

### Null Statements (`pass`) & Reference Deletion (`del`)
**Explanation**: `pass` is a null statement placeholder required for empty code blocks. `del` destroys variable bindings and removes elements from mutable sequences/dictionaries in-place.

**Syntax**: `pass` / `del var_name` / `del dict[key]`

```python
mock_tx = transactions[0].copy()
print('Keys before del:', list(mock_tx.keys())[:3])
del mock_tx['device_type']
print('Keys after del device_type:', list(mock_tx.keys())[:3])
```

### Q1: Memory Deallocation and Garbage Collection with `del`
**Explanation**: Explain that `del x` does NOT immediately destroy the object in RAM; it decrements the object's reference count. Memory is freed only when refcount reaches 0.

**Syntax**: `sys.getrefcount(obj)`

```python
a = [1, 2, 3]
print('Reference count of a:', sys.getrefcount(a) - 1)
b = a
print('Reference count after b = a:', sys.getrefcount(a) - 1)
del b
print('Reference count after del b:', sys.getrefcount(a) - 1)
```
