## Section 3: Fintech Interview Questions

### Q1: Format transaction region and amounts using capitalizations and Decimal math on the first transaction.

````python
# Solution:
from decimal import Decimal
with open(csv_path, 'r') as f:
    f.readline()
    row = f.readline().strip().split(',')
    transaction_region = row[9].strip().upper()
    transaction_amount = Decimal(row[3]) if row[3] not in ('', 'NaN') else Decimal('0.0')
    print(f'Region: {transaction_region} | Payout: ${transaction_amount:.2f}')
````

### Q2: Extract the numeric ID digits from card_type value using search indices on the first transaction.

````python
# Solution:
with open(csv_path, 'r') as f:
    f.readline()
    row = f.readline().strip().split(',')
    card_provider_name = row[4]
    # Extract card details
    print('Card:', card_provider_name, 'Length:', len(card_provider_name))
````
