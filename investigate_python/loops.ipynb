### 1. Simple iterating sequences (for loop)
**Explanation**: Iterates over sequence arrays.

**Syntax**:
```python
for item in sequence:
    # block
```



````python
card_providers_list = ['Visa', 'MasterCard']
for provider in card_providers_list: print(provider)
````

### 2. Lazy ranges generation
**Explanation**: Generates numbers lazily with O(1) memory footprint.

**Syntax**:
```python
range(start, stop, step)
```

**Visual Explanation (Data with Baraa Style)**:
```mermaid
graph TD
    lazy[range Object] -->|yields next on request| index(loop_index)
```

````python
memory_range_object = range(1, 5)
print(list(memory_range_object))
````

### 3. Loop increments scanning (while loop)
**Explanation**: Iterates until condition is Falsy.

**Syntax**:
```python
while condition:
    # block
```



````python
iteration_counter = 0
while iteration_counter < 2:
    print(iteration_counter)
    iteration_counter += 1
````

### 4. Loop Control: Immediate exit (break)
**Explanation**: Exits active loop frame instantly.

**Syntax**:
```python
break
```



````python
for loop_index in range(5):
    if loop_index == 2: break
    print(loop_index)
````

### 5. Loop Control: Skipping iterations (continue)
**Explanation**: Skips current loop iteration step.

**Syntax**:
```python
continue
```



````python
for loop_index in range(3):
    if loop_index == 1: continue
    print(loop_index)
````

### 6. Loop Control: Pass placeholders
**Explanation**: Placeholder statements.

**Syntax**:
```python
pass
```



````python
for loop_index in range(2):
    pass
````

### 7. Normal completion loops checks (for-else)
**Explanation**: Executes else block only if loop completes without hitting break statements.

**Syntax**:
```python
for item in sequence:
    pass
else:
    # runs if no break
```

**Visual Explanation (Data with Baraa Style)**:
```mermaid
graph TD
    loop[for loop iterates] -->|no break hit| else_b(Execute else block)
    loop -->|break hit| skipped(Else block skipped)
```


````python
odd_numbers_sequence = [1, 3, 5]
for value in odd_numbers_sequence:
    if value % 2 == 0: break
else: print('No break occurred')
````

### 8. While loop-else block execution
**Explanation**: While loops else block execution.

**Syntax**:
```python
while condition:
    pass
else:
    # block
```



````python
iteration_counter = 0
while iteration_counter < 2:
    iteration_counter += 1
else: print('While else hit')
````

### 9. Index enumeration tracking
**Explanation**: Tracks iteration indexes.

**Syntax**:
```python
for index, value in enumerate(sequence):
```



````python
for index, provider in enumerate(['Visa']): print(index, provider)
````

### 10. Zip parallel sequences iterating
**Explanation**: Iterates sequences in parallel.

**Syntax**:
```python
zip(collection_one, collection_two)
```



````python
for index_value, name_value in zip([1], ['Visa']): print(index_value, name_value)
````

### 11. Nested loops structure
**Explanation**: Nest loops to process multi-dimensional data.

**Syntax**:
```python
for outer_item in outer_list:
    for inner_item in inner_list:
```



````python
for outer_index in [1]:
    for inner_index in ['Visa']: print(outer_index, inner_index)
````

### 12. Iterating dictionaries in loops
**Explanation**: Iterate keys and values.

**Syntax**:
```python
for key, value in dictionary_object.items():
```



````python
simple_status_dict = {'TX101': 'Completed'}
for tx_id in simple_status_dict: print(tx_id)
````

### 13. Unpacking tuple structures in loops
**Explanation**: Unpack nested structures.

**Syntax**:
```python
for first_field, second_field in sequence_of_tuples:
```



````python
for tx_id, amount in [('TX101', 120.0)]: print(tx_id, amount)
````

### 14. Infinite loop scenarios
**Explanation**: Loops run forever unless broken.

**Syntax**:
```python
while True:
    if condition: break
```



````python
infinite_counter = 0
while True:
    infinite_counter += 1
    if infinite_counter >= 2: break
print(infinite_counter)
````

### 15. Modifying collections during iteration
**Explanation**: Modifying collections during iterations leads to errors; copy list first.

**Syntax**:
```python
for item in list_object.copy():
```



````python
original_payout_list = [100.0, 200.0]
for amount in original_payout_list.copy():
    original_payout_list.remove(amount)
print('Modified safely:', original_payout_list)
````

## Section 3: Fintech Interview Questions

### Q1: Extract regions for the first 10 transactions using enumerate, skipping 'Failed' status entries.

````python
# Solution:
with open(csv_path, 'r') as f:
    f.readline()
    for i in range(10):
        row = f.readline().strip().split(',')
        if row[5] == 'Failed': continue
        print(f'Idx: {i} | Tx: {row[0]} | Region: {row[9]}')
````

### Q2: Search for transaction regions in the first 20 entries, using loop-else to print if no matching transaction > $1900 exists.

````python
# Solution:
with open(csv_path, 'r') as f:
    f.readline()
    has_high_value_transaction = False
    for _ in range(20):
        row = f.readline().strip().split(',')
        transaction_amount = float(row[3]) if row[3] not in ('', 'NaN') else 0.0
        if transaction_amount > 1900.0:
            print('Match found:', row[0])
            has_high_value_transaction = True
            break
    if not has_high_value_transaction:
        print('Search complete: No transaction > $1900 found.')
````
