### 1. Arithmetic Operators Basics
**Explanation**: Performs mathematical calculations.

**Syntax**:
```python
operand_one + operand_two
operand_one - operand_two
```



````python
base_amount, additional_fee = 10, 5
print('Sum:', base_amount + additional_fee)
````

### 2. Floor Division
**Explanation**: Divides numbers and rounds down to the nearest integer.

**Syntax**:
```python
operand_one // operand_two
```



````python
total_items, batch_size = 15, 4
print('15 // 4 is:', total_items // batch_size)
````

### 3. Modulo Division Remainders
**Explanation**: Calculates division remainders.

**Syntax**:
```python
operand_one % operand_two
```



````python
total_items, batch_size = 15, 4
print('15 % 4 remainder is:', total_items % batch_size)
````

### 4. Exponentiation
**Explanation**: Powers base value by exponents.

**Syntax**:
```python
base_value ** exponent_value
```



````python
base_multiplier, power_exponent = 2, 10
print('2 to power 10:', base_multiplier ** power_exponent)
````

### 5. Comparison Operators
**Explanation**: Exposes relational checks.

**Syntax**:
```python
value_one == value_two
value_one != value_two
```



````python
transaction_amount = 50.0
print('Is amount > 30.0?:', transaction_amount > 30.0)
````

### 6. Logical Operators
**Explanation**: Binds multiple boolean checks.

**Syntax**:
```python
condition_one and condition_two
```



````python
logical_condition_one, logical_condition_two = True, False
print('True and False:', logical_condition_one and logical_condition_two)
````

### 7. Logical Short-Circuiting
**Explanation**: Evaluates expressions left-to-right, short-circuiting if right-side is irrelevant.

**Syntax**:
```python
expression_one or expression_two
```

**Visual Explanation (Data with Baraa Style)**:
```mermaid
graph TD
    eval[Evaluate: True or Print]
    eval -->|left is True, short circuits| return(True)
    eval -.->|never executed| print(print function call)
```


````python
validation_flag = True or print('This won\'t print')
print('Flag:', validation_flag)
````

### 8. Bitwise operations setting flags
**Explanation**: Bitwise OR sets flag states.

**Syntax**:
```python
combined_mask = flag_one | flag_two
```



````python
read_permission_bit = 1 << 0
print('Flag mask read:', bin(read_permission_bit))
````

### 9. Bitwise checking flags status
**Explanation**: Bitwise AND evaluates flag status.

**Syntax**:
```python
has_flag = value & flag_mask
```

**Visual Explanation (Data with Baraa Style)**:
```mermaid
graph TD
    mask[user_permissions_mask: 0b11] -->|AND check Bit 1| res(0b10 - True)
```


````python
user_permissions_mask = 3
write_permission_bit = 1 << 1
print('Has write flag?:', bool(user_permissions_mask & write_permission_bit))
````

### 10. Bitwise shifting
**Explanation**: Shifts integer bits left or right.

**Syntax**:
```python
integer_value << shift_count
integer_value >> shift_count
```



````python
bit_base_value = 2
print('2 shifted left by 3:', bit_base_value << 3)
````

### 11. Bitwise XOR and NOT
**Explanation**: Logical XOR evaluates parity; NOT flips bits.

**Syntax**:
```python
value_one ^ value_two
~value
```



````python
bitwise_operand_one, bitwise_operand_two = 12, 5
print('Bitwise XOR:', bitwise_operand_one ^ bitwise_operand_two, 'NOT:', ~0)
````

### 12. Membership checks
**Explanation**: Checks if item exists inside collection.

**Syntax**:
```python
item_value in collection_object
```



````python
target_region = 'North'
allowed_regions_list = ['North', 'South']
print('Is region allowed?:', target_region in allowed_regions_list)
````

### 13. Identity matching references
**Explanation**: Checks reference address match.

**Syntax**:
```python
object_one is object_two
```



````python
null_reference_one, null_reference_two = None, None
print('None is None?:', null_reference_one is null_reference_two)
````

### 14. Assignment operators
**Explanation**: Performs arithmetic updates in-place.

**Syntax**:
```python
variable_name += value
```



````python
running_sum_variable = 10
running_sum_variable += 5
print('running_sum_variable:', running_sum_variable)
````

### 15. Operators precedence and groupings
**Explanation**: Parentheses override operator priority rankings.

**Syntax**:
```python
(expression_one) * expression_two
```



````python
precedence_calculation = (2 + 3) * 4
print('Value:', precedence_calculation)
````

## Section 3: Fintech Interview Questions

### Q1: Validate if transaction amount is between $100 and $1000 and region is 'North' using logical operations.

````python
# Solution:
with open(csv_path, 'r') as f:
    f.readline()
    row = f.readline().strip().split(',')
    transaction_amount = float(row[3]) if row[3] not in ('', 'NaN') else 0.0
    transaction_region = row[9].strip()
    is_matching_criteria = 100.0 <= transaction_amount <= 1000.0 and transaction_region == 'North'
    print('Matching criteria?:', is_matching_criteria)
````

### Q2: Perform bitwise validation storing high amount (> $1500) at Bit 0 and fraud status at Bit 1.

````python
# Solution:
with open(csv_path, 'r') as f:
    f.readline()
    row = f.readline().strip().split(',')
    transaction_amount = float(row[3]) if row[3] not in ('', 'NaN') else 0.0
    is_fraudulent_flag = int(row[10])
    bitwise_status_mask = 0
    if transaction_amount > 1500.0: bitwise_status_mask |= 1
    if is_fraudulent_flag == 1: bitwise_status_mask |= 2
    print('Flags mask:', bin(bitwise_status_mask))
````
