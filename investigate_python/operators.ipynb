### 1. Arithmetic Operators Basics
**Explanation**: Python supports standard arithmetic operations: addition (`+`), subtraction (`-`), multiplication (`*`), and true division (`/`). In Python 3, true division `/` always returns a `float` even if the division is exact (e.g. `4 / 2` evaluates to `2.0`), guaranteeing consistent type signatures across operations.

**Syntax**: `a + b` / `a - b` / `a * b` / `a / b`

```python
base_amount, additional_fee = 10, 5
print('Sum:', base_amount + additional_fee)
```

### 2. Floor Division (`//`)
**Explanation**: The floor division operator `//` divides two numbers and rounds down to the nearest mathematical integer (towards negative infinity, floor operation). For positive numbers, `7 // 2` evaluates to `3`. For negative numbers, `-7 // 2` evaluates to `-4` (not `-3`), which is a frequent gotcha in algorithms and interview questions.

**Syntax**: `quotient = dividend // divisor`

```python
total_items, batch_size = 15, 4
print('15 // 4 is:', total_items // batch_size)
```

### 3. Modulo Division Remainders (`%`)
**Explanation**: The modulo operator `%` calculates the remainder of division satisfying the invariant `a = (a // b) * b + (a % b)`. Because Python uses floor division (rounding towards negative infinity), the result of `a % b` always shares the same sign as the divisor `b`. For instance, `-7 % 3` evaluates to `2` (since `(-7 // 3) * 3 + 2 = -9 + 2 = -7`).

**Syntax**: `remainder = dividend % divisor`

```python
total_items, batch_size = 15, 4
print('15 % 4 remainder is:', total_items % batch_size)
```

### 4. Exponentiation (`**` and `pow()`)
**Explanation**: The exponentiation operator `**` raises a base to a power. The built-in `pow(base, exp, mod)` function is heavily used in cryptography and modular arithmetic because `pow(base, exp, mod)` computes `(base ** exp) % mod` using fast modular exponentiation in O(log exp) time without allocating massive intermediate integers.

**Syntax**: `base ** exp` / `pow(base, exp, mod)`

```python
base_multiplier, power_exponent = 2, 10
print('2 to power 10:', base_multiplier ** power_exponent)
```

### 5. Comparison Operators & Chaining
**Explanation**: Python provides comparison operators: `==`, `!=`, `<`, `<=`, `>`, `>=`. Python uniquely supports comparison chaining such as `10 < x < 50`, which evaluates as `(10 < x) and (x < 50)` but guarantees that the middle expression `x` is evaluated only once, eliminating redundant computations and side effects.

**Syntax**: `min_val <= x <= max_val`

```python
transaction_amount = 50.0
print('Is amount > 30.0?:', transaction_amount > 30.0)
```

### 6. Logical Operators (`and`, `or`, `not`)
**Explanation**: Logical operators evaluate truthiness according to boolean logic. `not` always returns a strict `bool` (`True` or `False`). However, `and` and `or` return the actual operand object that determined the outcome, enabling clean fallback patterns like `display_name = user_name or 'Anonymous'`.

**Syntax**: `x and y` / `x or y` / `not x`

```python
logical_condition_one, logical_condition_two = True, False
print('True and False:', logical_condition_one and logical_condition_two)
```

### 7. Logical Short-Circuiting
**Explanation**: Python evaluates `and` and `or` from left to right and stops evaluation as soon as the outcome is determined. In `a and b`, if `a` is falsy, Python returns `a` immediately without evaluating `b`. In `a or b`, if `a` is truthy, Python returns `a` immediately. This is vital in production for guarding against `NoneType` errors: `if user and user.is_active:` never calls `.is_active` on `None`.

**Syntax**: `if obj is not None and obj.is_valid(): pass`

```python
validation_flag = True or print('This won\'t print')
print('Flag:', validation_flag)
```

### 8. Bitwise Operations: Setting Flags (`|`)
**Explanation**: Bitwise operations operate on the individual binary bits of integer operands. The bitwise OR operator `|` is used in permission systems and low-level protocols to combine or set specific bit flags (e.g. `flags = READ_PERM | WRITE_PERM`). Each bit represents an independent boolean state.

**Syntax**: `flags = current_flags | FLAG_BIT`

```python
read_permission_bit = 1 << 0
print('Flag mask read:', bin(read_permission_bit))
```

### 9. Bitwise Operations: Checking & Clearing Flags (`&` and `~`)
**Explanation**: The bitwise AND operator `&` isolates specific bits to check if a flag is active (`if flags & FLAG_BIT:`). Combined with the bitwise NOT operator `~` (inversion), it allows clearing specific flags (`flags = flags & ~FLAG_BIT`) without altering any other bit states in the bitmask.

**Syntax**: `is_set = bool(flags & FLAG_BIT)` / `flags &= ~FLAG_BIT`

```python
user_permissions_mask = 3
write_permission_bit = 1 << 1
print('Has write flag?:', bool(user_permissions_mask & write_permission_bit))
```

### 10. Bitwise Shifts (`<<` and `>>`)
**Explanation**: Left shift `<<` multiplies an integer by powers of 2 (`x << n` == `x * (2**n)`), and right shift `>>` performs floor division by powers of 2 (`x >> n` == `x // (2**n)`). Bit shifting is used in hashing algorithms, binary encoding, and performance-critical DSP math because shift operations execute in a single CPU cycle.

**Syntax**: `val << shift_amount` / `val >> shift_amount`

```python
bit_base_value = 2
print('2 shifted left by 3:', bit_base_value << 3)
```

### 11. Bitwise XOR (`^`)
**Explanation**: The bitwise XOR operator `^` returns 1 only where corresponding bits differ. In algorithmic interviews, XOR is famous for solving the 'Single Number' problem in O(N) time and O(1) space because `x ^ x == 0` and `x ^ 0 == x`, allowing duplicates to cancel each other out.

**Syntax**: `diff_bits = a ^ b`

```python
bitwise_operand_one, bitwise_operand_two = 12, 5
print('Bitwise XOR:', bitwise_operand_one ^ bitwise_operand_two, 'NOT:', ~0)
```

### 12. Membership Operators (`in`, `not in`)
**Explanation**: The `in` operator tests whether an item exists in a sequence or collection. The performance depends heavily on the data structure: O(N) linear scan for lists, strings, and tuples, but O(1) constant time average lookup for sets and dictionaries using hash lookups. Knowing this difference is a staple 5-year interview question.

**Syntax**: `item in collection` / `item not in collection`

```python
target_region = 'North'
allowed_regions_list = ['North', 'South']
print('Is region allowed?:', target_region in allowed_regions_list)
```

### 13. Identity Matching (`is` vs `==`)
**Explanation**: `is` evaluates whether two variables point to the exact same object in memory by comparing their memory addresses (`id(a) == id(b)`). In contrast, `==` invokes `__eq__` to compare values. Never use `is` to compare string or number values; reserve `is` exclusively for singleton comparisons like `is None` and `is not None`.

**Syntax**: `obj is None` / `obj is not None`

```python
null_reference_one, null_reference_two = None, None
print('None is None?:', null_reference_one is null_reference_two)
```

### 14. Augmented Assignment Operators (`+=`, `-=`, `*=`)
**Explanation**: Augmented assignment operators evaluate in-place mutation where supported by the object type via dunder methods like `__iadd__`. For mutable objects like lists, `list_a += [1]` modifies the list in-place (equivalent to `.extend()`). For immutable objects like integers and strings, `x += 1` creates a new object and re-binds `x`.

**Syntax**: `x += delta` / `items += [new_item]`

```python
running_sum_variable = 10
running_sum_variable += 5
print('running_sum_variable:', running_sum_variable)
```

### 15. Operator Precedence & Grouping
**Explanation**: Python evaluates expressions following strict operator precedence: Parentheses `()` > Exponentiation `**` > Bitwise NOT `~` > Multiplication/Division `* / // %` > Addition/Subtraction `+ -` > Bitwise Shifts `<< >>` > Bitwise AND `&` > Bitwise XOR `^` > Bitwise OR `|` > Comparisons `== != < > is in` > Logical NOT `not` > Logical AND `and` > Logical OR `or`. Use explicit parentheses for clarity in production code.

**Syntax**: `result = (a + b) * (c or d)`

```python
precedence_calculation = (2 + 3) * 4
print('Value:', precedence_calculation)
```

### Q1: Multi-Attribute Range & Status Validation
**Explanation**: **Scenario**: Validate transactions meeting complex criteria (amount between $100 and $1,000, region equals 'North', and status is not 'Failed') using chained comparisons and short-circuit evaluation.

**Syntax**: `100.0 <= amount <= 1000.0 and region == 'North'`
