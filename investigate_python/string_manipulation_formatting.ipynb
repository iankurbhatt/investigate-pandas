```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
```

### 🔹 String Indexing & Slicing
- **What it does:** Extracts characters and sub-strings via 0-based coordinate indexing and step slicing.
- **Syntax:** `s[start:stop:step]`
- **Operation:** `tx_id = transactions[0]['transaction_id']`
- **Key Note:** Remember: Basic slicing creates a *view* into the original array. Modifying a view changes the original array! Use `.copy()` when you need an isolated duplicate.

```python
tx_id = transactions[0]['transaction_id']
print('First 2 characters:', tx_id[:2])
print('Reversed ID:', tx_id[::-1])
```

### 🔹 Searching: `.find()`
- **What it does:** Returns lowest index of substring, or -1 if not found (does NOT raise error).
- **Syntax:** `s.find(sub)`
- **Operation:** `print('Find "TX" index:', tx_id.find('TX'))`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
print('Find "TX" index:', tx_id.find('TX'))
print('Find "XYZ" (missing):', tx_id.find('XYZ'))
```

### 🔹 Line Splitting: `.splitlines()`

### 🔹 Joining Sequences: `str.join(iterable)`
- **What it does:** Joins elements of an iterable into a single string using separator string in $O(N)$ time.
- **Syntax:** `delimiter.join(iterable)`
- **Operation:** `card_list = ['Visa', 'MasterCard', 'Amex']`
- **Key Note:** Always check `df.shape` before and after merging to verify you didn't accidentally introduce Cartesian duplication or drop rows.

```python
card_list = ['Visa', 'MasterCard', 'Amex']
print('Joined Cards:', ' | '.join(card_list))
```

### 🔹 Verification: `.startswith()` & `.endswith()`
- **What it does:** Tests string prefix and suffix matches.
- **Syntax:** `s.startswith(prefix)`
- **Operation:** `print('Starts with TX?:', tx_id.startswith('TX'))`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
print('Starts with TX?:', tx_id.startswith('TX'))
print('Ends with digits?:', tx_id[-1].isdigit())
```

### 🔹 Verification: `.isalnum()` & `.isdigit()`
- **What it does:** Validates character classes (alphanumeric, numeric digits).
- **Syntax:** `s.isalnum()`
- **Operation:** `print('Is alphanumeric (TX1001)?:', 'TX1001'.isalnum())`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
print('Is alphanumeric (TX1001)?:', 'TX1001'.isalnum())
print('Is pure digits (1001)?:', '1001'.isdigit())
```

### 🔹 Formatting with F-Strings: `f"{var}"`
- **What it does:** Formatted string interpolation with precision, padding, and alignment specifiers.
- **Syntax:** `f"{val:,.2f}"`
- **Operation:** `amt = float(transactions[0]['transaction_amount'])`
- **Key Note:** Pandas `.str` accessor methods automatically skip missing (`NaN`) values instead of raising an `AttributeError`.

```python
amt = float(transactions[0]['transaction_amount'])
print(f'Formatted currency: ${amt:,.2f}')
print(f'Padded hex representation: {255:#04x}')
```

### 🔹 Formatting with `.format()`
- **What it does:** Positional and keyword string formatting template syntax.
- **Syntax:** `"Tx: {}".format(tx_id)`
- **Operation:** `print('Formatted via .format(): TxID={0}, Amount=${1:.2f}'.format(tx_id, amt))`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
print('Formatted via .format(): TxID={0}, Amount=${1:.2f}'.format(tx_id, amt))
```

### 🔍 Scenario: Q1: Memory Buffer Allocation in `str.join` vs Repeated `+=`
- **Objective:** Q1: Memory Buffer Allocation in `str.join` vs Repeated `+=`
- **Approach:** Demonstrate how `str.join()` computes exact final length and allocates memory once in $O(N)$ time.
- **Syntax:** `''.join(seq)`

```python
words = [t['transaction_id'] for t in transactions[:500]]
res = ','.join(words)
print('Joined 500 transaction IDs. String length:', len(res))
```
