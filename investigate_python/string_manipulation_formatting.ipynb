This interactive revision guide uses `data/raw_transactions.csv` with individual dedicated cells per method.

```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
```

### String Immutability & Slicing
**Explanation**: Strings are immutable sequences of Unicode characters. Slicing `s[start:stop:step]` returns new string objects.

**Syntax**: `s[0:4]` / `s[::-1]`

```python
tx_id = transactions[0]['transaction_id']
print('Transaction ID:', tx_id)
print('Prefix (first 2 chars):', tx_id[:2])
print('Numeric suffix:', tx_id[2:])
print('Reversed String:', tx_id[::-1])
```

### Case Transformation & Trimming
**Explanation**: Standardizes string casing and removes whitespace padding.

**Syntax**: `s.lower()` / `s.upper()` / `s.strip()`

```python
raw_region = '   north america   '
print('Trimmed & Uppercase:', raw_region.strip().upper())
print('Title Case:', raw_region.strip().title())
```

### Searching & Replacing: `find`, `count`, `replace`
**Explanation**: Locates substrings and substitutes text patterns.

**Syntax**: `s.find(sub)` / `s.count(sub)` / `s.replace(old, new)`

```python
date_str = transactions[0]['transaction_date']
print('Original Date:', date_str)
print('Replaced separators:', date_str.replace('-', '/'))
print('Count of colons:', date_str.count(':'))
```

### Splitting & Joining: `split` and `str.join()`
**Explanation**: Splits delimited strings into lists and joins sequences with custom delimiters in $O(N)$ time.

**Syntax**: `s.split(sep)` / `delimiter.join(iterable)`

```python
sample_ids = [t['transaction_id'] for t in transactions[:4]]
joined_csv_line = ','.join(sample_ids)
print('Joined CSV string:', joined_csv_line)
print('Split back into list:', joined_csv_line.split(','))
```

### String Verification & F-String Formatting
**Explanation**: Validates string prefixes/character classes and formats numeric values using f-string specifiers.

**Syntax**: `f"{val:,.2f}"` / `s.startswith(prefix)` / `s.isalnum()`

```python
amt = float(transactions[0]['transaction_amount'])
print(f"Formatted Amount: ${amt:,.2f}")
print(f"Left-aligned padded: {'TX_ALERT':<15} Status: {'APPROVED':>10}")
print("Is alphanumeric:", tx_id.isalnum())
```

### Q1: Fast String Concatenation: `str.join` vs Repeated `+=`
**Explanation**: Explain why repeated string concatenation with `+=` has $O(N^2)$ time complexity due to immutability, while `str.join()` allocates memory once in $O(N)$ time.

**Syntax**: `''.join(list_of_strings)` ($O(N)$)

```python
words = [t['transaction_id'] for t in transactions[:1000]]
t0 = time.perf_counter()
res_join = ','.join(words)
t_join = time.perf_counter() - t0
print(f'str.join() on 1000 strings: {t_join*1000:.3f} ms')
```
