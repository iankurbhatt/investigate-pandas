This interactive revision guide uses `data/raw_transactions.csv` with 1-to-1 atomic method cells.

```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
```

### String Indexing & Slicing
**Explanation**: Extracts characters and sub-strings via 0-based coordinate indexing and step slicing.

**Syntax**: `s[start:stop:step]`

```python
tx_id = transactions[0]['transaction_id']
print('First 2 characters:', tx_id[:2])
print('Reversed ID:', tx_id[::-1])
```

### Case Transformation: `.lower()`
**Explanation**: Converts all characters to lowercase.

**Syntax**: `s.lower()`

```python
print('Lowercased card:', transactions[0]['card_type'].lower())
```

### Case Transformation: `.upper()`
**Explanation**: Converts all characters to uppercase.

**Syntax**: `s.upper()`

```python
print('Uppercased region:', transactions[0]['region'].upper())
```

### Case Transformation: `.title()`
**Explanation**: Capitalizes the first character of each word.

**Syntax**: `s.title()`

```python
print('Title cased:', 'north america financial'.title())
```

### Case Transformation: `.capitalize()`
**Explanation**: Capitalizes ONLY the very first character of the entire string.

**Syntax**: `s.capitalize()`

```python
print('Capitalized:', 'visa debit card'.capitalize())
```

### Trimming: `.strip()`
**Explanation**: Removes leading and trailing whitespace padding.

**Syntax**: `s.strip()`

```python
print('Stripped:', repr('   North America   '.strip()))
```

### Trimming: `.lstrip()` & `.rstrip()`
**Explanation**: `.lstrip()` removes leading whitespace; `.rstrip()` removes trailing whitespace.

**Syntax**: `s.lstrip()` / `s.rstrip()`

```python
print('Lstripped:', repr('   North'.lstrip()))
print('Rstripped:', repr('North   '.rstrip()))
```

### Text Padding: `.center()`
**Explanation**: Centers string within specified width padded with characters.

**Syntax**: `s.center(width, fillchar)`

```python
print('Centered Header:', 'RECEIPT'.center(30, '='))
```

### Searching: `.find()`
**Explanation**: Returns lowest index of substring, or -1 if not found (does NOT raise error).

**Syntax**: `s.find(sub)`

```python
print('Find "TX" index:', tx_id.find('TX'))
print('Find "XYZ" (missing):', tx_id.find('XYZ'))
```

### Searching: `.index()`
**Explanation**: Returns lowest index of substring, but raises ValueError if substring is absent.

**Syntax**: `s.index(sub)`

```python
print('Index of "TX":', tx_id.index('TX'))
```

### Occurrences: `.count()`
**Explanation**: Returns total non-overlapping occurrences of substring.

**Syntax**: `s.count(sub)`

```python
date_str = transactions[0]['transaction_date']
print('Count of hyphens in date:', date_str.count('-'))
```

### Substitutions: `.replace()`
**Explanation**: Replaces occurrences of substring with replacement string.

**Syntax**: `s.replace(old, new)`

```python
print('Replaced separators:', date_str.replace('-', '/'))
```

### Splitting: `.split()`
**Explanation**: Splits string into a list of substrings along specified delimiter.

**Syntax**: `s.split(sep)`

```python
print('Split date parts:', date_str.split('-'))
```

### Line Splitting: `.splitlines()`
**Explanation**: Splits multi-line text into a list of lines breaking at universal newlines.

**Syntax**: `s.splitlines()`

```python
multi_line_feed = 'TX1,100\nTX2,200\nTX3,300'
print('Split lines:', multi_line_feed.splitlines())
```

### Joining Sequences: `str.join(iterable)`
**Explanation**: Joins elements of an iterable into a single string using separator string in $O(N)$ time.

**Syntax**: `delimiter.join(iterable)`

```python
card_list = ['Visa', 'MasterCard', 'Amex']
print('Joined Cards:', ' | '.join(card_list))
```

### Verification: `.startswith()` & `.endswith()`
**Explanation**: Tests string prefix and suffix matches.

**Syntax**: `s.startswith(prefix)` / `s.endswith(suffix)`

```python
print('Starts with TX?:', tx_id.startswith('TX'))
print('Ends with digits?:', tx_id[-1].isdigit())
```

### Verification: `.isalnum()` & `.isdigit()`
**Explanation**: Validates character classes (alphanumeric, numeric digits).

**Syntax**: `s.isalnum()` / `s.isdigit()`

```python
print('Is alphanumeric (TX1001)?:', 'TX1001'.isalnum())
print('Is pure digits (1001)?:', '1001'.isdigit())
```

### Formatting with F-Strings: `f"{var}"`
**Explanation**: Formatted string interpolation with precision, padding, and alignment specifiers.

**Syntax**: `f"{val:,.2f}"`

```python
amt = float(transactions[0]['transaction_amount'])
print(f'Formatted currency: ${amt:,.2f}')
print(f'Padded hex representation: {255:#04x}')
```

### Formatting with `.format()`
**Explanation**: Positional and keyword string formatting template syntax.

**Syntax**: `"Tx: {}".format(tx_id)`

```python
print('Formatted via .format(): TxID={0}, Amount=${1:.2f}'.format(tx_id, amt))
```

### Q1: Memory Buffer Allocation in `str.join` vs Repeated `+=`
**Explanation**: Demonstrate how `str.join()` computes exact final length and allocates memory once in $O(N)$ time.

**Syntax**: `''.join(seq)`

```python
words = [t['transaction_id'] for t in transactions[:500]]
res = ','.join(words)
print('Joined 500 transaction IDs. String length:', len(res))
```
