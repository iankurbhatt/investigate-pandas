# String Manipulation & Text Formatting: Beginner Guide

### 🌟 What is String Manipulation & Text Formatting in Python?
Strings represent text in Python. Because strings are immutable (cannot be modified in-place), Python provides an extensive set of methods to slice, search, split, join, sanitize, and format text dynamically using modern f-strings (`f'{var}'`).

This interactive guide loads and works directly with `data/raw_transactions.csv`, giving you real-world hands-on practice.

### 📚 Key Concepts Covered in this Notebook:
- **Indexing & Slicing**: Covers character coordinates extraction.
- **Case Transformation**: Covers `.lower()`, `.upper()`, `.title()`, and `.capitalize()`.
- **Trimming & Padding**: Covers `.strip()`, `.lstrip()`, `.rstrip()`, and `.center()`.
- **Searching & Replacing**: Covers `.find()`, `.index()`, `.count()`, and `.replace()`.
- **Splitting & Joining**: Covers `.split()`, `.splitlines()`, and `str.join()`.
- **Verification**: Covers `.startswith()`, `.endswith()`, `.isalnum()`, and `.isdigit()`.
- **Formatting**: Covers f-strings and `.format()`.


```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
```

### 🔹 String Indexing & Slicing
Extracts characters and sub-strings via 0-based coordinate indexing and step slicing. It provides a robust, standardized way to process, clean, and analyze datasets reliably and efficiently. **Tip:** Remember: Basic slicing creates a *view* into the original array. Modifying a view changes the original array! Use `.copy()` when you need an isolated duplicate.

**Syntax:** `s[start:stop:step]`


```python
tx_id = transactions[0]['transaction_id']
print('First 2 characters:', tx_id[:2])
print('Reversed ID:', tx_id[::-1])
```

### 🔹 Formatting with F-Strings: `f"{var}"`
Formatted string interpolation with precision, padding, and alignment specifiers. It provides a robust, standardized way to process, clean, and analyze datasets reliably and efficiently. **Tip:** Pandas `.str` accessor methods automatically skip missing (`NaN`) values instead of raising an `AttributeError`.

**Syntax:** `f"{val:,.2f}"`


```python
amt = float(transactions[0]['transaction_amount'])
print(f'Formatted currency: ${amt:,.2f}')
print(f'Padded hex representation: {255:#04x}')
```

**Syntax:** `"Tx: {}".format(tx_id)`


```python
print('Formatted via .format(): TxID={0}, Amount=${1:.2f}'.format(tx_id, amt))
```

### 🔍 Scenario: Q1: Memory Buffer Allocation in `str.join` vs Repeated `+=`

**Approach:** Demonstrate how `str.join()` computes exact final length and allocates memory once in $O(N)$ time.
**Syntax:** `''.join(seq)`


```python
words = [t['transaction_id'] for t in transactions[:500]]
res = ','.join(words)
print('Joined 500 transaction IDs. String length:', len(res))
```
