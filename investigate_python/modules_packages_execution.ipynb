# Modules, Packages & Dynamic Execution: Beginner Guide

### 🌟 What Are Modules, Packages & Dynamic Execution?
As codebases grow, organizing code into separate **Modules** (`.py` files) and **Packages** (folders with `__init__.py`) keeps code modular. Understanding `sys.path`, import mechanisms, and execution guards (`if __name__ == '__main__':`) ensures your code can be imported as a library or run as a standalone script.

This interactive guide loads and works directly with `data/raw_transactions.csv`, giving you real-world hands-on practice.

### 📚 Key Concepts Covered in this Notebook:
- **Importing Variations**: Covers `import mod`, `from mod import name`, and `import mod as alias`.
- **Module Resolution**: Covers `sys.path` search order.
- **Execution Entrypoint**: Covers `if __name__ == '__main__':`.
- **Dynamic Evaluation**: Covers `eval()` and `exec()`.


```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os
import json
import re
import collections
from datetime import datetime, timedelta

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
```

### 🔹 Dynamic Expression Evaluation: `eval()`
Parses and evaluates a single Python expression string returning the computed result. Saving and loading data properly ensures results are persistent, shareable across teams, and ready for downstream analysis. **Tip:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

**Syntax:** `eval('x + y', {'x': 1, 'y': 2})`


```python
amt = float(transactions[0]['transaction_amount'])
res = eval('amt * 1.05 + 0.30', {}, {'amt': amt})
print(f'Dynamically evaluated fee: ${res:.2f}')
```

### 🔹 Dynamic Code Execution: `exec()`
Executes arbitrary multi-line Python code blocks dynamically. Saving and loading data properly ensures results are persistent, shareable across teams, and ready for downstream analysis. **Tip:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

**Syntax:** `exec('statement')`


```python
code_block = """
risk_flag = 'HIGH_RISK' if amt > 500 else 'LOW_RISK'
"""
locs = {'amt': amt}
exec(code_block, {}, locs)
print('Dynamic exec result:', locs['risk_flag'])
```
