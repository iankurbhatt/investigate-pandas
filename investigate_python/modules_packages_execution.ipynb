### 📌 Overview & Architectural Context
Welcome to **19. Modules, Packages & Script Execution Lifecycles**. Python modules are singletons loaded and cached in `sys.modules`. This notebook covers module imports, `__name__ == '__main__'` entrypoint guards, package namespace initialization (`__init__.py`, `__all__`), relative vs absolute imports, and dynamic module loading with `importlib`.

### 📚 Key Concepts Covered in this Notebook:
- [x] 🔹 Module Import: `import mod`
- [x] 🔹 Specific Import: `from mod import name`
- [x] 🔹 Aliased Import: `import mod as alias`
- [x] 🔹 Module Resolution: `sys.path`
- [x] 🔹 Execution Guard: `if __name__ == '__main__':`
- [x] 🔹 Dynamic Expression Evaluation: `eval()`
- [x] 🔹 Dynamic Code Execution: `exec()`


```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os
import json
import re
import collections
from datetime import datetime, timedelta

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
```

### 🔹 Module Import: `import mod`
- **What it does:** Loads a module into `sys.modules` and binds module object to local namespace.
- **Syntax:** `import mod`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Demonstrates Module Import with practical fintech data structures and variables in the following code block.


```python
import math
print('math.pi:', math.pi)
```

### 🔹 Specific Import: `from mod import name`
- **What it does:** Imports specific attributes/functions directly into the local namespace.
- **Syntax:** `from mod import name`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Demonstrates Specific Import with practical fintech data structures and variables in the following code block.


```python
from math import sqrt
print('sqrt(16):', sqrt(16))
```

### 🔹 Aliased Import: `import mod as alias`
- **What it does:** Imports and renames module to prevent local namespace collisions.
- **Syntax:** `import mod as alias`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Demonstrates Aliased Import with practical fintech data structures and variables in the following code block.


```python
import collections as col
print('Aliased module col:', col.__name__)
```

### 🔹 Execution Guard: `if __name__ == '__main__':`
- **What it does:** Ensures code runs only when executed directly from CLI, not when imported.
- **Syntax:** `if __name__ == '__main__':`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Demonstrates Execution Guard with practical fintech data structures and variables in the following code block.


```python
print(f'Current execution context __name__: {__name__}')
```

### 🔹 Dynamic Expression Evaluation: `eval()`
- **What it does:** Parses and evaluates a single Python expression string returning the computed result.
- **Syntax:** `eval()`
  - **Parameters:**
    - `row_label` (*hashable*): Row label.
    - `col_label` (*hashable*): Column label.
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Applies Dynamic Expression Evaluation on fintech records using columns `transaction_amount` to demonstrate real-world execution.


```python
amt = float(transactions[0]['transaction_amount'])
res = eval('amt * 1.05 + 0.30', {}, {'amt': amt})
print(f'Dynamically evaluated fee: ${res:.2f}')
```

### 🔹 Dynamic Code Execution: `exec()`
- **What it does:** Executes arbitrary multi-line Python code blocks dynamically.
- **Syntax:** `exec()`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Demonstrates Dynamic Code Execution with practical fintech data structures and variables in the following code block.


```python
code_block = """
risk_flag = 'HIGH_RISK' if amt > 500 else 'LOW_RISK'
"""
locs = {'amt': amt}
exec(code_block, {}, locs)
print('Dynamic exec result:', locs['risk_flag'])
```
