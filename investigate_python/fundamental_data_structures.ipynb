- **Dictionaries (Mappings)**: Covers `.get()`, `.keys()`, `.values()`, `.items()`, `.update()`, `.pop()`, `.setdefault()`, and `dict.fromkeys()`.


```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
print('Sample Record:', transactions[0])
```

### 🔹 List: `.append()`
- **What it does:** Adds a single element to the end of the list with $O(1)$ amortized time complexity.
- **Syntax:** `.append()`
- **Key Note:** Lists are mutable (changeable). If you want an immutable sequence that cannot be accidentally modified, use a `tuple`.
- **Dataset Application & Code Demonstration:** Applies List on fintech records using columns `transaction_amount` to demonstrate real-world execution.


```python
amt_list = [float(t['transaction_amount']) for t in transactions[:3]]
amt_list.append(999.99)
print('After .append():', amt_list)
```

### 🔹 List: `.extend()`
- **What it does:** Appends elements from an iterable to the end of the list in $O(K)$ time.
- **Syntax:** `.extend()`
- **Key Note:** Lists are mutable (changeable). If you want an immutable sequence that cannot be accidentally modified, use a `tuple`.
- **Dataset Application & Code Demonstration:** Demonstrates List with practical fintech data structures and variables in the following code block.


```python
amt_list.extend([150.0, 250.0])
print('After .extend():', amt_list)
```

### 🔹 List: `.insert()`
- **What it does:** Inserts an element at a specified index in $O(N)$ time (shifting subsequent elements).
- **Syntax:** `.insert()`
- **Key Note:** Lists are mutable (changeable). If you want an immutable sequence that cannot be accidentally modified, use a `tuple`.
- **Dataset Application & Code Demonstration:** Demonstrates List with practical fintech data structures and variables in the following code block.


```python
amt_list.insert(0, 50.00)
print('After .insert(0, 50.00):', amt_list)
```

### 🔹 List: `.remove()`
- **What it does:** Removes the first occurrence of a value from the list (raises ValueError if not present).
- **Syntax:** `.remove()`
- **Key Note:** Lists are mutable (changeable). If you want an immutable sequence that cannot be accidentally modified, use a `tuple`.
- **Dataset Application & Code Demonstration:** Demonstrates List with practical fintech data structures and variables in the following code block.


```python
amt_list.remove(50.00)
print('After .remove(50.00):', amt_list)
```

### 🔹 List: `.pop()`
- **What it does:** Removes and returns element at specified index (default last element in $O(1)$ time).
- **Syntax:** `.pop()`
- **Key Note:** Lists are mutable (changeable). If you want an immutable sequence that cannot be accidentally modified, use a `tuple`.
- **Dataset Application & Code Demonstration:** Demonstrates List with practical fintech data structures and variables in the following code block.


```python
popped = amt_list.pop()
print(f'Popped value: {popped}, List after pop: {amt_list}')
```

### 🔹 List: `.sort()`
- **What it does:** Sorts list in-place using Timsort algorithm ($O(N \log N)$).
- **Syntax:** `.sort()`
- **Key Note:** Lists are mutable (changeable). If you want an immutable sequence that cannot be accidentally modified, use a `tuple`.
- **Dataset Application & Code Demonstration:** Demonstrates List with practical fintech data structures and variables in the following code block.


```python
amt_list.sort(reverse=True)
print('After .sort(reverse=True):', amt_list)
```

### 🔹 List: `.reverse()`
- **What it does:** Reverses list elements in-place in $O(N)$ time.
- **Syntax:** `.reverse()`
- **Key Note:** Lists are mutable (changeable). If you want an immutable sequence that cannot be accidentally modified, use a `tuple`.
- **Dataset Application & Code Demonstration:** Demonstrates List with practical fintech data structures and variables in the following code block.


```python
amt_list.reverse()
print('After .reverse():', amt_list)
```

### 🔹 List: Slicing (`arr[start:stop:step]`)
- **What it does:** Extracts a shallow copy slice of list elements.
- **Syntax:** `arr[start:stop:step]`
- **Key Note:** Remember: Basic slicing creates a *view* into the original array. Modifying a view changes the original array! Use `.copy()` when you need an isolated duplicate.
- **Dataset Application & Code Demonstration:** Applies List on fintech records using columns `transaction_id` to demonstrate real-world execution.


```python
all_ids = [t['transaction_id'] for t in transactions[:10]]
print('Slice [0:5:2]:', all_ids[0:5:2])
print('Reversed slice [::-1]:', all_ids[:3][::-1])
```

### 🔹 Tuples: Packing & Immutability
- **What it does:** Tuples are immutable sequences whose pointers cannot be replaced or reallocated.
- **Syntax:** `function(*args, **kwargs)`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Applies Tuples on fintech records using columns `transaction_amount`, `transaction_id` to demonstrate real-world execution.


```python
sample_tup = (transactions[0]['transaction_id'], float(transactions[0]['transaction_amount']))
print('Packed Tuple:', sample_tup)
```

### 🔹 Tuples: Unpacking
- **What it does:** Unpacks tuple components into individual variables.
- **Syntax:** `function(*args, **kwargs)`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Demonstrates Tuples with practical fintech data structures and variables in the following code block.


```python
tx_id, amount = sample_tup
print(f'Unpacked: {tx_id} -> ${amount}')
```

### 🔹 Tuples: `.count()`
- **What it does:** Returns total occurrences of a value in the tuple.
- **Syntax:** `.count()`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Applies Tuples on fintech records using columns `card_type` to demonstrate real-world execution.


```python
tup_cards = tuple([t['card_type'] for t in transactions[:10]])
print('Visa count in tuple:', tup_cards.count('Visa'))
```

### 🔹 Tuples: `.index()`
- **What it does:** Returns 0-based index of first occurrence of a value.
- **Syntax:** `.index()`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Demonstrates Tuples with practical fintech data structures and variables in the following code block.


```python
print('First Visa index in tuple:', tup_cards.index('Visa') if 'Visa' in tup_cards else 'N/A')
```

### 🔹 Sets: `.add()`
- **What it does:** Adds a single hashable element to the set in $O(1)$ average time.
- **Syntax:** `.add()`
- **Key Note:** Sets only store unique elements and provide $O(1)$ instant lookup time, making `item in my_set` extremely fast.
- **Dataset Application & Code Demonstration:** Demonstrates Sets with practical fintech data structures and variables in the following code block.


```python
card_set = set()
card_set.add('Visa')
card_set.add('MasterCard')
print('Set after .add():', card_set)
```

### 🔹 Sets: `.update()`
- **What it does:** Adds elements from an iterable to the set in-place.
- **Syntax:** `.update()`
- **Key Note:** Ensure your column is converted to datetime with `pd.to_datetime()` before calling `.dt` properties like `.dt.year` or `.dt.day_name()`.
- **Dataset Application & Code Demonstration:** Demonstrates Sets with practical fintech data structures and variables in the following code block.


```python
card_set.update(['Amex', 'Discover', 'Visa'])
print('Set after .update():', card_set)
```

### 🔹 Sets: `.remove()`
- **What it does:** Removes an element from the set (raises KeyError if element is not found).
- **Syntax:** `.remove()`
- **Key Note:** Sets only store unique elements and provide $O(1)$ instant lookup time, making `item in my_set` extremely fast.
- **Dataset Application & Code Demonstration:** Demonstrates Sets with practical fintech data structures and variables in the following code block.


```python
card_set.remove('Discover')
print('Set after .remove():', card_set)
```

### 🔹 Sets: `.discard()`
- **What it does:** Removes an element from the set safely WITHOUT raising an error if absent.
- **Syntax:** `.discard()`
- **Key Note:** Sets only store unique elements and provide $O(1)$ instant lookup time, making `item in my_set` extremely fast.
- **Dataset Application & Code Demonstration:** Demonstrates Sets with practical fintech data structures and variables in the following code block.


```python
card_set.discard('NON_EXISTENT_CARD')
print('Set after .discard() (safe):', card_set)
```

### 🔹 Sets: `.union()` & `|`
- **What it does:** Returns a new set containing all distinct elements from both sets.

- **Key Note:** Sets only store unique elements and provide $O(1)$ instant lookup time, making `item in my_set` extremely fast.

### 🔹 Sets: `.intersection()` & `&`

### 🔹 Sets: `.difference()` & `-`

### 🔹 Dictionaries: `.get()`
- **What it does:** Returns key value safely, returning default if key is not found (avoids KeyError).
- **Syntax:** `.get()`
- **Key Note:** Use `dict.get(key, default)` when looking up keys that might not exist, avoiding unexpected `KeyError` crashes.
- **Dataset Application & Code Demonstration:** Applies Dictionaries on fintech records using columns `card_type` to demonstrate real-world execution.


```python
tx_dict = transactions[0]
print('Existing key .get():', tx_dict.get('card_type'))
print('Missing key .get():', tx_dict.get('non_existent_key', 'DEFAULT_VAL'))
```

### 🔹 Dictionaries: `.keys()`
- **What it does:** Returns a dynamic dict_keys view object of all dictionary keys.
- **Syntax:** `.keys()`
- **Key Note:** Use `dict.get(key, default)` when looking up keys that might not exist, avoiding unexpected `KeyError` crashes.
- **Dataset Application & Code Demonstration:** Demonstrates Dictionaries with practical fintech data structures and variables in the following code block.


```python
print('Dictionary keys view:', list(tx_dict.keys())[:4])
```

### 🔹 Dictionaries: `.values()`
- **What it does:** Returns a dynamic dict_values view object of all values.
- **Syntax:** `.values()`
- **Key Note:** Use `dict.get(key, default)` when looking up keys that might not exist, avoiding unexpected `KeyError` crashes.
- **Dataset Application & Code Demonstration:** Demonstrates Dictionaries with practical fintech data structures and variables in the following code block.


```python
print('Dictionary values view (first 4):', list(tx_dict.values())[:4])
```

### 🔹 Dictionaries: `.items()`
- **What it does:** Returns a dynamic dict_items view object of `(key, value)` pairs.
- **Syntax:** `.items()`
- **Key Note:** Use `dict.get(key, default)` when looking up keys that might not exist, avoiding unexpected `KeyError` crashes.
- **Dataset Application & Code Demonstration:** Demonstrates Dictionaries with practical fintech data structures and variables in the following code block.


```python
for k, v in list(tx_dict.items())[:3]:
    print(f'{k} -> {v}')
```

### 🔹 Dictionaries: `.update()`
- **What it does:** Merges key-value pairs from another dictionary or iterable in-place.
- **Syntax:** `.update()`
- **Key Note:** Ensure your column is converted to datetime with `pd.to_datetime()` before calling `.dt` properties like `.dt.year` or `.dt.day_name()`.
- **Dataset Application & Code Demonstration:** Demonstrates Dictionaries with practical fintech data structures and variables in the following code block.


```python
copy_tx = tx_dict.copy()
copy_tx.update({'audit_status': 'VERIFIED', 'risk_score': 0.12})
print('After .update() audit_status:', copy_tx['audit_status'])
```

### 🔹 Dictionaries: `.pop()`
- **What it does:** Removes specified key and returns its value (or default if missing).
- **Syntax:** `.pop()`
- **Key Note:** Use `dict.get(key, default)` when looking up keys that might not exist, avoiding unexpected `KeyError` crashes.
- **Dataset Application & Code Demonstration:** Demonstrates Dictionaries with practical fintech data structures and variables in the following code block.


```python
removed_val = copy_tx.pop('audit_status')
print('Popped value:', removed_val)
```

### 🔹 Dictionaries: `.setdefault()`
- **What it does:** Returns key value if present; otherwise inserts key with specified default value and returns it.
- **Syntax:** `.setdefault()`
- **Key Note:** Use `dict.get(key, default)` when looking up keys that might not exist, avoiding unexpected `KeyError` crashes.
- **Dataset Application & Code Demonstration:** Applies Dictionaries on fintech records using columns `card_type`, `transaction_amount` to demonstrate real-world execution.


```python
grouped_by_card = {}
for t in transactions[:10]:
    grouped_by_card.setdefault(t['card_type'], []).append(float(t['transaction_amount']))
print('Grouped via .setdefault():', {k: len(v) for k, v in grouped_by_card.items()})
```

### 🔹 Dictionaries: `dict.fromkeys()`
- **What it does:** Creates a new dictionary with keys from an iterable and values set to specified value.
- **Syntax:** `dict.fromkeys()`
- **Key Note:** Use `dict.get(key, default)` when looking up keys that might not exist, avoiding unexpected `KeyError` crashes.
- **Dataset Application & Code Demonstration:** Demonstrates Dictionaries with practical fintech data structures and variables in the following code block.


```python
regions = ['North', 'East', 'South', 'West']
region_totals = dict.fromkeys(regions, 0.0)
print('Dictionary constructed with dict.fromkeys():', region_totals)
```
