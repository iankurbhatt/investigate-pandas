This interactive revision guide uses `data/raw_transactions.csv` with individual dedicated cells per method.

```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
print('Sample Record:', transactions[0])
```

### List Operations (`append`, `extend`, `pop`, `sort`, `slicing`)
**Explanation**: Python lists are dynamic arrays of pointers. `.append()` has $O(1)$ amortized cost; `.insert(0, x)` has $O(N)$ cost; `.pop()` from end is $O(1)$.

**Syntax**: `lst.append()` / `lst.extend()` / `lst.pop()` / `lst.sort()`

```python
amt_list = [float(t['transaction_amount']) for t in transactions[:5]]
amt_list.append(999.50)
amt_list.extend([100.0, 200.0])
popped_val = amt_list.pop() # Removes last in O(1)
amt_list.sort(reverse=True) # Timsort O(N log N)
print('Sorted List Head:', amt_list[:4])
```

### Tuples: Immutability, Packing, `.count()`, `.index()`
**Explanation**: Tuples are immutable fixed-size sequences. Once allocated, pointers cannot be replaced or resized, making them hashable and safe for dictionary keys.

**Syntax**: `tup = (1, 2)` / `tup.count(x)` / `tup.index(x)`

```python
tx_tuple = tuple(transactions[0].values())
print('Tuple representation of transaction:', tx_tuple[:4])
print('Index of "Visa":', tx_tuple.index('Visa') if 'Visa' in tx_tuple else 'N/A')
```

### Set Operations: `add`, `update`, `union`, `intersection`, `difference`
**Explanation**: Sets use open-addressing hash tables for $O(1)$ average membership lookups. Mathematical operators: `|` (union), `&` (intersection), `-` (difference).

**Syntax**: `set.add()` / `set.intersection()` / `set1 & set2`

```python
na_cards = set([t['card_type'] for t in transactions if t['region'] == 'North'])
eu_cards = set([t['card_type'] for t in transactions if t['region'] == 'East'])
print('Union of Cards (|):', na_cards | eu_cards)
print('Intersection (&):', na_cards & eu_cards)
print('Difference (-):', na_cards - eu_cards)
```

### Dictionary Operations: `get`, `setdefault`, `pop`, `update`
**Explanation**: CPython 3.7+ preserves insertion order with compact hash tables. `.get(key, default)` avoids KeyErrors; `.setdefault(key, default)` initializes missing keys.

**Syntax**: `d.get(k, default)` / `d.setdefault(k, [])` / `d.update()`

```python
card_groups = {}
for t in transactions[:20]:
    card_groups.setdefault(t['card_type'], []).append(float(t['transaction_amount']))
for card, amounts in card_groups.items():
    print(f'{card}: {len(amounts)} transactions, Total: ${sum(amounts):,.2f}')
```

### Q1: Dictionary vs List Lookup Complexity ($O(1)$ vs $O(N)$)
**Explanation**: Demonstrate the algorithmic speed advantage of hash-table lookups (`key in dict`) over linear sequence scans (`key in list`).

**Syntax**: `key in hash_set` ($O(1)$) vs `key in list` ($O(N)$)

```python
cust_list = [t['customer_id'] for t in transactions]
cust_set = set(cust_list)
target = cust_list[-1]

t0 = time.perf_counter()
found_list = target in cust_list
t_list = time.perf_counter() - t0

t0 = time.perf_counter()
found_set = target in cust_set
t_set = time.perf_counter() - t0
print(f'List scan: {t_list*1000:.3f} ms vs Set lookup: {t_set*1000:.3f} ms ({t_list/t_set:.1f}x speedup)')
```
