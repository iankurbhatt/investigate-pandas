# Fundamental Data Structures (5+ Years Interview Guide)
Exhaustive revision guide with dedicated separate cells for every List, Tuple, Set, and Dict method on raw_transactions.csv.

### Key 5-Year Interview Concepts Covered:
- **Lists (Mutable)**: Dedicated separate cells for `.append()`, `.extend()`, `.insert()`, `.remove()`, `.pop()`, `.sort()`, `.reverse()`, and slicing.
- **Tuples (Immutable)**: Dedicated separate cells for packing, unpacking, `.count()`, and `.index()`.
- **Sets (Unique)**: Dedicated separate cells for `.add()`, `.update()`, `.remove()`, `.discard()`, `.union()`, `.intersection()`, and `.difference()`.
- **Dictionaries (Mappings)**: Dedicated separate cells for `.get()`, `.keys()`, `.values()`, `.items()`, `.update()`, `.pop()`, `.setdefault()`, and `dict.fromkeys()`.

This interactive revision guide uses `data/raw_transactions.csv` with 1-to-1 atomic method cells.

```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
print('Sample Record:', transactions[0])
```

### List: `.append()`
**Explanation**: Adds a single element to the end of the list with $O(1)$ amortized time complexity.

**Syntax**: `lst.append(item)`

```python
amt_list = [float(t['transaction_amount']) for t in transactions[:3]]
amt_list.append(999.99)
print('After .append():', amt_list)
```

### List: `.extend()`
**Explanation**: Appends elements from an iterable to the end of the list in $O(K)$ time.

**Syntax**: `lst.extend(iterable)`

```python
amt_list.extend([150.0, 250.0])
print('After .extend():', amt_list)
```

### List: `.insert()`
**Explanation**: Inserts an element at a specified index in $O(N)$ time (shifting subsequent elements).

**Syntax**: `lst.insert(index, item)`

```python
amt_list.insert(0, 50.00)
print('After .insert(0, 50.00):', amt_list)
```

### List: `.remove()`
**Explanation**: Removes the first occurrence of a value from the list (raises ValueError if not present).

**Syntax**: `lst.remove(val)`

```python
amt_list.remove(50.00)
print('After .remove(50.00):', amt_list)
```

### List: `.pop()`
**Explanation**: Removes and returns element at specified index (default last element in $O(1)$ time).

**Syntax**: `val = lst.pop()`

```python
popped = amt_list.pop()
print(f'Popped value: {popped}, List after pop: {amt_list}')
```

### List: `.sort()`
**Explanation**: Sorts list in-place using Timsort algorithm ($O(N \log N)$).

**Syntax**: `lst.sort(reverse=True)`

```python
amt_list.sort(reverse=True)
print('After .sort(reverse=True):', amt_list)
```

### List: `.reverse()`
**Explanation**: Reverses list elements in-place in $O(N)$ time.

**Syntax**: `lst.reverse()`

```python
amt_list.reverse()
print('After .reverse():', amt_list)
```

### List: Slicing (`arr[start:stop:step]`)
**Explanation**: Extracts a shallow copy slice of list elements.

**Syntax**: `lst[start:stop:step]`

```python
all_ids = [t['transaction_id'] for t in transactions[:10]]
print('Slice [0:5:2]:', all_ids[0:5:2])
print('Reversed slice [::-1]:', all_ids[:3][::-1])
```

### Tuples: Packing & Immutability
**Explanation**: Tuples are immutable sequences whose pointers cannot be replaced or reallocated.

**Syntax**: `tup = (1, 2, 3)`

```python
sample_tup = (transactions[0]['transaction_id'], float(transactions[0]['transaction_amount']))
print('Packed Tuple:', sample_tup)
```

### Tuples: Unpacking
**Explanation**: Unpacks tuple components into individual variables.

**Syntax**: `a, b = tup`

```python
tx_id, amount = sample_tup
print(f'Unpacked: {tx_id} -> ${amount}')
```

### Tuples: `.count()`
**Explanation**: Returns total occurrences of a value in the tuple.

**Syntax**: `tup.count(val)`

```python
tup_cards = tuple([t['card_type'] for t in transactions[:10]])
print('Visa count in tuple:', tup_cards.count('Visa'))
```

### Tuples: `.index()`
**Explanation**: Returns 0-based index of first occurrence of a value.

**Syntax**: `tup.index(val)`

```python
print('First Visa index in tuple:', tup_cards.index('Visa') if 'Visa' in tup_cards else 'N/A')
```

### Sets: `.add()`
**Explanation**: Adds a single hashable element to the set in $O(1)$ average time.

**Syntax**: `set.add(item)`

```python
card_set = set()
card_set.add('Visa')
card_set.add('MasterCard')
print('Set after .add():', card_set)
```

### Sets: `.update()`
**Explanation**: Adds elements from an iterable to the set in-place.

**Syntax**: `set.update(iterable)`

```python
card_set.update(['Amex', 'Discover', 'Visa'])
print('Set after .update():', card_set)
```

### Sets: `.remove()`
**Explanation**: Removes an element from the set (raises KeyError if element is not found).

**Syntax**: `set.remove(item)`

```python
card_set.remove('Discover')
print('Set after .remove():', card_set)
```

### Sets: `.discard()`
**Explanation**: Removes an element from the set safely WITHOUT raising an error if absent.

**Syntax**: `set.discard(item)`

```python
card_set.discard('NON_EXISTENT_CARD')
print('Set after .discard() (safe):', card_set)
```

### Sets: `.union()` & `|`
**Explanation**: Returns a new set containing all distinct elements from both sets.

**Syntax**: `set1.union(set2)` / `set1 | set2`

```python
set_a = {'North', 'East'}
set_b = {'East', 'South'}
print('Union (set_a | set_b):', set_a | set_b)
```

### Sets: `.intersection()` & `&`
**Explanation**: Returns a new set containing only elements common to both sets.

**Syntax**: `set1.intersection(set2)` / `set1 & set2`

```python
print('Intersection (set_a & set_b):', set_a & set_b)
```

### Sets: `.difference()` & `-`
**Explanation**: Returns elements present in first set but absent from second set.

**Syntax**: `set1.difference(set2)` / `set1 - set2`

```python
print('Difference (set_a - set_b):', set_a - set_b)
```

### Dictionaries: `.get()`
**Explanation**: Returns key value safely, returning default if key is not found (avoids KeyError).

**Syntax**: `d.get(key, default)`

```python
tx_dict = transactions[0]
print('Existing key .get():', tx_dict.get('card_type'))
print('Missing key .get():', tx_dict.get('non_existent_key', 'DEFAULT_VAL'))
```

### Dictionaries: `.keys()`
**Explanation**: Returns a dynamic dict_keys view object of all dictionary keys.

**Syntax**: `d.keys()`

```python
print('Dictionary keys view:', list(tx_dict.keys())[:4])
```

### Dictionaries: `.values()`
**Explanation**: Returns a dynamic dict_values view object of all values.

**Syntax**: `d.values()`

```python
print('Dictionary values view (first 4):', list(tx_dict.values())[:4])
```

### Dictionaries: `.items()`
**Explanation**: Returns a dynamic dict_items view object of `(key, value)` pairs.

**Syntax**: `d.items()`

```python
for k, v in list(tx_dict.items())[:3]:
    print(f'{k} -> {v}')
```

### Dictionaries: `.update()`
**Explanation**: Merges key-value pairs from another dictionary or iterable in-place.

**Syntax**: `d.update(other_dict)`

```python
copy_tx = tx_dict.copy()
copy_tx.update({'audit_status': 'VERIFIED', 'risk_score': 0.12})
print('After .update() audit_status:', copy_tx['audit_status'])
```

### Dictionaries: `.pop()`
**Explanation**: Removes specified key and returns its value (or default if missing).

**Syntax**: `val = d.pop(key)`

```python
removed_val = copy_tx.pop('audit_status')
print('Popped value:', removed_val)
```

### Dictionaries: `.setdefault()`
**Explanation**: Returns key value if present; otherwise inserts key with specified default value and returns it.

**Syntax**: `d.setdefault(key, default)`

```python
grouped_by_card = {}
for t in transactions[:10]:
    grouped_by_card.setdefault(t['card_type'], []).append(float(t['transaction_amount']))
print('Grouped via .setdefault():', {k: len(v) for k, v in grouped_by_card.items()})
```

### Dictionaries: `dict.fromkeys()`
**Explanation**: Creates a new dictionary with keys from an iterable and values set to specified value.

**Syntax**: `dict.fromkeys(keys, initial_val)`

```python
regions = ['North', 'East', 'South', 'West']
region_totals = dict.fromkeys(regions, 0.0)
print('Dictionary constructed with dict.fromkeys():', region_totals)
```
