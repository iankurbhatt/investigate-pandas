This interactive revision guide uses `data/raw_transactions.csv` with 1-to-1 atomic method cells.

```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
print('Sample Record:', transactions[0])
```

### Arithmetic Addition: `+`
**Explanation**: Computes sum of transaction amounts or concatenates sequences.

**Syntax**: `a + b`

```python
amt = float(transactions[0]['transaction_amount'])
print('Amount + $5.50 surcharge:', round(amt + 5.50, 2))
```

### Arithmetic Subtraction: `-`
**Explanation**: Computes difference between amounts.

**Syntax**: `a - b`

```python
print('Amount - $20 discount:', round(amt - 20.0, 2))
```

### Arithmetic Multiplication: `*`
**Explanation**: Multiplies transaction amount by scaling factors.

**Syntax**: `a * b`

```python
print('Amount * 3% tax:', round(amt * 0.03, 2))
```

### Arithmetic True Division: `/`
**Explanation**: Always returns a floating point number.

**Syntax**: `a / b`

```python
print('Amount / 2 equal split:', round(amt / 2, 2))
```

### Arithmetic Floor Division: `//`
**Explanation**: Discards fractional part and rounds down toward negative infinity.

**Syntax**: `a // b`

```python
print('Floor division by 100 (whole $100 denominations):', amt // 100)
```

### Arithmetic Modulo: `%`
**Explanation**: Returns remainder after division.

**Syntax**: `a % b`

```python
print('Modulo 100 (cents/remainder):', round(amt % 100, 2))
```

### Arithmetic Exponentiation: `**`
**Explanation**: Raises base number to exponent power.

**Syntax**: `a ** b`

```python
print('Growth compound factor 1.05 ** 12 (1 year):', round(1.05 ** 12, 4))
```

### Value Equality & Inequality: `==` & `!=`
**Explanation**: `==` compares object values; `!=` verifies values are not equal.

**Syntax**: `a == b` / `a != b`

```python
card = transactions[0]['card_type']
print('Is card Visa?:', card == 'Visa')
print('Is card != Amex?:', card != 'Amex')
```

### Relational Comparisons: `<`, `>`, `<=`, `>=`
**Explanation**: Evaluates ordering relations between numeric magnitudes.

**Syntax**: `a > b` / `a <= b`

```python
print('Is amount > $500?:', amt > 500.0)
print('Is amount <= $2000?:', amt <= 2000.0)
```

### Logical Conjunction: `and`
**Explanation**: Short-circuits: returns the first falsy operand, or the last truthy operand if all are true.

**Syntax**: `cond1 and cond2`

```python
is_high_risk = (amt > 500.0) and (int(transactions[0]['is_fraud']) == 1)
print('High-value fraud transaction?:', is_high_risk)
```

### Logical Disjunction: `or`
**Explanation**: Short-circuits: returns the first truthy operand, or the last falsy operand if all are false.

**Syntax**: `cond1 or cond2`

```python
is_flagged = (amt > 1500.0) or (transactions[0]['card_type'] == 'Discover')
print('Flagged for secondary check?:', is_flagged)
```

### Logical Negation: `not`
**Explanation**: Inverts boolean truth value.

**Syntax**: `not cond`

```python
is_clean = not bool(int(transactions[0]['is_fraud']))
print('Is transaction verified legitimate?:', is_clean)
```

### Identity Testing: `is` & `is not`
**Explanation**: Tests whether two variables reference the exact same memory address pointer (`id(a) == id(b)`).

**Syntax**: `a is None` / `a is b`

```python
none_val = None
print('Is value None?:', none_val is None)
```

### Membership Testing: `in` & `not in`
**Explanation**: Tests if an element exists inside an iterable/container.

**Syntax**: `item in container`

```python
allowed_cards = ['Visa', 'MasterCard', 'Amex']
print('Is card in allowed list?:', card in allowed_cards)
```

### Bitwise AND: `&`
**Explanation**: Performs bitwise conjunction across binary bits.

**Syntax**: `a & b`

```python
FLAG_AUTH = 0b0001
FLAG_SETTLED = 0b0010
status = 0b0011
print('Is FLAG_AUTH bit set?:', bool(status & FLAG_AUTH))
```

### Bitwise OR: `|`
**Explanation**: Performs bitwise disjunction (combines flag masks).

**Syntax**: `a | b`

```python
combined_mask = FLAG_AUTH | FLAG_SETTLED
print('Combined Bitmask Integer:', bin(combined_mask))
```

### Bitwise XOR: `^`
**Explanation**: Performs bitwise exclusive-OR (toggles bits).

**Syntax**: `a ^ b`

```python
toggled = status ^ FLAG_AUTH
print('Toggled status:', bin(toggled))
```

### Bitwise NOT: `~`
**Explanation**: Inverts all bits (two's complement: `~x = -x - 1`).

**Syntax**: `~a`

```python
print('Bitwise NOT ~1:', ~1)
```

### Bitwise Left Shift: `<<`
**Explanation**: Shifts bits to left (multiplies by $2^n$).

**Syntax**: `a << 2`

```python
print('1 << 3 (2^3 = 8):', 1 << 3)
```

### Bitwise Right Shift: `>>`
**Explanation**: Shifts bits to right (divides by $2^n$).

**Syntax**: `a >> 2`

```python
print('16 >> 2 (16 / 4 = 4):', 16 >> 2)
```

### Walrus Operator: `:=` (Assignment Expression)
**Explanation**: Assigns values to variables inside expressions (PEP 572).

**Syntax**: `if (n := len(x)) > 5:`

```python
if (high_amt := float(transactions[0]['transaction_amount'])) > 100.0:
    print(f'Captured high transaction amount inline via Walrus := : ${high_amt:.2f}')
```
