# Operators, Expressions & Walrus Operator: Beginner Guide

### 🌟 What Are Python Operators & the Walrus Operator?
**Operators** are the mathematical and logical building blocks of Python, enabling arithmetic calculations (`+`, `-`, `*`, `/`, `//`, `%`), value comparisons (`==`, `<`, `>`), logical evaluations (`and`, `or`, `not`), and bitwise operations. Python 3.8+ also introduced the **Walrus Operator (`:=`)**, which allows you to compute and assign a variable directly inside an expression or loop condition.

This interactive guide loads and works directly with `data/raw_transactions.csv`, giving you real-world hands-on practice.

### 📚 Key Concepts Covered in this Notebook:
- **Arithmetic Operators**: Covers `+`, `-`, `*`, `/`, `//`, `%`, and `**`.
- **Comparison & Identity**: Covers `==`, `!=`, `<`, `>`, `is`, and `in`.
- **Logical Operators**: Covers `and`, `or`, and `not`.
- **Bitwise Operators**: Covers `&`, `|`, `^`, `~`, `<<`, and `>>`.
- **Walrus Operator**: Covers inline assignment `:=`.


```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
print('Sample Record:', transactions[0])
```

### 🔹 Bitwise AND: `&`
`🔹 Bitwise AND: `&`` provides an essential way to process and transform data in Python. It provides a robust, standardized way to process, clean, and analyze datasets reliably and efficiently. **Tip:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

**Syntax:** `a & b`


```python
FLAG_AUTH = 0b0001
FLAG_SETTLED = 0b0010
status = 0b0011
print('Is FLAG_AUTH bit set?:', bool(status & FLAG_AUTH))
```

### 🔹 Bitwise OR: `|`
`🔹 Bitwise OR: `|`` provides an essential way to process and transform data in Python. It provides a robust, standardized way to process, clean, and analyze datasets reliably and efficiently. **Tip:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

**Syntax:** `a | b`


```python
combined_mask = FLAG_AUTH | FLAG_SETTLED
print('Combined Bitmask Integer:', bin(combined_mask))
```

`🔹 Bitwise XOR: `^`` provides an essential way to process and transform data in Python. ### 🔹 Bitwise NOT: `~`
Inverts all bits (two's complement: `~x = -x - 1`). It provides a robust, standardized way to process, clean, and analyze datasets reliably and efficiently. **Tip:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

### 🔹 Bitwise Left Shift: `<<`
Shifts bits to left (multiplies by $2^n$). It provides a robust, standardized way to process, clean, and analyze datasets reliably and efficiently. **Tip:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

**Syntax:** `a << 2`


```python
print('1 << 3 (2^3 = 8):', 1 << 3)
```

### 🔹 Bitwise Right Shift: `>>`
Shifts bits to right (divides by $2^n$). It provides a robust, standardized way to process, clean, and analyze datasets reliably and efficiently. **Tip:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

**Syntax:** `a >> 2`


```python
print('16 >> 2 (16 / 4 = 4):', 16 >> 2)
```

### 🔹 Walrus Operator: `:=` (Assignment Expression)
Assigns values to variables inside expressions (PEP 572). Clean variable assignments and unpacking make your code readable, concise, and easy to maintain. **Tip:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

**Syntax:** `if (n := len(x)) > 5:`


```python
if (high_amt := float(transactions[0]['transaction_amount'])) > 100.0:
    print(f'Captured high transaction amount inline via Walrus := : ${high_amt:.2f}')
```
