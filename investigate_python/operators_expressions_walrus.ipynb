```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
print('Sample Record:', transactions[0])
```

```python
amt = float(transactions[0]['transaction_amount'])
print('Amount + $5.50 surcharge:', round(amt + 5.50, 2))
```

### 🔹 Logical Conjunction: `and`
- **What it does:** Short-circuits: returns the first falsy operand, or the last truthy operand if all are true.
- **Syntax:** `and`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Applies Logical Conjunction on fintech records using columns `is_fraud` to demonstrate real-world execution.


```python
is_high_risk = (amt > 500.0) and (int(transactions[0]['is_fraud']) == 1)
print('High-value fraud transaction?:', is_high_risk)
```

### 🔹 Logical Disjunction: `or`

```python
is_flagged = (amt > 1500.0) or (transactions[0]['card_type'] == 'Discover')
print('Flagged for secondary check?:', is_flagged)
```

### 🔹 Bitwise AND: `&`
- **What it does:** `🔹 Bitwise AND: `&`` provides an essential way to process and transform data in Python.
- **Syntax:** `&`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Demonstrates Bitwise AND with practical fintech data structures and variables in the following code block.


```python
FLAG_AUTH = 0b0001
FLAG_SETTLED = 0b0010
status = 0b0011
print('Is FLAG_AUTH bit set?:', bool(status & FLAG_AUTH))
```

### 🔹 Bitwise OR: `|`
- **What it does:** `🔹 Bitwise OR: `|`` provides an essential way to process and transform data in Python.
- **Syntax:** `|`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Demonstrates Bitwise OR with practical fintech data structures and variables in the following code block.


```python
combined_mask = FLAG_AUTH | FLAG_SETTLED
print('Combined Bitmask Integer:', bin(combined_mask))
```

### 🔹 Bitwise XOR: `^`
- **What it does:** `🔹 Bitwise XOR: `^`` provides an essential way to process and transform data in Python.
- **Syntax:** `^`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Demonstrates Bitwise XOR with practical fintech data structures and variables in the following code block.


```python
toggled = status ^ FLAG_AUTH
print('Toggled status:', bin(toggled))
```

- **Dataset Application & Code Demonstration:** Demonstrates Bitwise NOT with practical fintech data structures and variables in the following code block.

### 🔹 Bitwise Left Shift: `<<`
- **What it does:** Shifts bits to left (multiplies by $2^n$).
- **Syntax:** `<<`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Demonstrates Bitwise Left Shift with practical fintech data structures and variables in the following code block.


```python
print('1 << 3 (2^3 = 8):', 1 << 3)
```

### 🔹 Bitwise Right Shift: `>>`
- **What it does:** Shifts bits to right (divides by $2^n$).
- **Syntax:** `>>`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
print('16 >> 2 (16 / 4 = 4):', 16 >> 2)
```

### 🔹 Walrus Operator: `:=` (Assignment Expression)
- **What it does:** Assigns values to variables inside expressions (PEP 572).
- **Syntax:** `:=`
  - **Parameters:**
    - `row_label` (*hashable*): Row label.
    - `col_label` (*hashable*): Column label.
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Applies Walrus Operator on fintech records using columns `transaction_amount` to demonstrate real-world execution.


```python
if (high_amt := float(transactions[0]['transaction_amount'])) > 100.0:
    print(f'Captured high transaction amount inline via Walrus := : ${high_amt:.2f}')
```
