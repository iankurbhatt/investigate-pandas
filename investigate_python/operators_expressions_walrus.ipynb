This interactive revision guide uses `data/raw_transactions.csv` with individual dedicated cells per method.

```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
print('Sample Record:', transactions[0])
```

### Arithmetic Evaluation (`+`, `-`, `*`, `/`, `//`, `%`, `**`)
**Explanation**: Floor division `//` discards the fractional part (rounds toward negative infinity). Modulo `%` computes remainder.

**Syntax**: `a // b` / `a % b` / `a ** 2`

```python
amt = float(transactions[0]['transaction_amount'])
print('Amount:', amt)
print('Floor division by 100 (whole hundreds):', amt // 100)
print('Modulo 100 (cents/remainder):', round(amt % 100, 2))
print('Squared:', round(amt ** 2, 2))
```

### Comparison, Logical Short-Circuiting & Identity
**Explanation**: `==` tests value equality; `is` tests memory address identity (`id(a) == id(b)`). Logical `and`/`or` short-circuits: returns the first truthy/falsy operand without evaluating the second.

**Syntax**: `a == b` / `a is b` / `x and y` / `item in container`

```python
tx1 = transactions[0]
tx2 = transactions[0]
print('tx1 == tx2 (value equality):', tx1 == tx2)
print('tx1 is tx2 (same object in memory):', tx1 is tx2)
print('Membership check ("Visa" in tx1["card_type"]):', 'Visa' in tx1['card_type'])
```

### Bitwise Operations (`&`, `|`, `^`, `~`, `<<`, `>>`)
**Explanation**: Bitwise operators manipulate integers at the binary level. Standard for flag bitmasking and permission modeling in trading systems.

**Syntax**: `a & b` / `a | b` / `a ^ b` / `a << 1`

```python
FLAG_FRAUD = 1 << 0   # 1 (0001)
FLAG_HIGH_VAL = 1 << 1 # 2 (0010)
FLAG_DEVICE = 1 << 2   # 4 (0100)

# Combine flags
status_mask = FLAG_FRAUD | FLAG_HIGH_VAL
print('Bitmask status integer:', status_mask)
print('Has FLAG_FRAUD bit set?:', bool(status_mask & FLAG_FRAUD))
print('Has FLAG_DEVICE bit set?:', bool(status_mask & FLAG_DEVICE))
```

### The Walrus Operator (`:=`) Assignment Expressions
**Explanation**: The walrus operator `:=` (PEP 572) assigns values to variables inside expressions, eliminating redundant function calls in conditionals, comprehensions, and while-loops.

**Syntax**: `if (n := len(x)) > threshold: ...`

```python
high_val_txs = []
for t in transactions[:20]:
    if (val := float(t['transaction_amount'])) > 1000.0:
        high_val_txs.append((t['transaction_id'], val))
print(f'Found {len(high_val_txs)} high-value transactions using walrus operator:')
print(high_val_txs[:3])
```

### Q1: Short-Circuit Evaluation and Defensive Default Assignment
**Explanation**: Demonstrate how `or` short-circuits to assign default values to missing dictionary keys without raising KeyErrors.

**Syntax**: `val = raw_dict.get('key') or 'DEFAULT'`

```python
raw_payload = {'user': 'Alice', 'region': None}
active_region = raw_payload.get('region') or 'North America'  # None is falsy, defaults to 'North America'
print('Resolved Region via short-circuiting:', active_region)
```
