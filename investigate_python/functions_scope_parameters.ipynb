```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
```

### Function Definition & Docstrings: `def` and `__doc__`
**Explanation**: Functions encapsulate reusable logic blocks. Docstrings document function contracts accessible via `func.__doc__` and `help(func)`.

**Syntax**: `def calculate_fee(amount: float) -> float: """Docstring"""`

```python
def calculate_fee(amount: float, rate: float = 0.02) -> float:
    """Calculates merchant processing interchange fee."""
    return round(amount * rate, 2)

print('Docstring:', calculate_fee.__doc__)
print('Fee for $500:', calculate_fee(500.0))
```

### Variable Arguments: `*args` and `**kwargs`
**Explanation**: `*args` collects arbitrary positional arguments into a `tuple`. `**kwargs` collects arbitrary keyword arguments into a `dict`.

**Syntax**: `def log_tx(*args, **kwargs): ...`

```python
def log_transaction(tx_id, *flags, **metadata):
    print(f'Logging {tx_id}: Flags={flags}, Metadata={metadata}')

tx = transactions[0]
log_transaction(tx['transaction_id'], 'HIGH_VALUE', 'VIP', region=tx['region'], card=tx['card_type'])
```

### Scope Resolution (LEGB) & `nonlocal` / `global`
**Explanation**: Python resolves names in order: Local -> Enclosing -> Global -> Built-in. Use `nonlocal` to bind enclosing state in closures; `global` to bind module-level state.

**Syntax**: `nonlocal var` / `global var`

```python
def make_running_total():
    total = 0.0
    def add(amount):
        nonlocal total
        total += amount
        return total
    return add

running_acc = make_running_total()
for t in transactions[:3]:
    print(f"Adding ${float(t['transaction_amount']):.2f} -> Running Total: ${running_acc(float(t['transaction_amount'])):,.2f}")
```

### Anonymous Lambda Functions
**Explanation**: Inline throwaway functions defined with `lambda params: expr`. Standard for sorting keys and callbacks.

**Syntax**: `lambda x: x['transaction_amount']`

```python
sorted_sample = sorted(transactions[:5], key=lambda t: float(t['transaction_amount']), reverse=True)
for t in sorted_sample:
    print(f"{t['transaction_id']}: ${float(t['transaction_amount']):.2f}")
```

### Q1: Mutable Default Argument Gotcha (`def f(x=[])`)
**Explanation**: Explain why mutable default arguments are evaluated once at function definition time, leading to shared state across invocations.

**Syntax**: `def append_to(item, target=None): if target is None: target = []`

```python
def safe_collector(tx_id, container=None):
    if container is None:
        container = []
    container.append(tx_id)
    return container

print('Call 1:', safe_collector('TX_1'))
print('Call 2:', safe_collector('TX_2'))
```
