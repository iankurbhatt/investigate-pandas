```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
```

### 🔹 Function Definition: `def func(params):`
- **What it does:** Creates a function object and binds it to a name in the current namespace.
- **Syntax:** `def calculate_total(amt, fee): return amt + fee`
- **Operation:** `def calculate_total(amount, fee):`
- **Key Note:** Always remember to include `self` as the first parameter in instance methods so Python knows which object instance is executing.

```python
def calculate_total(amount, fee):
    return amount + fee

print('Function result:', calculate_total(100.0, 2.50))
```

### 🔹 Docstrings: `func.__doc__`
- **What it does:** Docstrings document function contracts accessible via `__doc__` and `help()`.
- **Syntax:** `"""Docstring explanation"""`
- **Operation:** `def audit_transaction(tx_id: str) -> bool:`
- **Key Note:** Pandas `.str` accessor methods automatically skip missing (`NaN`) values instead of raising an `AttributeError`.

```python
def audit_transaction(tx_id: str) -> bool:
    """Validates financial transaction against compliance rules."""
    return tx_id.startswith('TX')

print('Accessed Docstring:', repr(audit_transaction.__doc__))
```

### 🔹 Positional Arguments
- **What it does:** Arguments passed into function parameters strictly mapped by position order.
- **Syntax:** `func(val1, val2)`
- **Operation:** `def format_tx(tx_id, amt):`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
def format_tx(tx_id, amt):
    return f'{tx_id}: ${amt}'

print('Positional args call:', format_tx(transactions[0]['transaction_id'], transactions[0]['transaction_amount']))
```

### 🔹 Keyword Arguments
- **What it does:** Arguments passed explicitly with parameter names, allowing arbitrary call order.
- **Syntax:** `func(param=val)`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
print('Keyword args call:', format_tx(amt=transactions[0]['transaction_amount'], tx_id=transactions[0]['transaction_id']))
```

### 🔹 Default Parameters
- **What it does:** Provides fallback parameter values when omitted in the caller's invocation.
- **Syntax:** `def func(amt, rate=0.02): ...`
- **Operation:** `def apply_surcharge(amount, rate=0.025):`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
def apply_surcharge(amount, rate=0.025):
    return round(amount * (1 + rate), 2)

print('Default rate (2.5%):', apply_surcharge(1000.0))
print('Overridden rate (5%):', apply_surcharge(1000.0, rate=0.05))
```

### 🔹 Variable Positional Arguments: `*args`
- **What it does:** Collects arbitrary additional positional arguments into a `tuple`.
- **Syntax:** `def func(*args): ...`
- **Operation:** `def sum_all_fees(*fees):`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
def sum_all_fees(*fees):
    print('Received *args tuple:', fees)
    return sum(fees)

print('Total Fees ($):', sum_all_fees(1.50, 2.75, 0.50, 3.25))
```

### 🔹 Variable Keyword Arguments: `**kwargs`
- **What it does:** Collects arbitrary additional keyword arguments into a `dict`.
- **Syntax:** `def func(**kwargs): ...`
- **Operation:** `def log_metadata(**metadata):`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
def log_metadata(**metadata):
    print('Received **kwargs dict:', metadata)

log_metadata(tx_id=transactions[0]['transaction_id'], region=transactions[0]['region'], card=transactions[0]['card_type'])
```

### 🔹 Scope Resolution (LEGB Rule)
- **What it does:** Python searches names in order: Local -> Enclosing -> Global -> Built-in.
- **Syntax:** `Local -> Enclosing -> Global -> Built-in`
- **Operation:** `global_tax = 0.08`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
global_tax = 0.08
def outer_scope():
    enclosing_tax = 0.05
    def inner_scope():
        local_tax = 0.02
        return local_tax, enclosing_tax, global_tax
    return inner_scope()

print('LEGB Scope resolution values (local, enclosing, global):', outer_scope())
```

### 🔹 Modifying Global Namespace: `global`
- **What it does:** Binds a variable inside a local function frame to the module-level global namespace.
- **Syntax:** `global var_name`
- **Operation:** `processed_count = 0`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
processed_count = 0
def increment_processed():
    global processed_count
    processed_count += 1

increment_processed()
increment_processed()
print('Global processed_count:', processed_count)
```

### 🔹 Modifying Enclosing Namespace: `nonlocal`
- **What it does:** Binds a variable to the nearest enclosing scope (excluding global) inside nested closures.
- **Syntax:** `nonlocal var_name`
- **Operation:** `def make_counter():`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
def make_counter():
    count = 0
    def inc():
        nonlocal count
        count += 1
        return count
    return inc

counter = make_counter()
print('Closure counter calls:', counter(), counter(), counter())
```

### 🔹 Anonymous Lambda Functions: `lambda`
- **What it does:** Inline anonymous function declaration (`lambda params: expression`).
- **Syntax:** `lambda x: x * 2`
- **Operation:** `get_amt = lambda tx_dict: float(tx_dict['transaction_amount'])`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
get_amt = lambda tx_dict: float(tx_dict['transaction_amount'])
print('Lambda extracted amount:', get_amt(transactions[0]))
```

### 🔍 Scenario: Q1: Python Late Binding Closures in Loops
- **Objective:** Q1: Python Late Binding Closures in Loops
- **Approach:** Explain why lambdas created inside loops bind variables by reference (late binding) and how default parameter binding `lambda x=i: x` solves it.
- **Syntax:** `[lambda x=i: x for i in range(5)]`

```python
funcs = [lambda x=i: x for i in range(3)]
print('Evaluated closures with default binding:', [f() for f in funcs])
```
