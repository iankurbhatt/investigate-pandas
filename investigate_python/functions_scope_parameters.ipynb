```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
```

### 🔹 Default Parameters
Provides fallback parameter values when omitted in the caller's invocation. It provides a robust, standardized way to process, clean, and analyze datasets reliably and efficiently. **Tip:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

**Syntax:** `def func(amt, rate=0.02): ...`


```python
def apply_surcharge(amount, rate=0.025):
    return round(amount * (1 + rate), 2)

print('Default rate (2.5%):', apply_surcharge(1000.0))
print('Overridden rate (5%):', apply_surcharge(1000.0, rate=0.05))
```

### 🔹 Variable Positional Arguments: `*args`

### 🔹 Scope Resolution (LEGB Rule)
Python searches names in order: Local -> Enclosing -> Global -> Built-in. Saving and loading data properly ensures results are persistent, shareable across teams, and ready for downstream analysis. **Tip:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

**Syntax:** `Local -> Enclosing -> Global -> Built-in`


```python
global_tax = 0.08
def outer_scope():
    enclosing_tax = 0.05
    def inner_scope():
        local_tax = 0.02
        return local_tax, enclosing_tax, global_tax
    return inner_scope()

print('LEGB Scope resolution values (local, enclosing, global):', outer_scope())
```

### 🔹 Modifying Global Namespace: `global`
Binds a variable inside a local function frame to the module-level global namespace. It provides a robust, standardized way to process, clean, and analyze datasets reliably and efficiently. **Tip:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

**Syntax:** `global var_name`


```python
processed_count = 0
def increment_processed():
    global processed_count
    processed_count += 1

increment_processed()
increment_processed()
print('Global processed_count:', processed_count)
```

### 🔹 Modifying Enclosing Namespace: `nonlocal`
Binds a variable to the nearest enclosing scope (excluding global) inside nested closures. It provides a robust, standardized way to process, clean, and analyze datasets reliably and efficiently. **Tip:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

**Syntax:** `nonlocal var_name`


```python
def make_counter():
    count = 0
    def inc():
        nonlocal count
        count += 1
        return count
    return inc

counter = make_counter()
print('Closure counter calls:', counter(), counter(), counter())
```

**Approach:** Explain why lambdas created inside loops bind variables by reference (late binding) and how default parameter binding `lambda x=i: x` solves it.
**Syntax:** `[lambda x=i: x for i in range(5)]`
