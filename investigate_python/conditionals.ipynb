### 1. Single condition checks (if)
**Explanation**: Conditionally execute a block.

**Syntax**:
```python
if condition:
    # block
```



````python
is_payout_valid = True
if is_payout_valid: print('Executed')
````

### 2. Two-way alternative branches (if-else)
**Explanation**: Executes fallback path when condition fails.

**Syntax**:
```python
if condition:
    # block
else:
    # fallback
```



````python
is_transaction_completed = False
if is_transaction_completed: print('Success') 
else: print('Fallback path')
````

### 3. Multi-branch conditionals ladders
**Explanation**: Checks sequential paths top-down.

**Syntax**:
```python
if condition_one:
    # b1
elif condition_two:
    # b2
else:
    # fallback
```



````python
transaction_amount = 5
if transaction_amount > 10: print('Tier 1')
elif transaction_amount > 3: print('Tier 2')
else: print('Tier 3')
````

### 4. Nested Conditionals routing
**Explanation**: Nest conditional checks inside outer paths.

**Syntax**:
```python
if condition_one:
    if condition_two:
        # nested
```



````python
is_authenticated = True
is_limit_cleared = True
if is_authenticated:
    if is_limit_cleared: print('Nested paths passed')
````

### 5. Inline ternary conditional expressions
**Explanation**: Compact assignment conditional syntax.

**Syntax**:
```python
variable = true_value if condition else false_value
```



````python
transaction_status_label = 'Odd' if 5%2 else 'Even'
print(transaction_status_label)
````

### 6. Truthy/Falsy Rules for Boolean states
**Explanation**: Direct boolean check rules.

**Syntax**:
```python
if boolean_flag:
```



````python
is_fraud_flagged = True
if is_fraud_flagged: print('Truthy condition resolved')
````

### 7. Truthy/Falsy Rules for Empty Lists & Dictionaries
**Explanation**: Empty collections evaluate to False.

**Syntax**:
```python
if not collection_object:
```



````python
empty_ledger_list = []
if not empty_ledger_list: print('Empty collection is Falsy')
````

### 8. Truthy/Falsy Rules for Zeros and None
**Explanation**: Numerical 0 values and None are Falsy.

**Syntax**:
```python
if not numerical_value:
```



````python
zero_balance_amount = 0
if not zero_balance_amount: print('Zero amount is Falsy')
````

### 9. Structural Pattern Matching Basics
**Explanation**: Multi-branch pattern routing.

**Syntax**:
```python
match variable_name:
    case pattern_value:
        # block
```

**Visual Explanation (Data with Baraa Style)**:
```mermaid
graph TD
    match[match resolved_status_code]
    match -->|case 'OK'| run1(Print Success)
    match -->|case _| run2(Default Case)
```


````python
resolved_status_code = 'OK'
match resolved_status_code:
    case 'OK': print('Success status resolved')
    case _: print('Default case')
````

### 10. Match-Case Wildcard fallbacks
**Explanation**: Matches any unhandled patterns.

**Syntax**:
```python
match variable_name:
    case _:
        # fallback
```



````python
unrecognized_region = 'Unknown'
match unrecognized_region:
    case 'North': print('North region resolved')
    case _: print('Wildcard matched default')
````

### 11. Match-Case guard conditions
**Explanation**: Add conditionals inside match cases.

**Syntax**:
```python
case pattern_value if guard_condition:
```



````python
payout_limit_value = 10
match payout_limit_value:
    case value if value > 5: print('Matched guard condition filter')
````

### 12. Matching sequences in patterns
**Explanation**: Unpack and match sequence patterns.

**Syntax**:
```python
case [first_item, *remaining_items]:
```



````python
transaction_sequence = [1, 2, 3]
match transaction_sequence:
    case [1, *remaining_items]: print('Starts with 1, remaining items:', remaining_items)
````

### 13. Matching dictionaries in patterns
**Explanation**: Matches dict values by keys.

**Syntax**:
```python
case {'key_name': variable_name}:
```



````python
transaction_metadata_dict = {'status': 'Failed'}
match transaction_metadata_dict:
    case {'status': status_value}: print('Status was:', status_value)
````

### 14. Combining pattern options
**Explanation**: Use pipe operator to match multiple cases.

**Syntax**:
```python
case value_one | value_two:
```



````python
current_transaction_state = 'Failed'
match current_transaction_state:
    case 'Failed' | 'Reversed': print('Requires operational action')
````

### 15. Conditional boolean chaining
**Explanation**: Combine checks with logical priority.

**Syntax**:
```python
if condition_one and (condition_two or condition_three):
```



````python
if True and (False or True): print('Chained priority logical condition matched')
````

### Q1: Categorize transaction tier using structural pattern matching on region and amount from the first transaction.

### Q2: Use dictionary pattern matching to process the first transaction log row mapped as a dict.

````python
# Solution:
with open(csv_path, 'r') as f:
    f.readline()
    row = f.readline().strip().split(',')
    transaction_data_dict = {'id': row[0], 'status': row[5]}
    match transaction_data_dict:
        case {'status': 'Failed' | 'Reversed'}:
            print('Action Required for ID:', transaction_data_dict['id'])
        case _:
            print('Cleared status:', transaction_data_dict['status'])
````
