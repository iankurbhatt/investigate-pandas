# File Handling & Basic I/O (5+ Years Interview Guide)
Exhaustive revision guide to file streams, open modes, context managers (with), tell/seek buffer manipulation, and safe I/O on raw_transactions.csv.

### Key 5-Year Interview Concepts Covered:
- **Stream Opening & Modes**: Dedicated cell for `open(mode='r'/'w'/'a'/'rb')` and `encoding='utf-8'`.
- **Context Management**: Dedicated cell for `with open(...) as f:`.
- **Data Reading & Writing**: Dedicated cell for `.read()`, `.readline()`, `.readlines()`, `.write()`, and `.writelines()`.
- **Stream Positioning**: Dedicated cell for `.tell()`, `.seek()`, and `.flush()`.

This interactive revision guide uses `data/raw_transactions.csv` with individual dedicated cells per method.

```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
```

### Opening File Streams with Encodings: `open()`
**Explanation**: Opens OS file descriptors with specified modes (`'r'`, `'w'`, `'a'`, `'rb'`, `'wb'`) and Unicode encodings (`'utf-8'`).

**Syntax**: `open('file.txt', mode='r', encoding='utf-8')`

```python
f = open(csv_path, mode='r', encoding='utf-8')
header_line = f.readline().strip()
f.close()
print('CSV Header Line:', header_line)
```

### Context Management with `with open(...) as f:`
**Explanation**: Ensures deterministic file descriptor closure even when exceptions are raised.

**Syntax**: `with open(...) as f: ...`

```python
with open(csv_path, mode='r', encoding='utf-8') as f:
    first_record = f.readline().strip()
    second_record = f.readline().strip()
print('Read securely via Context Manager:\n', second_record)
```

### Reading & Writing Routines (`read`, `readline`, `writelines`)
**Explanation**: Reads chunks, lines, or writes sequences of strings to disk.

**Syntax**: `f.read(n)` / `f.readlines()` / `f.writelines(lines)`

```python
os.makedirs('scratch', exist_ok=True)
with open('scratch/sample_output.txt', mode='w', encoding='utf-8') as out_f:
    out_f.writelines([f"{t['transaction_id']}: ${t['transaction_amount']}\n" for t in transactions[:3]])

with open('scratch/sample_output.txt', mode='r', encoding='utf-8') as in_f:
    print('Read written file contents:\n', in_f.read().strip())
```

### Stream Positioning & Flushing: `tell()` and `seek()`
**Explanation**: `tell()` returns current byte offset in stream. `seek(offset, whence)` repositions cursor (0=start, 1=current, 2=end).

**Syntax**: `f.tell()` / `f.seek(0)` / `f.flush()`

```python
with open(csv_path, mode='r', encoding='utf-8') as f:
    print('Initial stream position:', f.tell())
    f.readline()
    pos_after_header = f.tell()
    print('Position after header:', pos_after_header)
    f.seek(0) # Rewind to start
    print('Position after seek(0):', f.tell())
```

### Q1: Large File Streaming Without Memory Spikes
**Explanation**: Demonstrate line-by-line file iteration (`for line in f:`) which streams chunks lazily without loading the entire multi-gigabyte file into RAM.

**Syntax**: `for line in file_stream: process(line)` ($O(1)$ RAM)

```python
line_count = 0
with open(csv_path, mode='r', encoding='utf-8') as f:
    for line in f:
        line_count += 1
print(f'Streamed {line_count} total lines with constant memory.')
```
