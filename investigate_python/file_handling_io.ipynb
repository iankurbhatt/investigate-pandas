```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
```

### 🔹 Stream Opening: Read Mode `'r'`
- **What it does:** Opens an existing text file for sequential reading.
- **Syntax:** `open(file, mode='r', encoding='utf-8')`
- **Operation:** `f = open(csv_path, mode='r', encoding='utf-8')`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
f = open(csv_path, mode='r', encoding='utf-8')
first_line = f.readline().strip()
f.close()
print('Read Mode First Line:', first_line)
```

### 🔹 Stream Opening: Write Mode `'w'`
- **What it does:** Creates a new file for writing, truncating/overwriting any existing content.
- **Syntax:** `open(file, mode='w', encoding='utf-8')`
- **Operation:** `os.makedirs('scratch', exist_ok=True)`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
os.makedirs('scratch', exist_ok=True)
with open('scratch/demo_write.txt', mode='w', encoding='utf-8') as f:
    f.write('TX_AUDIT_LOG_START\n')
print('Write mode file created.')
```

### 🔹 Stream Opening: Append Mode `'a'`
- **What it does:** Opens file for appending data to end without truncating existing content.
- **Syntax:** `open(file, mode='a', encoding='utf-8')`
- **Operation:** `with open('scratch/demo_write.txt', mode='a', encoding='utf-8') as f:`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
with open('scratch/demo_write.txt', mode='a', encoding='utf-8') as f:
    f.write('TX_LOG_ENTRY_2\n')
print('Append mode written.')
```

### 🔹 Stream Opening: Binary Mode `'rb'` & `'wb'`
- **What it does:** Reads and writes raw unencoded bytes directly to and from disk.
- **Syntax:** `open(file, mode='rb')`
- **Operation:** `with open(csv_path, mode='rb') as f:`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
with open(csv_path, mode='rb') as f:
    raw_bytes = f.read(30)
print('Read 30 raw binary bytes:', raw_bytes)
```

### 🔹 Context Management: `with open(...) as f:`
- **What it does:** Ensures deterministic stream closure via `__exit__` even upon unhandled exceptions.
- **Syntax:** `with open(...) as f: ...`
- **Operation:** `with open(csv_path, mode='r', encoding='utf-8') as f:`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
with open(csv_path, mode='r', encoding='utf-8') as f:
    header = f.readline().strip()
print('Opened securely via context manager:', header)
```

### 🔹 Data Reading: `.read()`
- **What it does:** Reads entire file (or up to `size` characters) into a single string.
- **Syntax:** `content = f.read(100)`
- **Operation:** `with open(csv_path, mode='r', encoding='utf-8') as f:`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
with open(csv_path, mode='r', encoding='utf-8') as f:
    chunk = f.read(80)
print('Read 80 characters:', repr(chunk))
```

### 🔹 Data Reading: `.readline()`
- **What it does:** Reads next single line including trailing newline `\n`.
- **Syntax:** `line = f.readline()`
- **Operation:** `with open(csv_path, mode='r', encoding='utf-8') as f:`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
with open(csv_path, mode='r', encoding='utf-8') as f:
    print('Line 1:', f.readline().strip())
    print('Line 2:', f.readline().strip())
```

### 🔹 Data Reading: `.readlines()`
- **What it does:** Reads all remaining lines into a list of strings.
- **Syntax:** `lines = f.readlines()`
- **Operation:** `with open(csv_path, mode='r', encoding='utf-8') as f:`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
with open(csv_path, mode='r', encoding='utf-8') as f:
    f.readline() # Skip header
    top3_lines = [f.readline().strip() for _ in range(3)]
print('Read 3 lines:', top3_lines)
```

### 🔹 Data Writing: `.write()`
- **What it does:** Writes string data into output stream buffer.
- **Syntax:** `f.write(str_data)`
- **Operation:** `with open('scratch/demo_write.txt', mode='w', encoding='utf-8') as f:`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
with open('scratch/demo_write.txt', mode='w', encoding='utf-8') as f:
    bytes_written = f.write(f"{transactions[0]['transaction_id']},{transactions[0]['transaction_amount']}\n")
print('Bytes written:', bytes_written)
```

### 🔹 Data Writing: `.writelines()`
- **What it does:** Writes an iterable sequence of strings to disk without automatically adding newlines.
- **Syntax:** `f.writelines(list_of_lines)`
- **Operation:** `lines_to_write = [f"{t['transaction_id']}\n" for t in transactions[:3]]`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
lines_to_write = [f"{t['transaction_id']}\n" for t in transactions[:3]]
with open('scratch/demo_write.txt', mode='w', encoding='utf-8') as f:
    f.writelines(lines_to_write)
print('Writelines completed.')
```

### 🔹 Stream Positioning: `.tell()`
- **What it does:** Returns current integer byte position offset in file stream.
- **Syntax:** `pos = f.tell()`
- **Operation:** `with open(csv_path, mode='r', encoding='utf-8') as f:`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
with open(csv_path, mode='r', encoding='utf-8') as f:
    print('Initial position:', f.tell())
    f.readline()
    print('Position after header line:', f.tell())
```

### 🔹 Stream Positioning: `.seek()`
- **What it does:** Repositions file stream cursor to specified byte offset (`0` = start).
- **Syntax:** `f.seek(0)`
- **Operation:** `with open(csv_path, mode='r', encoding='utf-8') as f:`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
with open(csv_path, mode='r', encoding='utf-8') as f:
    f.readline()
    f.seek(0) # Rewind to start
    print('Position after seek(0):', f.tell())
```

### 🔹 Stream Buffer: `.flush()`
- **What it does:** Forces internal memory buffers to write to OS kernel disk cache immediately.
- **Syntax:** `f.flush()`
- **Operation:** `with open('scratch/demo_write.txt', mode='w', encoding='utf-8') as f:`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

```python
with open('scratch/demo_write.txt', mode='w', encoding='utf-8') as f:
    f.write('CRITICAL_TRANSACTION_RECORD\n')
    f.flush()
print('Flushed buffer to OS.')
```

### 🔍 Scenario: Q1: Generator-Based Line Streaming for Out-of-Memory Files
- **Objective:** Q1: Generator-Based Line Streaming for Out-of-Memory Files
- **Approach:** Demonstrate constant $O(1)$ memory file streaming using `for line in f:`.
- **Syntax:** `for line in f: yield parse(line)`

```python
total_lines = 0
with open(csv_path, mode='r', encoding='utf-8') as f:
    for line in f:
        total_lines += 1
print(f'Streamed {total_lines} lines with O(1) constant RAM footprint.')
```
