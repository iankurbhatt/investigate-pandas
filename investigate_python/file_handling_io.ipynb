# File Handling & Basic I/O (5+ Years Interview Guide)
Exhaustive revision guide with dedicated separate cells for open modes, context managers, read, readline, readlines, write, writelines, tell, seek, and flush on raw_transactions.csv.

### Key 5-Year Interview Concepts Covered:
- **Stream Opening**: Dedicated separate cells for read mode (`'r'`), write mode (`'w'`), append mode (`'a'`), and binary mode (`'rb'`/`'wb'`).
- **Context Management**: Dedicated separate cell for `with open(...) as f:`.
- **Data Reading**: Dedicated separate cells for `.read()`, `.readline()`, and `.readlines()`.
- **Data Writing**: Dedicated separate cells for `.write()` and `.writelines()`.
- **Stream Positioning**: Dedicated separate cells for `.tell()`, `.seek()`, and `.flush()`.

This interactive revision guide uses `data/raw_transactions.csv` with 1-to-1 atomic method cells.

```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
```

### Stream Opening: Read Mode `'r'`
**Explanation**: Opens an existing text file for sequential reading.

**Syntax**: `open(file, mode='r', encoding='utf-8')`

```python
f = open(csv_path, mode='r', encoding='utf-8')
first_line = f.readline().strip()
f.close()
print('Read Mode First Line:', first_line)
```

### Stream Opening: Write Mode `'w'`
**Explanation**: Creates a new file for writing, truncating/overwriting any existing content.

**Syntax**: `open(file, mode='w', encoding='utf-8')`

```python
os.makedirs('scratch', exist_ok=True)
with open('scratch/demo_write.txt', mode='w', encoding='utf-8') as f:
    f.write('TX_AUDIT_LOG_START\n')
print('Write mode file created.')
```

### Stream Opening: Append Mode `'a'`
**Explanation**: Opens file for appending data to end without truncating existing content.

**Syntax**: `open(file, mode='a', encoding='utf-8')`

```python
with open('scratch/demo_write.txt', mode='a', encoding='utf-8') as f:
    f.write('TX_LOG_ENTRY_2\n')
print('Append mode written.')
```

### Stream Opening: Binary Mode `'rb'` & `'wb'`
**Explanation**: Reads and writes raw unencoded bytes directly to and from disk.

**Syntax**: `open(file, mode='rb')`

```python
with open(csv_path, mode='rb') as f:
    raw_bytes = f.read(30)
print('Read 30 raw binary bytes:', raw_bytes)
```

### Context Management: `with open(...) as f:`
**Explanation**: Ensures deterministic stream closure via `__exit__` even upon unhandled exceptions.

**Syntax**: `with open(...) as f: ...`

```python
with open(csv_path, mode='r', encoding='utf-8') as f:
    header = f.readline().strip()
print('Opened securely via context manager:', header)
```

### Data Reading: `.read()`
**Explanation**: Reads entire file (or up to `size` characters) into a single string.

**Syntax**: `content = f.read(100)`

```python
with open(csv_path, mode='r', encoding='utf-8') as f:
    chunk = f.read(80)
print('Read 80 characters:', repr(chunk))
```

### Data Reading: `.readline()`
**Explanation**: Reads next single line including trailing newline `\n`.

**Syntax**: `line = f.readline()`

```python
with open(csv_path, mode='r', encoding='utf-8') as f:
    print('Line 1:', f.readline().strip())
    print('Line 2:', f.readline().strip())
```

### Data Reading: `.readlines()`
**Explanation**: Reads all remaining lines into a list of strings.

**Syntax**: `lines = f.readlines()`

```python
with open(csv_path, mode='r', encoding='utf-8') as f:
    f.readline() # Skip header
    top3_lines = [f.readline().strip() for _ in range(3)]
print('Read 3 lines:', top3_lines)
```

### Data Writing: `.write()`
**Explanation**: Writes string data into output stream buffer.

**Syntax**: `f.write(str_data)`

```python
with open('scratch/demo_write.txt', mode='w', encoding='utf-8') as f:
    bytes_written = f.write(f"{transactions[0]['transaction_id']},{transactions[0]['transaction_amount']}\n")
print('Bytes written:', bytes_written)
```

### Data Writing: `.writelines()`
**Explanation**: Writes an iterable sequence of strings to disk without automatically adding newlines.

**Syntax**: `f.writelines(list_of_lines)`

```python
lines_to_write = [f"{t['transaction_id']}\n" for t in transactions[:3]]
with open('scratch/demo_write.txt', mode='w', encoding='utf-8') as f:
    f.writelines(lines_to_write)
print('Writelines completed.')
```

### Stream Positioning: `.tell()`
**Explanation**: Returns current integer byte position offset in file stream.

**Syntax**: `pos = f.tell()`

```python
with open(csv_path, mode='r', encoding='utf-8') as f:
    print('Initial position:', f.tell())
    f.readline()
    print('Position after header line:', f.tell())
```

### Stream Positioning: `.seek()`
**Explanation**: Repositions file stream cursor to specified byte offset (`0` = start).

**Syntax**: `f.seek(0)`

```python
with open(csv_path, mode='r', encoding='utf-8') as f:
    f.readline()
    f.seek(0) # Rewind to start
    print('Position after seek(0):', f.tell())
```

### Stream Buffer: `.flush()`
**Explanation**: Forces internal memory buffers to write to OS kernel disk cache immediately.

**Syntax**: `f.flush()`

```python
with open('scratch/demo_write.txt', mode='w', encoding='utf-8') as f:
    f.write('CRITICAL_TRANSACTION_RECORD\n')
    f.flush()
print('Flushed buffer to OS.')
```

### Q1: Generator-Based Line Streaming for Out-of-Memory Files
**Explanation**: Demonstrate constant $O(1)$ memory file streaming using `for line in f:`.

**Syntax**: `for line in f: yield parse(line)`

```python
total_lines = 0
with open(csv_path, mode='r', encoding='utf-8') as f:
    for line in f:
        total_lines += 1
print(f'Streamed {total_lines} lines with O(1) constant RAM footprint.')
```
