This interactive revision guide uses `data/raw_transactions.csv` with individual dedicated cells per method.

```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
print('Sample Record:', transactions[0])
```

### Type Inspection: `type()` vs `isinstance()`
**Explanation**: `type(obj)` returns the exact class of an object (ignoring inheritance). `isinstance(obj, classinfo)` checks whether an object is an instance of a class or any of its subclasses (recommended for polymorphic checks).

**Syntax**: `type(obj)` / `isinstance(obj, (int, float))`

```python
tx = transactions[0]
amount_raw = tx['transaction_amount']
print('type(amount_raw):', type(amount_raw))
print('isinstance(amount_raw, str):', isinstance(amount_raw, str))
print('isinstance(amount_raw, (int, float, str)):', isinstance(amount_raw, (int, float, str)))
```

### Numeric Conversion: `int()` & `float()`
**Explanation**: Converts string transaction attributes to binary integer and floating-point primitives.

**Syntax**: `int(str_val)` / `float(str_val)`

```python
amount = float(tx['transaction_amount'])
age = int(tx['account_age_months'])
print(f'Converted: amount={amount} (type={type(amount)}), age={age} (type={type(age)})')
```

### String & Boolean Conversion: `str()` & `bool()`
**Explanation**: `str(obj)` converts any object into its string representation. `bool(obj)` evaluates truthiness (`0`, `None`, empty collections `[]`/`{}`/`''` evaluate to `False`; everything else evaluates to `True`).

**Syntax**: `str(amount)` / `bool(is_fraud)`

```python
is_fraud = bool(int(tx['is_fraud']))
print('String representation:', str(amount))
print('Boolean truth value of is_fraud:', is_fraud)
```

### Collection Conversion: `list()`, `tuple()`, `set()`, `dict()`
**Explanation**: Converts iterable structures across mutable sequences (`list`), immutable sequences (`tuple`), distinct sets (`set`), and key-value mappings (`dict`).

**Syntax**: `list(iterable)` / `tuple(iterable)` / `set(iterable)` / `dict(pairs)`

```python
card_types = [t['card_type'] for t in transactions[:10]]
print('List of cards:', card_types)
print('Tuple of cards:', tuple(card_types))
print('Set of unique cards:', set(card_types))
print('Dict of tx_id -> amount:', dict([(t['transaction_id'], float(t['transaction_amount'])) for t in transactions[:3]]))
```

### Chained Assignment (`x = y = 10`)
**Explanation**: Binds multiple variable names to the exact same object in memory simultaneously. Caution: If the object is mutable (like a list), mutating through `x` will mutate `y`.

**Syntax**: `a = b = initial_value`

```python
default_limit = max_limit = 5000.0
print('default_limit id:', id(default_limit), 'max_limit id:', id(max_limit))
print('Are both pointing to same object?:', default_limit is max_limit)
```

### Multiple Variable Unpacking (`a, b = 1, 2`)
**Explanation**: Unpacks elements of an iterable into separate variables in a single statement. Supports extended star unpacking (`first, *rest, last = items`).

**Syntax**: `a, b, c = sequence` / `first, *middle, last = sequence`

```python
tx_id, customer_id, amount_val = tx['transaction_id'], tx['customer_id'], float(tx['transaction_amount'])
print(f'Unpacked: {tx_id} | {customer_id} | ${amount_val}')
first_tx, *middle_txs, last_tx = [t['transaction_id'] for t in transactions[:5]]
print('Star Unpacking (first, middle count, last):', first_tx, len(middle_txs), last_tx)
```

### Basic I/O Formatting: `print(sep, end)`
**Explanation**: `print(*objects, sep=' ', end='\n', file=sys.stdout, flush=False)` outputs formatted text streams with customizable delimiters.

**Syntax**: `print(a, b, sep=' | ', end='\n')`

```python
print('TX_AUDIT', tx['transaction_id'], tx['transaction_amount'], tx['card_type'], sep=' | ')
```

### Q1: Small Integer Caching & String Interning Internals
**Explanation**: Demonstrate CPython's small integer cache (`[-5, 256]`) where integers share identical memory pointers (`is` returns `True`).

**Syntax**: `a is b  # True for ints in [-5, 256]`

```python
x = 250
y = 250
print('250 is 250 (cached in [-5, 256]):', x is y)
p = 1000
q = 1000
print('1000 is 1000 (outside cache, separate objects):', p is q)
```
