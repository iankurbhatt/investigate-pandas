```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
print('Sample Record:', transactions[0])
```

### 🔹 Type Inspection: `type(obj)`
- **What it does:** `type(obj)` returns the exact class type of the object in Python, directly inspecting the internal type definition.
- **Syntax:** `type(obj)`
- **Key Note:** Prefer `isinstance(obj, ClassName)` over `type(obj) == ClassName` when checking types in conditional logic.
- **Dataset Application & Code Demonstration:** Applies Type Inspection on fintech records using columns `transaction_amount` to demonstrate real-world execution.


```python
tx = transactions[0]
amount_raw = tx['transaction_amount']
print('Type of raw amount string:', type(amount_raw))
print('Type of whole transaction record:', type(tx))
```

### 🔹 Type Inspection: `isinstance(obj, classinfo)`
- **What it does:** `isinstance(obj, classinfo)` tests if an object is an instance of a class or a subclass.
- **Syntax:** `isinstance(obj, classinfo)`
  - **Optional Parameters:**
    - `memory_usage` (*bool, str, optional*): If True, include memory usage; 'deep' for deep introspection.
    - `show_counts` (*bool, optional*): Whether to show non-null counts.
- **Key Note:** Always remember to include `self` as the first parameter in instance methods so Python knows which object instance is executing.
- **Dataset Application & Code Demonstration:** Demonstrates Type Inspection with practical fintech data structures and variables in the following code block.


```python
print('Is amount string an instance of str?:', isinstance(amount_raw, str))
print('Is amount string an instance of (int, float)?:', isinstance(amount_raw, (int, float)))
```

### 🔹 Type Conversion: `int()`
- **What it does:** `int()` converts numbers or formatted numeric strings into standard arbitrary-precision integers.
- **Syntax:** `int()`
- **Key Note:** Prefer `isinstance(obj, ClassName)` over `type(obj) == ClassName` when checking types in conditional logic.
- **Dataset Application & Code Demonstration:** Applies Type Conversion on fintech records using columns `account_age_months` to demonstrate real-world execution.


```python
account_age = int(tx['account_age_months'])
print(f'Converted account age: {account_age} (type={type(account_age)})')
```

### 🔹 Type Conversion: `float()`
- **What it does:** `float()` converts numeric strings or integers into IEEE 754 64-bit double-precision floating point numbers.
- **Syntax:** `float()`
  - **Parameters:**
    - `row_label` (*hashable*): Row label.
    - `col_label` (*hashable*): Column label.
- **Key Note:** Prefer `isinstance(obj, ClassName)` over `type(obj) == ClassName` when checking types in conditional logic.
- **Dataset Application & Code Demonstration:** Applies Type Conversion on fintech records using columns `transaction_amount` to demonstrate real-world execution.


```python
amount = float(tx['transaction_amount'])
print(f'Converted amount: {amount} (type={type(amount)})')
```

### 🔹 Type Conversion: `str()`
- **What it does:** `str()` invokes the object's `__str__()` protocol to return an informal human-readable string representation.
- **Syntax:** `str()`
- **Key Note:** Prefer `isinstance(obj, ClassName)` over `type(obj) == ClassName` when checking types in conditional logic.
- **Dataset Application & Code Demonstration:** Demonstrates Type Conversion with practical fintech data structures and variables in the following code block.


```python
amount_str = str(amount)
print('Converted back to string:', repr(amount_str))
```

### 🔹 Type Conversion: `bool()`
- **What it does:** `bool()` evaluates truthiness: `0`, `0.0`, `None`, and empty collections `[]`/`{}` evaluate to `False`; all non-zero/non-empty objects evaluate to `True`.
- **Syntax:** `bool()`
- **Key Note:** Prefer `isinstance(obj, ClassName)` over `type(obj) == ClassName` when checking types in conditional logic.
- **Dataset Application & Code Demonstration:** Applies Type Conversion on fintech records using columns `is_fraud` to demonstrate real-world execution.


```python
is_fraud = bool(int(tx['is_fraud']))
print('Truth value of fraud flag:', is_fraud)
print('Truth value of empty string bool(""):', bool(''))
```

### 🔹 Type Conversion: `list()`
- **What it does:** `list()` constructs a new mutable dynamic array sequence from any iterable.
- **Syntax:** `list()`
- **Key Note:** Lists are mutable (changeable). If you want an immutable sequence that cannot be accidentally modified, use a `tuple`.
- **Dataset Application & Code Demonstration:** Applies Type Conversion on fintech records using columns `card_type` to demonstrate real-world execution.


```python
cards_list = list([t['card_type'] for t in transactions[:5]])
print('Constructed list:', cards_list)
```

### 🔹 Type Conversion: `tuple()`
- **What it does:** `tuple()` converts an iterable into an immutable fixed-size sequence.
- **Syntax:** `tuple()`
- **Key Note:** Prefer `isinstance(obj, ClassName)` over `type(obj) == ClassName` when checking types in conditional logic.
- **Dataset Application & Code Demonstration:** Demonstrates Type Conversion with practical fintech data structures and variables in the following code block.


```python
tx_tuple = tuple(tx.values())
print('Constructed immutable tuple (first 4 elements):', tx_tuple[:4])
```

### 🔹 Type Conversion: `set()`
- **What it does:** `set()` converts an iterable into an unordered collection of distinct, hashable elements.
- **Syntax:** `set()`
- **Key Note:** Sets only store unique elements and provide $O(1)$ instant lookup time, making `item in my_set` extremely fast.
- **Dataset Application & Code Demonstration:** Applies Type Conversion on fintech records using columns `region` to demonstrate real-world execution.


```python
unique_regions = set([t['region'] for t in transactions[:20]])
print('Constructed unique set:', unique_regions)
```

### 🔹 Type Conversion: `dict()`
- **What it does:** `dict()` constructs a key-value mapping from sequences of key-value pairs or keyword arguments.
- **Syntax:** `dict()`
- **Key Note:** Use `dict.get(key, default)` when looking up keys that might not exist, avoiding unexpected `KeyError` crashes.
- **Dataset Application & Code Demonstration:** Applies Type Conversion on fintech records using columns `transaction_amount`, `transaction_id` to demonstrate real-world execution.


```python
kv_pairs = [('tx_id', tx['transaction_id']), ('amount', float(tx['transaction_amount']))]
constructed_dict = dict(kv_pairs)
print('Constructed dictionary:', constructed_dict)
```

### 🔹 Single Variable Assignment: `x = val`
- **What it does:** Creates an object in memory and binds a variable reference name to that object's pointer.
- **Syntax:** `x = val`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Demonstrates Single Variable Assignment with practical fintech data structures and variables in the following code block.


```python
currency = 'USD'
print(f'Bound currency variable: {currency} at id {id(currency)}')
```

### 🔹 Chained Variable Assignment: `x = y = 10`
- **What it does:** Binds multiple variable names simultaneously to the exact same object memory address.
- **Syntax:** `x = y = 10`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Demonstrates Chained Variable Assignment with practical fintech data structures and variables in the following code block.


```python
daily_limit = monthly_limit = 10000.0
print('daily_limit id:', id(daily_limit), 'monthly_limit id:', id(monthly_limit))
print('Are pointers identical (daily_limit is monthly_limit)?:', daily_limit is monthly_limit)
```

### 🔹 Multiple Variable Unpacking: `a, b = 1, 2` & `*rest`
- **What it does:** Unpacks elements of an iterable into separate variables in a single step.
- **Syntax:** `a, b = 1, 2`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Applies Multiple Variable Unpacking on fintech records using columns `transaction_id` to demonstrate real-world execution.


```python
first_id, second_id = transactions[0]['transaction_id'], transactions[1]['transaction_id']
print(f'Unpacked single pair: {first_id}, {second_id}')

head_tx, *middle_txs, tail_tx = [t['transaction_id'] for t in transactions[:6]]
print(f'Star Unpacking -> Head: {head_tx}, Middle Count: {len(middle_txs)}, Tail: {tail_tx}')
```

### 🔹 Basic I/O Formatting: `print(val, sep=' ', end='\n')`
- **What it does:** `print(*objects, sep=' ', end='\n')` outputs string data with custom item separators and line terminators.
- **Syntax:** `print(val, sep=' ', end='\n')`
  - **Parameters:**
    - `row_label` (*hashable*): Row label.
    - `col_label` (*hashable*): Column label.
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Applies Basic I/O Formatting on fintech records using columns `card_type`, `transaction_amount`, `transaction_id` to demonstrate real-world execution.


```python
print('AUDIT_REPORT', tx['transaction_id'], tx['transaction_amount'], tx['card_type'], sep=' | ', end=' [END RECORD]\n')
```
