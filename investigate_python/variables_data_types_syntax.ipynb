# Variables, Data Types & Basic Syntax: Beginner Guide

### 🌟 What Are Variables, Data Types & Syntax in Python?
In Python, a **variable** is a friendly name that references a piece of data stored in your computer's memory. Python automatically handles data types dynamically, including whole numbers (`int`), decimal amounts (`float`), text strings (`str`), and true/false conditions (`bool`). Understanding how Python creates, inspects (`type()`, `isinstance()`), and converts these types is the fundamental first step to writing reliable data processing code.

This interactive guide loads and works directly with `data/raw_transactions.csv`, giving you real-world hands-on practice.

### 📚 Key Concepts Covered in this Notebook:
- **Type Inspection**: Covers `type()` and `isinstance()`.
- **Type Conversion**: Covers `int()`, `float()`, `str()`, `bool()`, `list()`, `tuple()`, `set()`, and `dict()`.
- **Variable Assignment**: Covers single assignment, chained assignment (`x = y = 10`), and multiple unpacking (`a, b = 1, 2`).
- **Basic I/O**: Covers `print(sep, end)` and `input()` mechanics.


```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
print('Sample Record:', transactions[0])
```

### 🔹 Type Inspection: `type(obj)`
`type(obj)` returns the exact class type of the object in Python, directly inspecting the internal type definition. Checking and validating data types prevents subtle runtime errors and ensures subsequent mathematical or string operations behave properly. **Tip:** Prefer `isinstance(obj, ClassName)` over `type(obj) == ClassName` when checking types in conditional logic.

**Syntax:** `type(obj)`


```python
tx = transactions[0]
amount_raw = tx['transaction_amount']
print('Type of raw amount string:', type(amount_raw))
print('Type of whole transaction record:', type(tx))
```

### 🔹 Type Inspection: `isinstance(obj, classinfo)`
`isinstance(obj, classinfo)` tests if an object is an instance of a class or a subclass. Supports passing a tuple of candidate types. Checking and validating data types prevents subtle runtime errors and ensures subsequent mathematical or string operations behave properly. **Tip:** Always remember to include `self` as the first parameter in instance methods so Python knows which object instance is executing.

**Syntax:** `isinstance(obj, (int, float, str))`


```python
print('Is amount string an instance of str?:', isinstance(amount_raw, str))
print('Is amount string an instance of (int, float)?:', isinstance(amount_raw, (int, float)))
```

### 🔹 Type Conversion: `int()`
`int()` converts numbers or formatted numeric strings into standard arbitrary-precision integers. Checking and validating data types prevents subtle runtime errors and ensures subsequent mathematical or string operations behave properly. **Tip:** Prefer `isinstance(obj, ClassName)` over `type(obj) == ClassName` when checking types in conditional logic.

**Syntax:** `int(val)`


```python
account_age = int(tx['account_age_months'])
print(f'Converted account age: {account_age} (type={type(account_age)})')
```

### 🔹 Type Conversion: `float()`
`float()` converts numeric strings or integers into IEEE 754 64-bit double-precision floating point numbers. Checking and validating data types prevents subtle runtime errors and ensures subsequent mathematical or string operations behave properly. **Tip:** Prefer `isinstance(obj, ClassName)` over `type(obj) == ClassName` when checking types in conditional logic.

**Syntax:** `float(val)`


```python
amount = float(tx['transaction_amount'])
print(f'Converted amount: {amount} (type={type(amount)})')
```

### 🔹 Type Conversion: `str()`
`str()` invokes the object's `__str__()` protocol to return an informal human-readable string representation. Checking and validating data types prevents subtle runtime errors and ensures subsequent mathematical or string operations behave properly. **Tip:** Prefer `isinstance(obj, ClassName)` over `type(obj) == ClassName` when checking types in conditional logic.

**Syntax:** `str(val)`


```python
amount_str = str(amount)
print('Converted back to string:', repr(amount_str))
```

### 🔹 Type Conversion: `bool()`
`bool()` evaluates truthiness: `0`, `0.0`, `None`, and empty collections `[]`/`{}` evaluate to `False`; all non-zero/non-empty objects evaluate to `True`. Checking and validating data types prevents subtle runtime errors and ensures subsequent mathematical or string operations behave properly. **Tip:** Prefer `isinstance(obj, ClassName)` over `type(obj) == ClassName` when checking types in conditional logic.

**Syntax:** `bool(val)`


```python
is_fraud = bool(int(tx['is_fraud']))
print('Truth value of fraud flag:', is_fraud)
print('Truth value of empty string bool(""):', bool(''))
```

### 🔹 Type Conversion: `list()`
`list()` constructs a new mutable dynamic array sequence from any iterable. Checking and validating data types prevents subtle runtime errors and ensures subsequent mathematical or string operations behave properly. **Tip:** Lists are mutable (changeable). If you want an immutable sequence that cannot be accidentally modified, use a `tuple`.

**Syntax:** `list(iterable)`


```python
cards_list = list([t['card_type'] for t in transactions[:5]])
print('Constructed list:', cards_list)
```

### 🔹 Type Conversion: `tuple()`
`tuple()` converts an iterable into an immutable fixed-size sequence. Checking and validating data types prevents subtle runtime errors and ensures subsequent mathematical or string operations behave properly. **Tip:** Prefer `isinstance(obj, ClassName)` over `type(obj) == ClassName` when checking types in conditional logic.

**Syntax:** `tuple(iterable)`


```python
tx_tuple = tuple(tx.values())
print('Constructed immutable tuple (first 4 elements):', tx_tuple[:4])
```

### 🔹 Type Conversion: `set()`
`set()` converts an iterable into an unordered collection of distinct, hashable elements. Checking and validating data types prevents subtle runtime errors and ensures subsequent mathematical or string operations behave properly. **Tip:** Sets only store unique elements and provide $O(1)$ instant lookup time, making `item in my_set` extremely fast.

**Syntax:** `set(iterable)`


```python
unique_regions = set([t['region'] for t in transactions[:20]])
print('Constructed unique set:', unique_regions)
```

### 🔹 Type Conversion: `dict()`
`dict()` constructs a key-value mapping from sequences of key-value pairs or keyword arguments. Checking and validating data types prevents subtle runtime errors and ensures subsequent mathematical or string operations behave properly. **Tip:** Use `dict.get(key, default)` when looking up keys that might not exist, avoiding unexpected `KeyError` crashes.

**Syntax:** `dict(pairs)`


```python
kv_pairs = [('tx_id', tx['transaction_id']), ('amount', float(tx['transaction_amount']))]
constructed_dict = dict(kv_pairs)
print('Constructed dictionary:', constructed_dict)
```

### 🔹 Single Variable Assignment: `x = val`
Creates an object in memory and binds a variable reference name to that object's pointer. Clean variable assignments and unpacking make your code readable, concise, and easy to maintain. **Tip:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

**Syntax:** `x = val`


```python
currency = 'USD'
print(f'Bound currency variable: {currency} at id {id(currency)}')
```

### 🔹 Chained Variable Assignment: `x = y = 10`
Binds multiple variable names simultaneously to the exact same object memory address. Clean variable assignments and unpacking make your code readable, concise, and easy to maintain. **Tip:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

**Syntax:** `x = y = val`


```python
daily_limit = monthly_limit = 10000.0
print('daily_limit id:', id(daily_limit), 'monthly_limit id:', id(monthly_limit))
print('Are pointers identical (daily_limit is monthly_limit)?:', daily_limit is monthly_limit)
```

### 🔹 Multiple Variable Unpacking: `a, b = 1, 2` & `*rest`
Unpacks elements of an iterable into separate variables in a single step. Clean variable assignments and unpacking make your code readable, concise, and easy to maintain. **Tip:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

**Syntax:** `a, b = tup`


```python
first_id, second_id = transactions[0]['transaction_id'], transactions[1]['transaction_id']
print(f'Unpacked single pair: {first_id}, {second_id}')

head_tx, *middle_txs, tail_tx = [t['transaction_id'] for t in transactions[:6]]
print(f'Star Unpacking -> Head: {head_tx}, Middle Count: {len(middle_txs)}, Tail: {tail_tx}')
```

### 🔹 Basic I/O Formatting: `print(val, sep=' ', end='\n')`
`print(*objects, sep=' ', end='\n')` outputs string data with custom item separators and line terminators. It provides a robust, standardized way to process, clean, and analyze datasets reliably and efficiently. **Tip:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

**Syntax:** `print(a, b, sep=' | ', end='\n')`


```python
print('AUDIT_REPORT', tx['transaction_id'], tx['transaction_amount'], tx['card_type'], sep=' | ', end=' [END RECORD]\n')
```
