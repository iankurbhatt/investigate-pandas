```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
print('Sample Record:', transactions[0])
```

### 🔹 For-Loop Iteration: `for item in iterable:`
- **What it does:** Iterates sequentially across elements using Python's underlying iterator protocol.
- **Syntax:** `for item in iterable:`
- **Key Note:** Generators calculate values lazily on demand. This lets you process millions of rows without loading the entire dataset into memory at once.
- **Dataset Application & Code Demonstration:** Applies For-Loop Iteration on fintech records using columns `transaction_amount`, `transaction_id` to demonstrate real-world execution.


```python
for t in transactions[:3]:
    print(f'Processing {t["transaction_id"]}: ${float(t["transaction_amount"]):.2f}')
```

### 🔹 Sequence Generation: `range(stop)`
- **What it does:** Generates integer sequence from 0 to `stop - 1` with $O(1)$ constant memory.
- **Syntax:** `range(stop)`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Applies Sequence Generation on fintech records using columns `transaction_id` to demonstrate real-world execution.


```python
for i in range(3):
    print(f'Iteration Index {i}: {transactions[i]["transaction_id"]}')
```

### 🔹 Stepped Sequence Generation: `range(start, stop, step)`
- **What it does:** Generates stepped progressions skipping intermediate indices.
- **Syntax:** `range(start, stop, step)`
- **Key Note:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.
- **Dataset Application & Code Demonstration:** Applies Stepped Sequence Generation on fintech records using columns `transaction_id` to demonstrate real-world execution.


```python
for i in range(0, 6, 2):
    print(f'Stepped Offset {i}: {transactions[i]["transaction_id"]}')
```

### 🔹 While-Loop Execution: `while condition:`
- **What it does:** Repeats execution dynamically while a state predicate evaluates to True.
- **Syntax:** `while condition:`
- **Key Note:** Always remember to include `self` as the first parameter in instance methods so Python knows which object instance is executing.
- **Dataset Application & Code Demonstration:** Applies While-Loop Execution on fintech records using columns `transaction_id` to demonstrate real-world execution.


```python
count = 0
while count < 3:
    print(f'While Count {count}: {transactions[count]["transaction_id"]}')
    count += 1
```

### 🔹 Loop Interruption: `break`
- **What it does:** Terminates loop execution immediately upon hitting trigger condition.
- **Syntax:** `break`
- **Key Note:** Always remember to include `self` as the first parameter in instance methods so Python knows which object instance is executing.
- **Dataset Application & Code Demonstration:** Applies Loop Interruption on fintech records using columns `is_fraud`, `transaction_id` to demonstrate real-world execution.


```python
for t in transactions:
    if int(t['is_fraud']) == 1:
        print(f'First Fraud Intercepted: {t["transaction_id"]}. Breaking loop!')
        break
```

### 🔹 Loop Skipping: `continue`
- **What it does:** Skips remainder of current iteration and moves immediately to next item.
- **Syntax:** `continue`
- **Key Note:** Always remember to include `self` as the first parameter in instance methods so Python knows which object instance is executing.
- **Dataset Application & Code Demonstration:** Applies Loop Skipping on fintech records using columns `transaction_amount`, `transaction_id` to demonstrate real-world execution.


```python
for t in transactions[:5]:
    if float(t['transaction_amount']) < 500.0:
        continue
    print(f'High-Value Transaction: {t["transaction_id"]} (${float(t["transaction_amount"]):.2f})')
```

### 🔹 For-Loop Fallback Clause: `for ... else:`
- **What it does:** `else:` block executes ONLY if loop completes naturally without encountering a `break`.
- **Syntax:** `for ... else:`
- **Key Note:** Always remember to include `self` as the first parameter in instance methods so Python knows which object instance is executing.
- **Dataset Application & Code Demonstration:** Applies For-Loop Fallback Clause on fintech records using columns `transaction_id` to demonstrate real-world execution.


```python
search_id = 'NON_EXISTENT_TX'
for t in transactions[:5]:
    if t['transaction_id'] == search_id:
        print('Found!')
        break
else:
    print(f'Search ID {search_id} not found (for...else executed).')
```

### 🔹 While-Loop Fallback Clause: `while ... else:`
- **What it does:** `else:` block executes when while condition becomes False (and loop was not terminated by break).
- **Syntax:** `while ... else:`
- **Key Note:** Always remember to include `self` as the first parameter in instance methods so Python knows which object instance is executing.
- **Dataset Application & Code Demonstration:** Demonstrates While-Loop Fallback Clause with practical fintech data structures and variables in the following code block.


```python
n = 0
while n < 2:
    n += 1
else:
    print('While condition completed cleanly (while...else executed).')
```

### 🔍 Scenario: Q1: Memory Optimization: Python `range` vs `list(range)`
- **Objective:** Q1: Memory Optimization: Python `range` vs `list(range)`
- **Approach:** Explain why `range` objects consume only 48 bytes regardless of span size ($O(1)$ memory).
- **Syntax:** `sys.getsizeof(range(1000000))`

```python
print('Size of range(0, 10_000_000):', sys.getsizeof(range(0, 10_000_000)), 'bytes')
print('Size of list(range(100)):', sys.getsizeof(list(range(100))), 'bytes')
```
