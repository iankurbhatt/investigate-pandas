Exhaustive revision guide to for-loops, range generators, while-loops, break/continue flow control, and for/while...else fallback logic on raw_transactions.csv.

### Key 5-Year Interview Concepts Covered:
- **For-Loop Iteration**: Dedicated cell for `for item in iterable:`.
- **Sequence Generation**: Dedicated cell for `range(start, stop, step)`.
- **While-Loop Execution**: Dedicated cell for `while condition:`.
- **Loop Flow Control**: Dedicated cell for `break`, `continue`, and `for/while ... else:`.

This interactive revision guide uses `data/raw_transactions.csv` with individual dedicated cells per method.

```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
print('Sample Record:', transactions[0])
```

### For-Loop Iteration over Dictionaries & Lists
**Explanation**: Iterates across transaction records utilizing the Python iterator protocol under the hood.

**Syntax**: `for item in sequence:`

```python
total_sample_spend = 0.0
for t in transactions[:10]:
    total_sample_spend += float(t['transaction_amount'])
print(f'Summed spend across first 10 transactions: ${total_sample_sample_spend:,.2f}' if 'total_sample_sample_spend' in locals() else f'Total Spend: ${total_sample_spend:,.2f}')
```

### Sequence Generation with `range()`
**Explanation**: `range(start, stop, step)` generates memory-efficient arithmetic progressions without materializing lists in RAM ($O(1)$ memory).

**Syntax**: `range(start, stop, step)`

```python
print('Range object size in memory:', sys.getsizeof(range(0, 1_000_000)), 'bytes (constant O(1))')
for i in range(0, 5, 2):
    print(f'Batch Offset {i}: {transactions[i]["transaction_id"]}')
```

### While-Loop Execution with Sentinel Conditions
**Explanation**: Executes iterations dynamically until a stopping predicate evaluates to `False`.

**Syntax**: `while condition: ...`

```python
idx = 0
fraud_count = 0
while idx < len(transactions) and fraud_count < 3:
    if int(transactions[idx]['is_fraud']) == 1:
        fraud_count += 1
        print(f'Found Fraud #{fraud_count}: {transactions[idx]["transaction_id"]} (${float(transactions[idx]["transaction_amount"]):.2f})')
    idx += 1
```

### Loop Control: `break`, `continue`, and `for...else`
**Explanation**: `break` terminates the loop immediately. `continue` skips to the next iteration. `else:` executes ONLY if the loop completed naturally without hitting a `break`.

**Syntax**: `for x in iter: if cond: break else: ...`

```python
target_tx = 'TX999999'
for t in transactions[:10]:
    if t['transaction_id'] == target_tx:
        print('Found target!')
        break
else:
    print(f'Target {target_tx} not found in sample (for...else executed successfully).')
```

## Section: Senior Fintech Interview Scenarios (5+ Years Experience)

### Q1: When does `for...else` execute in Python?
**Explanation**: Explain the search-and-fallback idiom: `else` executes when no break occurs, eliminating the need for boolean flag variables (`found = True`).

**Syntax**: `for item in seq: if match(item): found = True; break` vs `for...else`

```python
def check_any_fraud(batch):
    for t in batch:
        if int(t['is_fraud']) == 1:
            return 'FRAUD_DETECTED'
    return 'ALL_CLEAN'
print('Batch 1 Audit Result:', check_any_fraud(transactions[:10]))
```
