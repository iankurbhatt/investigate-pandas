# Loops & Iteration Control: Beginner Guide

### 🌟 What is Iteration & Loop Control in Python?
**Loops** allow you to automate repetitive tasks by stepping through sequences (like lists of transactions or rows) using `for` loops or repeating actions until a condition changes using `while` loops. Loop control keywords like `break` (exit early) and `continue` (skip to the next iteration) let you precisely manage iteration flow.

This interactive guide loads and works directly with `data/raw_transactions.csv`, giving you real-world hands-on practice.

### 📚 Key Concepts Covered in this Notebook:
- **For-Loop Iteration**: Covers `for item in iterable:`.
- **Range Progression**: Covers `range(stop)` and `range(start, stop, step)`.
- **While-Loop Execution**: Covers `while condition:`.
- **Flow Interruption**: Covers `break` and `continue`.
- **Fallback Clauses**: Covers `for ... else:` and `while ... else:`.


```python
# Setup imports & dataset loading from raw_transactions.csv
import csv
import sys
import time
import os

csv_path = 'data/raw_transactions.csv' if os.path.exists('data/raw_transactions.csv') else '../data/raw_transactions.csv'
transactions = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        transactions.append(row)

print(f"Python Version: {sys.version.split()[0]}")
print(f"Loaded {len(transactions)} transaction records from {csv_path}")
print('Sample Record:', transactions[0])
```

### 🔹 For-Loop Iteration: `for item in iterable:`
Iterates sequentially across elements using Python's underlying iterator protocol. Saving and loading data properly ensures results are persistent, shareable across teams, and ready for downstream analysis. **Tip:** Generators calculate values lazily on demand. This lets you process millions of rows without loading the entire dataset into memory at once.

**Syntax:** `for item in seq:`


```python
for t in transactions[:3]:
    print(f'Processing {t["transaction_id"]}: ${float(t["transaction_amount"]):.2f}')
```

### 🔹 Sequence Generation: `range(stop)`
Generates integer sequence from 0 to `stop - 1` with $O(1)$ constant memory. Saving and loading data properly ensures results are persistent, shareable across teams, and ready for downstream analysis. **Tip:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

**Syntax:** `range(5)`


```python
for i in range(3):
    print(f'Iteration Index {i}: {transactions[i]["transaction_id"]}')
```

### 🔹 Stepped Sequence Generation: `range(start, stop, step)`
Generates stepped progressions skipping intermediate indices. Saving and loading data properly ensures results are persistent, shareable across teams, and ready for downstream analysis. **Tip:** Print intermediate variables with `print()` or inspect their type with `type()` to trace how data changes at each step.

**Syntax:** `range(0, 10, 2)`


```python
for i in range(0, 6, 2):
    print(f'Stepped Offset {i}: {transactions[i]["transaction_id"]}')
```

### 🔹 While-Loop Execution: `while condition:`
Repeats execution dynamically while a state predicate evaluates to True. Saving and loading data properly ensures results are persistent, shareable across teams, and ready for downstream analysis. **Tip:** Always remember to include `self` as the first parameter in instance methods so Python knows which object instance is executing.

**Syntax:** `while count < threshold:`


```python
count = 0
while count < 3:
    print(f'While Count {count}: {transactions[count]["transaction_id"]}')
    count += 1
```

### 🔹 Loop Interruption: `break`
Terminates loop execution immediately upon hitting trigger condition. Saving and loading data properly ensures results are persistent, shareable across teams, and ready for downstream analysis. **Tip:** Always remember to include `self` as the first parameter in instance methods so Python knows which object instance is executing.

**Syntax:** `if cond: break`


```python
for t in transactions:
    if int(t['is_fraud']) == 1:
        print(f'First Fraud Intercepted: {t["transaction_id"]}. Breaking loop!')
        break
```

### 🔹 Loop Skipping: `continue`
Skips remainder of current iteration and moves immediately to next item. Loops and iterators let you automate repetitive tasks over collections of items cleanly and systematically. **Tip:** Always remember to include `self` as the first parameter in instance methods so Python knows which object instance is executing.

**Syntax:** `if cond: continue`


```python
for t in transactions[:5]:
    if float(t['transaction_amount']) < 500.0:
        continue
    print(f'High-Value Transaction: {t["transaction_id"]} (${float(t["transaction_amount"]):.2f})')
```

### 🔹 For-Loop Fallback Clause: `for ... else:`
`else:` block executes ONLY if loop completes naturally without encountering a `break`. Loops and iterators let you automate repetitive tasks over collections of items cleanly and systematically. **Tip:** Always remember to include `self` as the first parameter in instance methods so Python knows which object instance is executing.

**Syntax:** `for item in seq: ... else: ...`


```python
search_id = 'NON_EXISTENT_TX'
for t in transactions[:5]:
    if t['transaction_id'] == search_id:
        print('Found!')
        break
else:
    print(f'Search ID {search_id} not found (for...else executed).')
```

### 🔹 While-Loop Fallback Clause: `while ... else:`
`else:` block executes when while condition becomes False (and loop was not terminated by break). Loops and iterators let you automate repetitive tasks over collections of items cleanly and systematically. **Tip:** Always remember to include `self` as the first parameter in instance methods so Python knows which object instance is executing.

**Syntax:** `while cond: ... else: ...`


```python
n = 0
while n < 2:
    n += 1
else:
    print('While condition completed cleanly (while...else executed).')
```

### 🔍 Scenario: Q1: Memory Optimization: Python `range` vs `list(range)`
